### 練習 01：建立空字典

**題目**：使用兩種方法建立空字典，並印出它們的型別。

**解題思路**：
1. Python 提供兩種建立空字典的方法：`{}` 和 `dict()`
2. 兩種方法建立的都是 `dict` 型別
3. 使用 `type()` 函數可以查看變數的型別
4. 這題主要是熟悉基本語法，沒有陷阱

````python
# 解答 01
# 1. 使用 {} 建立 d1
d1 = {}

# 2. 使用 dict() 建立 d2
d2 = dict()

# 3. 印出 d1 和 d2 的內容與型別
print("d1:", d1)
print("d1 型別:", type(d1))
print("d2:", d2)
print("d2 型別:", type(d2))
````

**輸出與重點解釋**：

```
d1: {}
d1 型別: <class 'dict'>
d2: {}
d2 型別: <class 'dict'>
```

**重點**：
- `{}` 和 `dict()` 建立的都是完全相同的空字典
- 實務上較常使用 `{}` 因為更簡潔
- ⚠️ 注意：空的 `{}` 是字典，不是 set（set 要用 `set()`）

**常見錯誤**：
- 誤以為 `{}` 是 set（實際上空 set 要用 `set()`）

### 練習 02：建立並讀取字典

**題目**：建立一個學生資料字典，並讀取特定資料。

**解題思路**：
1. 使用 `{}` 建立字典，格式為 `{'key': value}`
2. 使用 `d[key]` 讀取對應的 value
3. 這題的 key 都存在，所以不會有 KeyError
4. 可以直接使用方括號語法讀取

````python
# 解答 02
# 1. 建立 student 字典
student = {
    'name': 'Alice',
    'age': 20,
    'grade': 'A'
}

# 2. 印出姓名
print("姓名:", student['name'])

# 3. 印出年齡
print("年齡:", student['age'])
````

**輸出與重點解釋**：

```
姓名: Alice
年齡: 20
```

**重點**：
- 字典使用 `key: value` 的格式，用逗號分隔每個配對
- 讀取資料用 `d[key]` 語法，類似 list 的索引但使用 key 而非數字
- Key 可以是字串、數字、tuple 等不可變類型

**常見錯誤**：
- 忘記 key 要加引號（字串 key）
- 使用不存在的 key 會引發 KeyError

### 練習 03：使用 get() 避免錯誤

**題目**：使用 `get()` 方法安全地查詢字典中不存在的 key。

**解題思路**：
1. 當 key 不存在時，`d[key]` 會引發 KeyError
2. `get(key, default)` 方法可以指定預設值，避免錯誤
3. 如果 key 存在，回傳對應的 value
4. 如果 key 不存在，回傳預設值（本題為 0）

````python
# 解答 03
product = {'name': 'iPhone', 'price': 30000, 'brand': 'Apple'}

# 使用 get() 查詢 stock，預設值為 0
stock = product.get('stock', 0)

# 印出庫存數量
print("庫存數量:", stock)
````

**輸出與重點解釋**：

```
庫存數量: 0
```

**重點 - `d[key]` vs `d.get(key, default)` 的差異**：

| 方法 | key 存在 | key 不存在 | 使用時機 |
|------|----------|------------|----------|
| `d[key]` | 回傳 value | ❌ KeyError | 確定 key 存在 |
| `d.get(key)` | 回傳 value | 回傳 None | 不確定 key 是否存在 |
| `d.get(key, default)` | 回傳 value | 回傳 default | 需要預設值 |

**使用建議**：
- 從資料庫、API 取得的資料：用 `get()` 較安全
- 自己建立的字典且確定 key 存在：可用 `d[key]`
- 需要預設值時：一定用 `get(key, default)`

### 練習 04：get() 與直接查詢的差異

**題目**：比較使用 `d[key]` 和 `d.get(key)` 的差異。

**解題思路**：
1. 使用 `get('美國', '查無此國家')` 安全查詢
2. 因為 '美國' 不在字典中，會回傳預設值
3. 如果改用 `capitals['美國']` 會引發 KeyError
4. 這題展示 get() 的防呆功能

````python
# 解答 04
capitals = {'台灣': '台北', '日本': '東京', '韓國': '首爾'}

# 使用 get() 查詢美國，預設值為 '查無此國家'
result = capitals.get('美國', '查無此國家')
print("美國的首都:", result)

# 進階：測試直接查詢會發生什麼（已註解避免錯誤）
# print(capitals['美國'])  # 會引發 KeyError: '美國'
````

**輸出與重點解釋**：

```
美國的首都: 查無此國家
```

**重點**：
- `get()` 方法的第二個參數是預設值，當 key 不存在時回傳
- 如果只寫 `get(key)` 不寫預設值，則回傳 `None`
- 使用 `d[key]` 存取不存在的 key 會立即引發 `KeyError` 並中斷程式

**實務應用**：
```python
# 處理使用者輸入，可能有缺漏
user_data = {'name': 'Alice', 'age': 20}
email = user_data.get('email', 'No email provided')  # 安全取得，不會出錯
```

### 練習 05：新增資料

**題目**：在現有字典中新增資料。

**解題思路**：
1. 使用 `d[new_key] = value` 新增資料
2. 如果 key 不存在，會自動新增
3. 如果 key 已存在，則會更新該 key 的 value
4. 字典是可變的（mutable），可以直接修改

````python
# 解答 05
user = {'username': 'john123', 'email': 'john@example.com'}

# 新增 age 和 city
user['age'] = 25
user['city'] = 'Taipei'

# 印出完整字典
print(user)
````

**輸出與重點解釋**：

```
{'username': 'john123', 'email': 'john@example.com', 'age': 25, 'city': 'Taipei'}
```

**重點**：
- 新增資料很簡單：`d[key] = value`
- Python 3.7+ 字典會保持插入順序（insertion order）
- 新增的 key-value 會出現在字典最後面

**與 List 的差異**：
- List 新增用 `append()` 或 `insert()`
- Dict 新增直接用 `d[key] = value`，更直觀

### 練習 06：更新資料

**題目**：更新字典中已存在的資料。

**解題思路**：
1. 更新與新增用相同語法：`d[key] = new_value`
2. 如果 key 已存在，會覆蓋舊值
3. 如果 key 不存在，會新增該 key
4. Python 會自動判斷是新增還是更新

````python
# 解答 06
scores = {'Alice': 85, 'Bob': 70, 'Cindy': 92}

# 更新 Bob 和 Alice 的成績
scores['Bob'] = 75
scores['Alice'] = 90

# 印出更新後的字典
print(scores)
````

**輸出與重點解釋**：

```
{'Alice': 90, 'Bob': 75, 'Cindy': 92}
```

**重點**：
- 更新資料會**直接覆蓋**舊值，不會保留歷史
- 如果需要保留舊值，要先複製或儲存到另一個變數
- 字典的 key 必須唯一，不能有重複的 key

**新增 vs 更新**：
```python
d = {'a': 1}
d['a'] = 2  # 更新：key 'a' 已存在，值從 1 變成 2
d['b'] = 3  # 新增：key 'b' 不存在，新增此配對
```

### 練習 07：使用 len() 計算字典大小

**題目**：計算字典中有幾個 key-value 配對。

**解題思路**：
1. `len(d)` 回傳字典中 key-value 配對的數量
2. 與 list 的 `len()` 用法相同
3. 注意：len() 計算的是配對數，不是 key 或 value 的總數

````python
# 解答 07
inventory = {'apple': 50, 'banana': 30, 'orange': 45, 'grape': 20}

# 計算商品種類數量
count = len(inventory)

# 印出結果
print(f"庫存中有 {count} 種商品")
````

**輸出與重點解釋**：

```
庫存中有 4 種商品
```

**重點**：
- `len(d)` 計算的是 key-value **配對的數量**
- 等同於 `len(d.keys())` 或 `len(d.values())`
- 空字典的 len 為 0

**實用場景**：
```python
if len(cart) == 0:
    print("購物車是空的")
elif len(cart) > 10:
    print("購物車商品過多，請分批結帳")
```

### 練習 08：使用 in 檢查 key 是否存在

**題目**：檢查特定的 key 是否存在於字典中。

**解題思路**：
1. 使用 `key in d` 檢查 key 是否存在，回傳 True/False
2. 這是避免 KeyError 的另一種方法
3. 可以配合 if 條件判斷來執行不同邏輯
4. 注意：`in` 預設檢查的是 key，不是 value

````python
# 解答 08
menu = {'pizza': 200, 'burger': 150, 'pasta': 180}

# 檢查 pizza 是否存在
print("pizza 在菜單中:", 'pizza' in menu)

# 檢查 steak 是否存在
print("steak 在菜單中:", 'steak' in menu)

# 使用 if 判斷 burger
if 'burger' in menu:
    print("有漢堡")
````

**輸出與重點解釋**：

```
pizza 在菜單中: True
steak 在菜單中: False
有漢堡
```

**重點 - 三種檢查 key 存在的方法**：

```python
# 方法1：使用 in（推薦）
if 'key' in d:
    value = d['key']

# 方法2：使用 get()
value = d.get('key')
if value is not None:
    # 處理 value

# 方法3：使用 keys()（較少用）
if 'key' in d.keys():
    value = d['key']
```

**注意**：
- `in` 預設檢查 **key**，不是 value
- 如果要檢查 value，用 `value in d.values()`

### 練習 09：使用 keys()、values()、items()

**題目**：分別印出字典的所有 key、所有 value、以及所有 key-value 配對。

**解題思路**：
1. `keys()` 回傳所有的 key
2. `values()` 回傳所有的 value
3. `items()` 回傳所有的 key-value 配對（以 tuple 形式）
4. 這三個方法回傳的都是特殊的「視圖物件」，不是 list

````python
# 解答 09
colors = {'red': '紅色', 'blue': '藍色', 'green': '綠色'}

# 印出所有的 key
print("所有的 key:", colors.keys())

# 印出所有的 value
print("所有的 value:", colors.values())

# 印出所有的配對
print("所有的配對:", colors.items())
````

**輸出與重點解釋**：

```
所有的 key: dict_keys(['red', 'blue', 'green'])
所有的 value: dict_values(['紅色', '藍色', '綠色'])
所有的配對: dict_items([('red', '紅色'), ('blue', '藍色'), ('green', '綠色')])
```

**重點 - keys/values/items 回傳型別**：

| 方法 | 回傳型別 | 用途 | 特性 |
|------|----------|------|------|
| `keys()` | dict_keys | 取得所有 key | 可迭代、支援 in |
| `values()` | dict_values | 取得所有 value | 可迭代 |
| `items()` | dict_items | 取得所有配對 | 可迭代、解包 |

**重要觀念**：
- 這三個方法回傳的是**視圖物件**（view object），不是 list
- 視圖物件會**動態反映**字典的變化
- 如果需要 list，用 `list(d.keys())`
- 在 for 迴圈中可直接使用，不需要轉換

### 練習 10：轉換 keys、values 為 list

**題目**：將字典的 keys 和 values 轉換為 list。

**解題思路**：
1. 使用 `list()` 函數將視圖物件轉換為 list
2. `list(d.keys())` 得到 key 的 list
3. `list(d.values())` 得到 value 的 list
4. 轉換後就可以使用 list 的所有方法（索引、切片等）

````python
# 解答 10
grades = {'math': 95, 'english': 88, 'science': 92}

# 轉換 keys 為 list
subjects = list(grades.keys())

# 轉換 values 為 list
scores = list(grades.values())

# 印出結果
print("科目:", subjects)
print("分數:", scores)
````

**輸出與重點解釋**：

```
科目: ['math', 'english', 'science']
分數: [95, 88, 92]
```

**重點**：
- 使用 `list()` 可以將任何可迭代物件轉為 list
- 轉換後的 list 順序與字典順序相同（Python 3.7+）
- 轉換後就可以使用 list 特有的操作（索引、排序等）

**何時需要轉換**：
```python
# 需要索引存取
keys_list = list(d.keys())
first_key = keys_list[0]

# 需要排序
sorted_values = sorted(d.values())

# 只是迭代則不需要轉換
for key in d.keys():  # 不需要 list()
    print(key)
```

## 📊 Section A 總結（第 1-10 題）

### 🎯 核心技巧

1. **建立字典**：`{}` 或 `dict()`
2. **讀取資料**：
   - 確定存在：`d[key]`
   - 不確定存在：`d.get(key, default)`
3. **新增/更新**：`d[key] = value`（自動判斷）
4. **檢查存在**：`key in d`
5. **基本操作**：`len(d)`、`d.keys()`、`d.values()`、`d.items()`

### ⚠️ 易錯點

1. **KeyError 陷阱**：使用不存在的 key 會出錯
   - 解決：使用 `get()` 或先用 `in` 檢查

2. **視圖物件 vs List**：
   - `keys()`、`values()`、`items()` 回傳的不是 list
   - 需要 list 時要用 `list()` 轉換

3. **新增 vs 更新**：
   - 語法相同：`d[key] = value`
   - Python 自動判斷是新增還是更新

### 💡 最佳實踐

```python
# ✅ 好的寫法
value = d.get('key', 'default')  # 安全
if 'key' in d:  # 先檢查
    value = d['key']

# ❌ 危險的寫法
value = d['key']  # 可能 KeyError
```

### 練習 11：使用 del 刪除資料

**題目**：使用 `del` 刪除字典中的特定資料。

**解題思路**：
1. `del d[key]` 可以直接刪除指定的 key-value 配對
2. 如果 key 不存在，會引發 KeyError
3. `del` 不會回傳被刪除的值
4. 刪除後字典的大小會減少

````python
# 解答 11
cart = {'apple': 3, 'banana': 2, 'orange': 5, 'grape': 1}

# 刪除 grape 和 banana
del cart['grape']
del cart['banana']

# 印出刪除後的字典
print(cart)
````

**輸出與重點解釋**：

```
{'apple': 3, 'orange': 5}
```

**重點**：
- `del` 是 Python 的關鍵字，用於刪除物件
- `del d[key]` 會永久刪除該 key-value 配對
- 如果要刪除不確定存在的 key，先用 `in` 檢查：
  ```python
  if 'key' in d:
      del d['key']
  ```

**del vs pop() 的差異**：
- `del`：只刪除，不回傳值
- `pop()`：刪除並回傳被刪除的值（下一題）

### 練習 12：使用 pop() 刪除並取得值

**題目**：使用 `pop()` 刪除資料並取得被刪除的值。

**解題思路**：
1. `pop(key)` 會刪除指定的 key 並回傳其 value
2. 如果需要知道被刪除的值，用 `pop()` 而不是 `del`
3. `pop()` 也可以指定預設值：`pop(key, default)`
4. 這樣即使 key 不存在也不會出錯

````python
# 解答 12
settings = {'volume': 80, 'brightness': 70, 'language': 'zh-TW'}

# 使用 pop() 刪除 brightness 並取得值
old_brightness = settings.pop('brightness')

# 印出原亮度
print("原亮度:", old_brightness)

# 印出刪除後的字典
print("刪除後:", settings)
````

**輸出與重點解釋**：

```
原亮度: 70
刪除後: {'volume': 80, 'language': 'zh-TW'}
```

**重點 - del vs pop() 完整比較**：

| 方法 | 語法 | 回傳值 | key 不存在時 | 使用時機 |
|------|------|--------|--------------|----------|
| `del` | `del d[key]` | 無 | KeyError | 只想刪除 |
| `pop()` | `d.pop(key)` | 被刪除的 value | KeyError | 需要取得值 |
| `pop() + default` | `d.pop(key, default)` | value 或 default | 回傳 default | 安全刪除 |

**實用範例**：
```python
# 情境：移除暫存資料並記錄
temp_value = cache.pop('temp_key', None)
if temp_value:
    print(f"清除暫存: {temp_value}")
```

### 練習 13：使用 for 走訪字典的 key

**題目**：使用 for 迴圈走訪字典，印出每個 key 和對應的 value。

**解題思路**：
1. `for key in d:` 預設會走訪所有的 key
2. 在迴圈內用 `d[key]` 取得對應的 value
3. 也可以寫成 `for key in d.keys():` 但通常省略 `.keys()`
4. 這是最基本的字典走訪方式

````python
# 解答 13
fruits = {'apple': 50, 'banana': 30, 'orange': 40}

# 使用 for 走訪並印出
for key in fruits:
    print(f"{key}: {fruits[key]}元")
````

**輸出與重點解釋**：

```
apple: 50元
banana: 30元
orange: 40元
```

**重點 - 三種走訪字典的方式**：

```python
# 方式1：走訪 key（最常用）
for key in d:
    print(key, d[key])

# 方式2：明確使用 keys()（較少用）
for key in d.keys():
    print(key, d[key])

# 方式3：走訪 items()（推薦，見下題）
for key, value in d.items():
    print(key, value)
```

**注意**：
- Python 3.7+ 字典走訪順序等於插入順序
- 不要在走訪過程中修改字典大小（新增/刪除）

### 練習 14：使用 items() 走訪字典

**題目**：使用 `items()` 方法同時取得 key 和 value。

**解題思路**：
1. `d.items()` 回傳所有的 (key, value) tuple
2. 在 for 迴圈中可以直接解包：`for key, value in d.items()`
3. 這是最推薦的走訪方式，程式碼更簡潔易讀
4. 不需要在迴圈內再用 `d[key]` 取值

````python
# 解答 14
countries = {'Taiwan': 'Taipei', 'Japan': 'Tokyo', 'Korea': 'Seoul'}

# 使用 items() 走訪並印出
for country, capital in countries.items():
    print(f"{country} 的首都是 {capital}")
````

**輸出與重點解釋**：

```
Taiwan 的首都是 Taipei
Japan 的首都是 Tokyo
Korea 的首都是 Seoul
```

**重點 - items() 解包（unpacking）**：

```python
# items() 回傳的每個元素都是 tuple
for item in d.items():
    print(item)  # 輸出: ('key', 'value')

# 使用解包更優雅（推薦）
for key, value in d.items():
    print(key, value)
```

**為什麼推薦 items()**：
1. 程式碼更簡潔（不需要 `d[key]`）
2. 效能更好（不需要重複查找）
3. 可讀性更高（意圖更明確）

**常見錯誤**：
```python
# ❌ 錯誤：只用一個變數接收
for item in d.items():
    print(item[0], item[1])  # 麻煩且不直觀

# ✅ 正確：解包成兩個變數
for key, value in d.items():
    print(key, value)
```

### 練習 15：計算 values 的總和

**題目**：計算字典中所有 values 的總和。

**解題思路**：
1. 方法1：使用 for 迴圈走訪 values 並累加
2. 方法2：使用 `sum(d.values())` 一行完成
3. 兩種方法都可以，但 sum() 更簡潔
4. 這題展示如何對 values 進行統計運算

````python
# 解答 15
expenses = {'rent': 15000, 'food': 8000, 'transport': 3000, 'entertainment': 2000}

# 方法1：使用 for 迴圈
total = 0
for value in expenses.values():
    total += value
print(f"總支出: {total} 元")

# 方法2：使用 sum()（更簡潔）
total2 = sum(expenses.values())
print(f"總支出: {total2} 元")
````

### 練習 17：字典的相等性比較（順序不影響）

**題目**：測試字典的相等性，驗證順序不同但內容相同的字典是否相等。

**解題思路**：
1. Python 字典比較時，**只看 key-value 配對是否相同**
2. **順序不影響**相等性判斷（與 list 不同）
3. 只要所有 key 和對應的 value 都相同，字典就相等
4. 這是字典的重要特性，與 list 的順序敏感不同

````python
# 解答 17
# 建立三個字典
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'c': 3, 'a': 1, 'b': 2}  # 順序不同
dict3 = {'a': 1, 'b': 2}  # 少一個元素

# 比較並印出結果
print("dict1 == dict2:", dict1 == dict2)
print("dict1 == dict3:", dict1 == dict3)
````

**輸出與重點解釋**：

```
dict1 == dict2: True
dict1 == dict3: False
```

**重點 - Dict vs List 的相等性比較**：

| 資料結構 | 比較方式 | 順序影響 | 範例 |
|---------|---------|---------|------|
| List | 元素和順序 | ✅ 影響 | `[1,2,3] != [3,2,1]` |
| Dict | key-value 配對 | ❌ 不影響 | `{'a':1,'b':2} == {'b':2,'a':1}` |
| Set | 元素內容 | ❌ 不影響 | `{1,2,3} == {3,2,1}` |
| Tuple | 元素和順序 | ✅ 影響 | `(1,2,3) != (3,2,1)` |

**為什麼字典順序不影響**：
- 字典本質上是 **key 到 value 的映射**
- 只要映射關係相同，就是相同的字典
- Python 3.7+ 雖然保持插入順序，但比較時依然忽略順序

**實務意義**：
```python
# 比較兩個 API 回傳的 JSON 資料
api1_data = {'user': 'Alice', 'age': 20, 'city': 'Taipei'}
api2_data = {'age': 20, 'city': 'Taipei', 'user': 'Alice'}
print(api1_data == api2_data)  # True，順序無所謂
```

### 練習 18：找出最高分

**題目**：從成績字典中找出最高分及對應的學生姓名。

**解題思路**：
1. 先用 `max(scores.values())` 找出最高分數
2. 用 for 迴圈配合 if 條件找出分數等於最高分的學生
3. 也可以用 `max()` 配合 `key` 參數一次完成
4. 這題練習結合統計與條件篩選

````python
# 解答 18
scores = {'Alice': 85, 'Bob': 92, 'Cindy': 78, 'David': 95, 'Eva': 88}

# 方法1：分兩步驟
max_score = max(scores.values())
for name, score in scores.items():
    if score == max_score:
        top_student = name
        break

print(f"最高分: {top_student}, {max_score} 分")

# 方法2：使用 max() 的 key 參數（更優雅）
top_student2 = max(scores, key=scores.get)
print(f"最高分: {top_student2}, {scores[top_student2]} 分")
````

**輸出與重點解釋**：

```
最高分: David, 95 分
最高分: David, 95 分
```

**重點 - max() 的進階用法**：

```python
# 1. 找最大的 value
max_value = max(d.values())

# 2. 找最大 value 對應的 key（推薦）
max_key = max(d, key=d.get)

# 3. 同時取得 key 和 value
max_key = max(d, key=d.get)
max_value = d[max_key]

# 4. 或使用 items()
max_item = max(d.items(), key=lambda x: x[1])  # 回傳 (key, value) tuple
```

**key 參數的作用**：
- `max(d, key=d.get)` 表示：對 d 中的每個 key，用 `d.get(key)` 的結果來比較
- 等同於找出 value 最大的 key

**min() 用法相同**：
```python
# 找最低分
min_student = min(scores, key=scores.get)
print(f"最低分: {min_student}, {scores[min_student]} 分")
```

**常見錯誤**：
```python
# ❌ 錯誤：max(d) 只會比較 key（字母順序）
max(scores)  # 回傳 'Eva'（字母最大），不是分數最高

# ✅ 正確：要指定 key 參數
max(scores, key=scores.get)  # 回傳 'David'（分數最高）
```

## 📊 Section B 總結（第 11-20 題）

### 🎯 核心技巧

1. **刪除資料**：
   - `del d[key]`：只刪除，不回傳
   - `d.pop(key)`：刪除並回傳值
   - `d.pop(key, default)`：安全刪除

2. **走訪字典**：
   - `for key in d:`：走訪 key
   - `for key, value in d.items()`：走訪配對（推薦）
   - `for value in d.values()`：走訪 value

3. **統計運算**：
   - 總和：`sum(d.values())`
   - 平均：`sum(d.values()) / len(d)`
   - 最大/小：`max(d.values())` / `min(d.values())`
   - 找最大值的 key：`max(d, key=d.get)`

4. **字典比較**：
   - 順序不影響相等性（與 list 不同）
   - 只看 key-value 配對是否相同

5. **list of dict**：
   - 走訪：`for item in list_of_dict:`
   - 篩選：`[item for item in data if condition]`
   - 取欄位：`[item['field'] for item in data]`

### ⚠️ 易錯點

1. **max() 的陷阱**：
   - `max(d)` 比較的是 key，不是 value
   - 要用 `max(d, key=d.get)` 才能找到最大 value 的 key

2. **空字典的統計**：
   - `sum(d.values()) / len(d)` 若 d 為空會出錯
   - 要先檢查：`if len(d) > 0:`

3. **list of dict 的 KeyError**：
   - 不是每個 dict 都有相同的 key
   - 建議用 `get()` 安全存取

### 💡 最佳實踐

```python
# ✅ 刪除資料的選擇
if need_value:
    value = d.pop(key, None)  # 需要取得值
else:
    if key in d:  # 安全刪除
        del d[key]

# ✅ 走訪字典
for key, value in d.items():  # 最清晰
    print(key, value)

# ✅ 找極值
max_key = max(d, key=d.get)  # 找最大值對應的 key

# ✅ 篩選 list of dict
result = [item for item in data if item.get('age', 0) >= 20]
```

### 練習 23：字串字元計數器（使用 get）

**題目**：統計字串中每個字元出現的次數（不使用 defaultdict，改用 get 方法）。

**解題思路**：
1. 建立空字典作為計數器
2. 走訪字串中的每個字元
3. 使用 `counter[char] = counter.get(char, 0) + 1` 來計數
4. 這是實作計數器的經典模式，必須熟練掌握

````python
# 解答 23
text = "hello"
counter = {}

# 使用 for + get() 統計每個字元
for char in text:
    counter[char] = counter.get(char, 0) + 1

# 印出結果
print(counter)
````

**輸出與重點解釋**：

```
{'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

**重點 - 使用 get() 實作計數器的原理**：

```python
# 核心邏輯
counter[key] = counter.get(key, 0) + 1

# 拆解說明：
# 1. counter.get(key, 0)：如果 key 不存在，回傳 0
# 2. ... + 1：在原值（或 0）的基礎上加 1
# 3. counter[key] = ...：將新值寫回字典

# 等同於：
if key in counter:
    counter[key] += 1
else:
    counter[key] = 1
```

**為什麼必須用 get()**：
```python
# ❌ 錯誤：直接 += 會 KeyError
counter[char] += 1  # 如果 char 第一次出現會出錯

# ✅ 方法1：使用 get()（推薦）
counter[char] = counter.get(char, 0) + 1

# ✅ 方法2：先檢查
if char in counter:
    counter[char] += 1
else:
    counter[char] = 1

# ✅ 方法3：使用 setdefault()
counter.setdefault(char, 0)
counter[char] += 1
```

**實務應用**：
```python

for word in text.split():
    word_count[word] = word_count.get(word, 0) + 1

# 2. 投票統計（見下題）
votes = {}
for candidate in vote_list:
    votes[candidate] = votes.get(candidate, 0) + 1

# 3. 頻率分析
freq = {}
for item in data:
    freq[item] = freq.get(item, 0) + 1
```

**進階：找出最常出現的字元**：
```python

```

### 練習 24：投票統計系統

**題目**：統計投票結果，計算每個候選人得票數，並找出獲勝者。

**解題思路**：
1. 使用與上題相同的計數器模式
2. 走訪投票 list，對每個候選人累計票數
3. 使用 `max()` 配合 `key` 參數找出得票最多的候選人
4. 這題結合了計數器和極值查找

````python
# 解答 24
votes = ['Alice', 'Bob', 'Alice', 'Cindy', 'Bob', 'Alice', 'Bob', 'Cindy']
vote_count = {}

# 統計票數
for candidate in votes:
    vote_count[candidate] = vote_count.get(candidate, 0) + 1

# 印出投票結果
print("投票結果:")
for name, count in vote_count.items():
    print(f"{name}: {count} 票")

# 找出獲勝者
winner = max(vote_count, key=vote_count.get)
print(f"\n獲勝者: {winner} ({vote_count[winner]} 票)")
````

### 練習 25：Dict Comprehension 基礎

**題目**：使用 dict comprehension 從 list 建立字典。

**解題思路**：
1. dict comprehension 語法：`{key: value for item in iterable}`
2. 類似 list comprehension，但產生字典
3. 可以在一行內完成迴圈 + 字典建立
4. 這是 Pythonic 的寫法，簡潔且高效

````python
# 解答 25
numbers = [1, 2, 3, 4, 5]

# 使用 dict comprehension 建立平方字典
squares = {num: num**2 for num in numbers}

# 印出結果
print(squares)

# 也可以這樣寫（傳統寫法對照）
squares_traditional = {}
for num in numbers:
    squares_traditional[num] = num**2
print("傳統寫法:", squares_traditional)
````

**輸出與重點解釋**：

```
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
傳統寫法: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

**重點 - Dict Comprehension 語法**：

```python
# 基本格式
{key_expression: value_expression for item in iterable}

# 範例1：數字到平方
{x: x**2 for x in range(5)}
# 結果: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# 範例2：字串到長度
words = ['apple', 'banana', 'cherry']
{word: len(word) for word in words}
# 結果: {'apple': 5, 'banana': 6, 'cherry': 6}

# 範例3：從兩個 list 建立字典
keys = ['a', 'b', 'c']
values = [1, 2, 3]
{k: v for k, v in zip(keys, values)}
# 結果: {'a': 1, 'b': 2, 'c': 3}
```

**與 for 迴圈的比較**：

```python
# 傳統 for 迴圈（3-4 行）
result = {}
for item in data:
    result[item] = process(item)

# Dict comprehension（1 行）
result = {item: process(item) for item in data}
```

**何時使用 comprehension**：
- ✅ 簡單的轉換邏輯
- ✅ 一對一的映射關係
- ✅ 注重程式碼簡潔性
- ❌ 複雜的多步驟邏輯
- ❌ 需要條件判斷（雖然可以，但可讀性差）
- ❌ 需要錯誤處理

**常見模式**：
```python
# 1. 索引到值
{i: val for i, val in enumerate(['a', 'b', 'c'])}
# {0: 'a', 1: 'b', 2: 'c'}

# 2. 值到索引（反向）
{val: i for i, val in enumerate(['a', 'b', 'c'])}
# {'a': 0, 'b': 1, 'c': 2}

# 3. 字母到數字
{chr(i): i for i in range(65, 70)}
# {'A': 65, 'B': 66, 'C': 67, 'D': 68, 'E': 69}
```

### 練習 26：Dict Comprehension 進階（帶條件）

**題目**：使用 dict comprehension 篩選並建立字典。

**解題思路**：
1. 在 comprehension 中加入 if 條件：`{k: v for ... if condition}`
2. 只有符合條件的項目才會被加入字典
3. 條件篩選放在最後
4. 這是 comprehension 的進階用法，結合了篩選與轉換

````python
# 解答 26
fruits = ['apple', 'banana', 'cherry', 'date', 'elderberry']

# 使用 dict comprehension + 條件篩選
long_fruits = {fruit: len(fruit) for fruit in fruits if len(fruit) > 5}

# 印出結果
print(long_fruits)

# 傳統寫法對照
long_fruits_traditional = {}
for fruit in fruits:
    if len(fruit) > 5:
        long_fruits_traditional[fruit] = len(fruit)
print("傳統寫法:", long_fruits_traditional)
````

**輸出與重點解釋**：

```
{'banana': 6, 'cherry': 6, 'elderberry': 10}
傳統寫法: {'banana': 6, 'cherry': 6, 'elderberry': 10}
```

**重點 - 帶條件的 Dict Comprehension**：

```python
# 格式：filter 條件放在最後
{key: value for item in iterable if condition}

# 範例1：只要偶數
{x: x**2 for x in range(10) if x % 2 == 0}
# {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}

# 範例2：過濾特定值
data = {'a': 1, 'b': 0, 'c': 3, 'd': 0}
{k: v for k, v in data.items() if v != 0}
# {'a': 1, 'c': 3}

# 範例3：字串過濾
words = {'apple': 5, 'a': 1, 'banana': 6}
{k: v for k, v in words.items() if len(k) > 2}
# {'apple': 5, 'banana': 6}
```

**條件表達式（三元運算符）**：
```python
# if-else 放在 value 位置（不是 filter）
{x: 'even' if x % 2 == 0 else 'odd' for x in range(5)}
# {0: 'even', 1: 'odd', 2: 'even', 3: 'odd', 4: 'even'}

# 注意：這不是 filter，所有元素都會保留
```

**多重條件**：
```python
# AND 條件
{x: x**2 for x in range(20) if x % 2 == 0 if x > 10}
# 等同於
{x: x**2 for x in range(20) if x % 2 == 0 and x > 10}

# OR 條件（需要用 or）
{x: x for x in range(10) if x < 3 or x > 7}
```

**從現有字典篩選**：
```python
# 情境：從成績字典中篩選及格的學生
```

### 練習 27：資料清理 - 移除 None 值

**題目**：清理字典，移除 value 為 None 的項目。

**解題思路**：
1. 可以用 dict comprehension 配合條件篩選
2. 條件：`if value is not None`
3. 也可以用傳統 for 迴圈
4. 這是資料清理的基本技巧，常用於處理不完整的資料

````python
# 解答 27
data = {'name': 'Alice', 'age': None, 'city': 'Taipei', 'phone': None, 'email': 'alice@example.com'}

# 方法1：使用 dict comprehension（推薦）
cleaned_data = {k: v for k, v in data.items() if v is not None}

print(cleaned_data)

# 方法2：使用傳統 for 迴圈
cleaned_data2 = {}
for key, value in data.items():
    if value is not None:
        cleaned_data2[key] = value
print("方法2:", cleaned_data2)
````

### 練習 29：綜合應用 - 購物車統計

**題目**：統計購物車中每個商品的總價。

**解題思路**：
1. 步驟1：統計每個商品的總購買數量（使用計數器模式）
2. 步驟2：計算每個商品的總金額（數量 × 單價）
3. 步驟3：計算購物車總金額
4. 這題結合了 list of dict、計數器、統計運算等多個技巧

````python
# 解答 29
cart = [
    {'product': 'apple', 'quantity': 3},
    {'product': 'banana', 'quantity': 2},
    {'product': 'apple', 'quantity': 1},
    {'product': 'orange', 'quantity': 5},
    {'product': 'banana', 'quantity': 3}
]

prices = {'apple': 30, 'banana': 20, 'orange': 25}

# 步驟1: 統計每個商品的總數量
quantity_summary = {}
for item in cart:
    product = item['product']
    qty = item['quantity']
    quantity_summary[product] = quantity_summary.get(product, 0) + qty

# 步驟2: 計算每個商品的總金額
amount_summary = {}
for product, qty in quantity_summary.items():
    amount_summary[product] = qty * prices[product]

# 步驟3: 計算購物車總金額
total_amount = sum(amount_summary.values())

# 印出結果
print("商品統計:")
for product in quantity_summary:
    qty = quantity_summary[product]
    amount = amount_summary[product]
    print(f"{product}: {qty} 個, 總計 {amount} 元")

print(f"\n購物車總金額: {total_amount} 元")
````

**輸出與重點解釋**：

```
商品統計:
```

### 練習 30：綜合應用 - 學生成績報表

**題目**：產生學生成績統計報表。

**解題思路**：
1. 為每個學生計算總分和平均分
2. 找出總分最高的學生
3. 計算全班每科的平均分
4. 這是最綜合的題目，整合了所有前面學過的技巧

````python
# 解答 30
students_scores = [
    {'name': 'Alice', 'math': 85, 'english': 92, 'science': 88},
    {'name': 'Bob', 'math': 78, 'english': 85, 'science': 90},
    {'name': 'Cindy', 'math': 92, 'english': 88, 'science': 85},
    {'name': 'David', 'math': 88, 'english': 90, 'science': 92}
]

# 步驟1: 計算每個學生的總分和平均
print("個人成績:")
student_totals = {}
for student in students_scores:
    name = student['name']
    # 計算總分（排除 name 欄位）
    scores = [v for k, v in student.items() if k != 'name']
    total = sum(scores)
    average = total / len(scores)
    student_totals[name] = total
    print(f"{name} - 總分: {total}, 平均: {average:.2f}")

# 步驟2: 找出總分最高的學生
top_student = max(student_totals, key=student_totals.get)
print(f"\n最高分學生: {top_student} ({student_totals[top_student]} 分)")

# 步驟3: 計算全班每科平均
subjects = ['math', 'english', 'science']
print("\n全班各科平均:")
for subject in subjects:
    # 收集該科所有分數
    subject_scores = [s[subject] for s in students_scores]
    avg = sum(subject_scores) / len(subject_scores)
    print(f"{subject}: {avg:.2f}")
````

## ⚠️ 常見錯誤總整理

### 1. KeyError 陷阱
```python
# ❌ 危險
value = d['key']  # 可能出錯

# ✅ 安全
value = d.get('key', default)
```

### 2. max() 誤用
```python
# ❌ 錯誤：比較 key
max(d)

# ✅ 正確：比較 value
max(d, key=d.get)
```

### 3. 空字典統計
```python
# ❌ 危險
avg = sum(d.values()) / len(d)  # 空字典會出錯

# ✅ 安全
avg = sum(d.values()) / len(d) if len(d) > 0 else 0
```

### 4. Comprehension 過度複雜
```python
# ❌ 難讀
{k: v*2 if v>10 else v/2 for k,v in d.items() if k.startswith('a')}

# ✅ 改用 for 迴圈
result = {}
for k, v in d.items():
    if k.startswith('a'):
        result[k] = v * 2 if v > 10 else v / 2
```
