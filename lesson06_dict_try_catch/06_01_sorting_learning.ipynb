## 📖 sorted() 基本用法

`sorted()` 是 Python 內建的排序函式，會回傳一個**新的已排序串列**。

**語法**：`sorted(list)`

### 範例 1：數字排序

````python
# 數字串列排序（由小到大）
numbers = [5, 2, 8, 1, 9]
result = sorted(numbers)

print("原始串列：", numbers)
print("排序結果：", result)

# 預期輸出：
# 原始串列： [5, 2, 8, 1, 9]
# 排序結果： [1, 2, 5, 8, 9]
````

### 範例 2：字串排序

````python
# 字串串列排序（依字母順序）
fruits = ["banana", "apple", "cherry", "date"]
result = sorted(fruits)

print("原始串列：", fruits)
print("排序結果：", result)

# 預期輸出：
# 原始串列： ['banana', 'apple', 'cherry', 'date']
# 排序結果： ['apple', 'banana', 'cherry', 'date']
````

### 範例 3：字串型態的數字排序（重要！）

````python
# ⚠️ 注意：字串型態的數字是依「字典順序」排序，不是數值大小
lst = ['four', '9', '10']
result = sorted(lst)

print("原始串列：", lst)
print("排序結果：", result)
print("\n說明：'10' 會排在 '9' 前面")
print("因為比較的是第一個字元 '1' vs '9'")

# 預期輸出：
# 原始串列： ['four', '9', '10']
# 排序結果： ['10', '9', 'four']
# 
# 說明：'10' 會排在 '9' 前面
# 因為比較的是第一個字元 '1' vs '9'
````

### 💡 重點提醒

- `sorted()` **不會改變**原始串列
- 回傳一個**新的**已排序串列
- 預設是**由小到大**（ascending）

## 🔄 sort() vs sorted() 差異

Python 有兩個排序工具，它們的差異很重要！

| 特性 | `sorted()` | `sort()` |
|------|-----------|---------|
| 類型 | 函式 | 方法 |
| 回傳值 | 新串列 | None |
| 原串列 | 不變 | 改變 |
| 使用方式 | `sorted(list)` | `list.sort()` |

### 範例：sorted() 不改變原串列

````python
# 使用 sorted()
numbers = [5, 2, 8, 1, 9]
result = sorted(numbers)

print("原始串列：", numbers)  # 沒有改變
print("排序結果：", result)   # 新的串列

# 預期輸出：
# 原始串列： [5, 2, 8, 1, 9]
# 排序結果： [1, 2, 5, 8, 9]
````

### 範例：sort() 會改變原串列

````python
# 使用 sort()
numbers = [5, 2, 8, 1, 9]
numbers.sort()  # 直接改變 numbers

print("排序後：", numbers)  # 已經改變了

# 預期輸出：
# 排序後： [1, 2, 5, 8, 9]
````

### 範例 3：完整對照（同一個串列）

````python
# 完整對照範例
print("=== 對照實驗 ===")
print()

# 實驗 1：sorted()
list1 = [16, 1, 89, 26, 93]
print("使用 sorted():")
print("排序前：", list1)
sorted_result = sorted(list1)
print("回傳值：", sorted_result)
print("原串列：", list1, "← 沒變")
print()

# 實驗 2：sort()
list2 = [16, 1, 89, 26, 93]
print("使用 sort():")
print("排序前：", list2)
sort_result = list2.sort()
print("回傳值：", sort_result, "← None")
print("原串列：", list2, "← 已改變")

# 預期輸出：
# === 對照實驗 ===
# 
# 使用 sorted():
# 排序前： [16, 1, 89, 26, 93]
# 回傳值： [1, 16, 26, 89, 93]
# 原串列： [16, 1, 89, 26, 93] ← 沒變
# 
# 使用 sort():
# 排序前： [16, 1, 89, 26, 93]
# 回傳值： None ← None
# 原串列： [1, 16, 26, 89, 93] ← 已改變
````

### 🎯 選擇建議

- **需要保留原串列** → 使用 `sorted()`
- **想要直接修改原串列** → 使用 `sort()`

## 🔃 reverse 反向排序

有時候我們想要**由大到小**排序，可以使用 `reverse=True` 參數。

### 範例 1：數字由大到小

````python
# 預設：由小到大
numbers = [5, 2, 8, 1, 9]
print("由小到大：", sorted(numbers))

# 反向：由大到小
print("由大到小：", sorted(numbers, reverse=True))

# 預期輸出：
# 由小到大： [1, 2, 5, 8, 9]
# 由大到小： [9, 8, 5, 2, 1]
````

### 範例 3：sort() 也可以使用 reverse

````python
# sort() 方法也可以使用 reverse 參數
fruit = ['kiwi', 'pear', 'grape']

print("排序前：", fruit)
fruit.sort(reverse=True)  # 直接修改，由大到小
print("排序後：", fruit)

# 預期輸出：
# 排序前： ['kiwi', 'pear', 'grape']
# 排序後： ['pear', 'kiwi', 'grape']
````

### 💡 reverse 參數總結

```python
# sorted() + reverse
sorted(list)                # 由小到大
sorted(list, reverse=True)  # 由大到小

# sort() + reverse
list.sort()                 # 由小到大（改變原串列）
list.sort(reverse=True)     # 由大到小（改變原串列）
```

**記憶技巧**：
- `reverse=False` → 正常順序（預設）
- `reverse=True` → 反轉順序

### 範例 3：sort() 也可以使用 reverse

````python
# sort() 方法也可以使用 reverse 參數
fruit = ['kiwi', 'pear', 'grape']

print("排序前：", fruit)
fruit.sort(reverse=True)  # 直接修改，由大到小
print("排序後：", fruit)

# 預期輸出：
# 排序前： ['kiwi', 'pear', 'grape']
# 排序後： ['pear', 'kiwi', 'grape']
````

### 💡 reverse 參數總結

```python
# sorted() + reverse
sorted(list)                # 由小到大
sorted(list, reverse=True)  # 由大到小

# sort() + reverse
list.sort()                 # 由小到大（改變原串列）
list.sort(reverse=True)     # 由大到小（改變原串列）
```

**記憶技巧**：
- `reverse=False` → 正常順序（預設）
- `reverse=True` → 反轉順序

### 範例 2：字串由 Z 到 A

````python
# 字串反向排序
fruits = ["banana", "apple", "cherry", "date"]
print("A→Z：", sorted(fruits))
print("Z→A：", sorted(fruits, reverse=True))

# 預期輸出：
# A→Z： ['apple', 'banana', 'cherry', 'date']
# Z→A： ['date', 'cherry', 'banana', 'apple']
````

## 🎯 key 自訂排序規則

有時候預設的排序規則不符合需求，我們可以用 `key` 參數自訂排序邏輯。

**語法**：`sorted(list, key=function)`

`key` 參數接受一個函式，Python 會用這個函式的回傳值來決定排序順序。

### 基本概念

```python
sorted(list, key=某個函式)
```

- Python 會對串列中的每個元素執行「某個函式」
- 用函式的回傳值來排序
- 常用的內建函式：`len`、`str.lower`、`abs` 等

---

````python
# 基本範例：依絕對值排序
num = [-5, 2, -8, 1, 9]

print("一般排序：", sorted(num))
print("依絕對值排序：", sorted(num, key=abs))

# 預期輸出：
# 一般排序： [-8, -5, 1, 2, 9]
# 依絕對值排序： [1, 2, -5, -8, 9]
# 說明：abs(1)=1 < abs(2)=2 < abs(-5)=5 < abs(-8)=8 < abs(9)=9
````

### 💡 key 的核心概念

**重點**：
- key 函式的**回傳值**才是用來排序的依據
- 最後回傳的是**原始元素**，不是回傳值
- 可以使用內建函式，也可以自己寫函式

---

## 📏 key=len 依長度排序

`len` 是 Python 內建函式，可以計算字串長度。

### 範例：依字串長度排序

````python
# 依字串長度排序（由短到長）
words = ["python", "is", "awesome", "fun"]

result = sorted(words, key=len)
print("依長度排序：", result)

# 預期輸出：
# 依長度排序： ['is', 'fun', 'python', 'awesome']
# 說明：is(2) < fun(3) < python(6) < awesome(7)
````

### 範例 2：驗證每個字串的長度

````python
# 驗證每個字串的長度
words = ["python", "is", "awesome", "fun"]

for word in words:
    print(f"{word}: {len(word)} 個字元")

# 預期輸出：
# python: 6 個字元
# is: 2 個字元
# awesome: 7 個字元
# fun: 3 個字元
````

### 範例：由長到短排序

````python
# 結合 reverse=True
words = ["python", "is", "awesome", "fun"]

result = sorted(words, key=len, reverse=True)
print("由長到短：", result)

# 預期輸出：
# 由長到短： ['awesome', 'python', 'fun', 'is']
````

## 🔤 key=自訂函式（第二個字母）

如果內建函式無法滿足需求，我們可以自己寫一個函式。

### 範例：依第二個字母排序

````python
# 定義一個函式：回傳第二個字母
def second_letter(word):
    return word[1]

# 依第二個字母排序
words = ["apple", "banana", "cherry", "date"]
result = sorted(words, key=second_letter)

print("原始：", words)
print("依第二個字母排序：", result)

# 預期輸出：
# 原始： ['apple', 'banana', 'cherry', 'date']
# 依第二個字母排序： ['banana', 'date', 'cherry', 'apple']
# 說明：banana(a) < date(a) < cherry(h) < apple(p)
````

````python
# 驗證第二個字母
words = ["apple", "banana", "cherry", "date"]

for word in words:
    print(f"{word}: 第二個字母是 '{word[1]}'")

# 預期輸出：
# apple: 第二個字母是 'p'
# banana: 第二個字母是 'a'
# cherry: 第二個字母是 'h'
# date: 第二個字母是 'a'
````

### 範例 2：驗證第二個字母

### 💡 理解 key 的運作方式

當我們寫 `sorted(words, key=second_letter)` 時，Python 會：

1. 對每個單字執行 `second_letter()` 函式
   - `second_letter("apple")` → `"p"`
   - `second_letter("banana")` → `"a"`
   - `second_letter("cherry")` → `"h"`
   - `second_letter("date")` → `"a"`

2. 依照回傳值排序：`"a" < "a" < "h" < "p"`

3. 回傳排序後的原始單字

## 🎵 key=自訂函式（母音數量）

再來一個更複雜的例子：依母音（a, e, i, o, u）數量排序。

### 範例：計算母音數量並排序

````python
# 定義函式：計算母音數量
def count_vowels(word):
    vowels = "aeiou"
    count = 0
    for letter in word:
        if letter.lower() in vowels:
            count += 1
    return count

# 依母音數量排序
words = ["python", "hello", "world", "awesome"]

result = sorted(words, key=count_vowels)
print("原始：", words)
print("依母音數量排序：", result)

# 預期輸出：
# 原始： ['python', 'hello', 'world', 'awesome']
# 依母音數量排序： ['python', 'world', 'hello', 'awesome']
# 說明：python(1) < world(1) < hello(2) < awesome(4)
````

### 驗證一下

````python
# 驗證每個單字的母音數量
words = ["python", "hello", "world", "awesome"]

for word in words:
    print(f"{word}: {count_vowels(word)} 個母音")

# 預期輸出：
# python: 1 個母音
# hello: 2 個母音
# world: 1 個母音
# awesome: 4 個母音
````

## 📝 小結：重點整理

### 1️⃣ sorted() 基本用法
```python
sorted(list)              # 由小到大
sorted(list, reverse=True)  # 由大到小
```

### 2️⃣ sort() vs sorted()
- `sorted()` → 回傳新串列，不改變原串列
- `sort()` → 直接修改原串列，回傳 None

### 3️⃣ key 參數
```python
sorted(list, key=len)           # 依長度排序
sorted(list, key=自訂函式)        # 依自訂規則排序
```

### 4️⃣ 常見用法組合
```python
sorted(list, key=len, reverse=True)  # 由長到短
```

### 💡 記憶口訣

- **sorted** = **新**串列（兩個都有 s 開頭）
- **sort** = **改**原串列（直接修改，不囉嗦）
- **key** = **關鍵**（決定排序的關鍵函式）
