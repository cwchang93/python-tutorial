## 📖 什麼是 Dictionary？

Dictionary（字典）是 Python 中用來儲存 **key-value 配對** 的資料結構。

### 特點：
- 使用 `{}` 大括號建立
- 每個元素是「鍵:值」配對 (`key: value`)
- key 必須是唯一的
- 透過 key 來存取對應的 value

### 範例：個人資料字典

### 學習建議

- 核心先做：建立字典、讀取 key、`get()`、新增/更新資料、`in` 檢查。
- 可先延後：巢狀字典、list of dict、dict comprehension、資料清理與報表類題目。
- 這一堂要先讓學生理解『key 對應 value』，不要一開始就丟太多資料結構混合題。

````python
# 建立一個個人資料字典
profile = {'name': 'Rebel Wilson', 'year': 1980}

print(profile)
print(type(profile))
````

## 🔨 建立 Dictionary

有兩種方式可以建立字典：
1. 使用 `{}` 大括號
2. 使用 `dict()` 函數

````python
# 方法1：使用 {} 建立空字典
d1 = {}
print("d1:", d1)
print("d1 的型別:", type(d1))

print()

# 方法2：使用 dict() 建立空字典
d2 = dict()
print("d2:", d2)
print("d2 的型別:", type(d2))
````

````python
# 建立有資料的字典：各國首都
capitals = {
    '台灣': '台北',
    '美國': '華盛頓',
    '土耳其': '安卡拉',
    '日本': None,
    '澳洲': '坎培拉'
}

print(capitals)
````

## 🔍 查詢資料 d[key]

使用 `字典名稱[key]` 來取得對應的 value。

⚠️ **注意**：如果 key 不存在，會引發 `KeyError`！

````python
# 查詢台灣的首都
print(capitals['台灣'])
````

````python
# ⚠️ 先猜猜看：執行這行會發生什麼事？
# 提示：字典中有「韓國」這個 key 嗎？

print(capitals['韓國'])
````

### 💥 KeyError 錯誤

當你嘗試存取不存在的 key 時，Python 會拋出 `KeyError` 錯誤。

這時候我們需要使用 `get()` 方法來安全查詢！

---

## 🛡️ 安全查詢 get()

`get()` 方法的優點：
- 如果 key 不存在，**不會產生錯誤**
- 可以設定預設值（當 key 不存在時回傳）

### 語法：
```python
字典.get(key)           # 找不到回傳 None
字典.get(key, 預設值)   # 找不到回傳預設值
```

````python
# 使用 get() 查詢不存在的 key
result = capitals.get('韓國')
print("查詢韓國:", result)
print("型別:", type(result))
````

````python
# 使用 get() 並設定預設值
result = capitals.get('韓國', '找不到這個國家')
print(result)
````

### 📝 小結：d[key] vs get()

| 方法 | key 存在 | key 不存在 | 使用時機 |
|------|----------|------------|----------|
| `d[key]` | 回傳 value | ❌ KeyError | 確定 key 存在時 |
| `d.get(key)` | 回傳 value | ✅ 回傳 None | 不確定 key 是否存在 |
| `d.get(key, default)` | 回傳 value | ✅ 回傳 default | 需要預設值時 |

## ➕ 新增 / 更新資料

使用 `字典[key] = value` 來新增或更新資料：
- 如果 key **不存在**，會**新增**這個鍵值對
- 如果 key **已存在**，會**更新**其值

````python
# 新增資料：韓國的首都是首爾
capitals['韓國'] = '首爾'
print("新增後:", capitals)
````

````python
# 更新資料：將美國首都改為更正式的名稱
capitals['美國'] = 'Washington, D.C.'
print("更新後:", capitals)
````

## ➖ 刪除資料（del / pop）

有兩種刪除方法：
1. **del**：直接刪除，不回傳值
2. **pop()**：刪除並回傳該 value

### 準備測試資料：歷史事件

````python
# 建立歷史事件字典
events = {
    '美國獨立': 1776,
    '法國大革命': 1789,
    '武昌起義': 1911,
    '美西戰爭': 1898
}

print("原始資料:", events)
````

````python
# 使用 del 刪除
del events['美西戰爭']
print("刪除美西戰爭後:", events)
````

````python
# ⚠️ 先猜猜看：執行這行會發生什麼？
# 提示：美西戰爭已經被刪除了！

print(events['美西戰爭'])
````

````python
# 使用 pop() 刪除並取得值
k = events.pop('美國獨立')

print("pop 回傳的值:", k)
print("刪除後的字典:", events)
````

### 📝 小結：del vs pop()

| 方法 | 語法 | 回傳值 | 使用時機 |
|------|------|--------|----------|
| `del` | `del d[key]` | 無 | 只想刪除，不需要值 |
| `pop()` | `d.pop(key)` | 回傳被刪除的 value | 刪除前想取得該值 |

## 🔄 Dictionary 與 List 的比較（==）

### List 的比較：順序很重要

````python
# List 比較
a = [1, 2, 3]
b = [2, 3, 1]  # 元素相同，順序不同
c = [1, 2, 3]  # 元素和順序都相同

print("a == b:", a == b)  # False，因為順序不同
print("a == c:", a == c)  # True，元素和順序都相同
````

### Dictionary 的比較：順序不重要

字典只要 key-value 配對相同，**順序不同也視為相等**！

````python
# Dictionary 比較
x = {'A': 1, 'B': 2, 'C': 3}
y = {'B': 2, 'C': 3, 'A': 1}  # 順序不同，但 key-value 配對相同
z = {'A': 1, 'B': 2, 'C': 3}  # 完全相同

print("x == y:", x == y)  # True，字典不在乎順序
print("x == z:", x == z)  # True
````

## 🔁 走訪 Dictionary（for 迴圈）

### 方法1：直接走訪（預設走訪 keys）

````python
# 建立測試字典
d = {'name': 'Rebel Wilson', 'year': 1980}

# 方法1a：直接用 for key in d
print("方法1a：直接走訪")
for key in d:
    print(key, d[key])
````

````python
# 方法1b：明確使用 keys()
print("方法1b：使用 keys()")
for key in d.keys():
    print(key, d[key])
````

### 方法2：走訪 values

````python
# 只印出所有的 value
print("方法2a：使用 values()")
for value in d.values():
    print(value)
````

### 方法3：使用 items()（推薦）

使用 `items()` 可以同時取得 key 和 value，**最方便**！

````python
# 方法3b：使用 items() 同時取得 key 和 value
print("方法3b：使用 items()（推薦）")
for key, value in d.items():
    print(key, value)
````

### 練習 4：加分題 💡

已知密度字典：`density = {'Au': 19.3, 'Fe': 7.87, 'Hg': 13.53, 'Cu': 8.92, 'Ag': 10.49}`

請使用 `for` 迴圈配合 `values()` 方法，計算所有密度值的總和。

預期輸出：
```
所有密度的總和: 60.11
```

````python
# TODO：練習 4（加分題）
density = {'Au': 19.3, 'Fe': 7.87, 'Hg': 13.53, 'Cu': 8.92, 'Ag': 10.49}

# 使用 for + values() 計算總和
````

````python
# ✅ 參考解答：練習 4
density = {'Au': 19.3, 'Fe': 7.87, 'Hg': 13.53, 'Cu': 8.92, 'Ag': 10.49}

# 方法1：使用 for + values()
total = 0
for value in density.values():
    total += value

print("所有密度的總和:", total)

# 方法2：使用 sum() 函數（更簡潔）
total2 = sum(density.values())
print("所有密度的總和:", total2)
````
