## 第 1 題:Python 基礎語法 - Two Sum(兩數之和)

### 題目描述
給定一個整數陣列 `nums` 和一個目標值 `target`,請找出陣列中兩個數字,使它們的和等於目標值,並回傳這兩個數字的**索引**。

### 範例
```python
輸入:nums = [2, 7, 11, 15], target = 9
輸出:[0, 1]
說明:因為 nums[0] + nums[1] = 2 + 7 = 9

輸入:nums = [3, 2, 4], target = 6
輸出:[1, 2]
說明:因為 nums[1] + nums[2] = 2 + 4 = 6

輸入:nums = [3, 3], target = 6
輸出:[0, 1]
```

### 要求
1. 每個輸入只會有**唯一解**
2. 不能使用同一個元素兩次
3. 請使用字典來優化效能

### 提示
- 可以用字典記錄已經遍歷過的數字及其索引
- 對於每個數字 `num`,檢查 `target - num` 是否在字典中
- 時間複雜度可以達到 O(n)

````python
# 請在此作答

# ========== 方法 1：雙重迴圈（暴力解法，初學者友善）==========
def two_sum_simple(nums, target):
    """
    找出兩個數字的索引,使其和等於目標值（使用雙重迴圈）
    
    參數:
        nums: 整數陣列
        target: 目標值
    
    回傳:
        包含兩個索引的列表
        
    時間複雜度: O(n²) - 兩層迴圈
    空間複雜度: O(1) - 不需要額外空間
    """
    # 外層迴圈：遍歷每個數字
    for i in range(len(nums)):
        # 內層迴圈：從 i+1 開始，避免使用同一個元素兩次
        for j in range(i + 1, len(nums)):
            # 檢查兩個數字的和是否等於目標值
            if nums[i] + nums[j] == target:
                return [i, j]
    
    # 如果沒有找到答案
    return None


# ========== 方法 2：字典優化（推薦解法）==========
def two_sum(nums, target):
    """
    找出兩個數字的索引,使其和等於目標值（使用字典優化）
    
    參數:
        nums: 整數陣列
        target: 目標值
    
    回傳:
        包含兩個索引的列表
        
    時間複雜度: O(n) - 只需遍歷一次
    空間複雜度: O(n) - 需要字典儲存
    """
    # 建立字典來記錄已經遍歷過的數字及其索引
    num_dict = {}
    
    # 遍歷陣列
    for i, num in enumerate(nums):
        # 計算需要的另一個數字
        complement = target - num
        
        # 檢查這個數字是否已經在字典中
        if complement in num_dict:
            # 找到答案，回傳兩個索引
            return [num_dict[complement], i]
        
        # 將當前數字和索引存入字典
        num_dict[num] = i
    
    # 如果沒有找到答案
    return None


# ========== 測試兩種方法 ==========
test_cases = [
    ([2, 7, 11, 15], 9),
    ([3, 2, 4], 6),
    ([3, 3], 6)
]

print("========== 方法 1：雙重迴圈（簡單易懂）==========\n")
for nums, target in test_cases:
    result = two_sum_simple(nums, target)
    print(f"輸入: nums = {nums}, target = {target}")
    print(f"輸出: {result}")
    if result:
        print(f"驗證: nums[{result[0]}] + nums[{result[1]}] = {nums[result[0]]} + {nums[result[1]]} = {nums[result[0]] + nums[result[1]]}\n")

print("\n" + "="*60 + "\n")
print("========== 方法 2：字典優化（效能更好）==========\n")
for nums, target in test_cases:
    result = two_sum(nums, target)
    print(f"輸入: nums = {nums}, target = {target}")
    print(f"輸出: {result}")
    if result:
        print(f"驗證: nums[{result[0]}] + nums[{result[1]}] = {nums[result[0]]} + {nums[result[1]]} = {nums[result[0]] + nums[result[1]]}\n")

print("\n" + "="*60)
print("💡 學習重點：")
print("  • 方法 1 容易理解，但當資料量大時會比較慢")
print("  • 方法 2 使用空間換取時間，效能更好")
print("  • 實際應用中建議使用方法 2")
````

## 第 2 題：演算法思維 - 找出陣列中的「孤獨數字」

### 題目描述
給定一個整數陣列，其中每個數字都出現**兩次**，只有一個數字出現**一次**。請找出這個只出現一次的數字。

### 範例
```python
輸入：[4, 1, 2, 1, 2]
輸出：4

輸入：[2, 2, 1]
輸出：1

輸入：[7, 3, 5, 3, 7]
輸出：5
```

### 要求
1. 不能使用內建的 `count()` 方法直接數每個數字出現次數
2. 請用字典（dictionary）來記錄每個數字的出現次數
3. 找出只出現一次的數字

### 提示
- 使用字典記錄：`{數字: 出現次數}`
- 遍歷字典，找出次數為 1 的數字
- 進階挑戰：能否用 XOR 位元運算來解？

````python
# 請在此作答
def find_single_number(nums):
    """
    找出陣列中只出現一次的數字
    
    參數:
        nums: 整數陣列
    
    回傳:
        只出現一次的數字
    """
    # 方法1：使用字典記錄每個數字的出現次數
    count_dict = {}
    
    # 統計每個數字出現的次數
    for num in nums:
        if num in count_dict:
            count_dict[num] += 1
        else:
            count_dict[num] = 1
    
    # 找出只出現一次的數字
    for num, count in count_dict.items():
        if count == 1:
            return num
    
    # 方法2（進階）：使用 XOR 位元運算
    # result = 0
    # for num in nums:
    #     result ^= num
    # return result
    # 原理：a ^ a = 0, a ^ 0 = a, XOR 具有交換律和結合律
    # 所有出現兩次的數字會互相抵消為 0，剩下只出現一次的數字

# 測試
test_cases = [
    [4, 1, 2, 1, 2],
    [2, 2, 1],
    [7, 3, 5, 3, 7]
]

for nums in test_cases:
    result = find_single_number(nums)
    print(f"輸入：{nums}")
    print(f"輸出：{result}\n")
````

## 第 3 題:物件導向程式設計 (OOP) - 銀行帳戶系統

### 題目描述
設計一個簡單的銀行帳戶類別 `BankAccount`。

### 要求

#### BankAccount 類別
- 屬性:
  - `account_number`(帳號)
  - `owner`(戶名)
  - `balance`(餘額,預設為 0)

- 方法:
  - `__init__(account_number, owner, balance=0)`:建構子
  - `deposit(amount)`:存款
    - 檢查金額是否大於 0
    - 更新餘額並印出訊息
  - `withdraw(amount)`:提款
    - 檢查金額是否大於 0
    - 檢查餘額是否足夠
    - 更新餘額並印出訊息
  - `get_balance()`:查詢餘額
  - `__str__()`:回傳帳戶資訊字串

### 測試要求
1. 建立一個帳戶(初始餘額 1000)
2. 存入 500 元
3. 提款 300 元
4. 嘗試提款 2000 元(餘額不足)
5. 查詢最終餘額

````python
# 請在此作答

class BankAccount:
    """銀行帳戶類別"""
    
    def __init__(self, account_number, owner, balance=0):
        """
        建構子
        參數:
            account_number: 帳號
            owner: 戶名
            balance: 初始餘額（預設為 0）
        """
        self.account_number = account_number
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        """
        存款
        參數:
            amount: 存款金額
        """
        if amount <= 0:
            print("❌ 存款金額必須大於 0")
            return
        
        self.balance += amount
        print(f"✅ 存款成功！存入 ${amount:,} 元，目前餘額：${self.balance:,} 元")
    
    def withdraw(self, amount):
        """
        提款
        參數:
            amount: 提款金額
        """
        if amount <= 0:
            print("❌ 提款金額必須大於 0")
            return
        
        if amount > self.balance:
            print(f"❌ 餘額不足！目前餘額：${self.balance:,} 元，無法提領 ${amount:,} 元")
            return
        
        self.balance -= amount
        print(f"✅ 提款成功！提領 ${amount:,} 元，目前餘額：${self.balance:,} 元")
    
    def get_balance(self):
        """查詢餘額"""
        print(f"💰 帳戶餘額：${self.balance:,} 元")
        return self.balance
    
    def __str__(self):
        """回傳帳戶資訊字串"""
        return f"帳號：{self.account_number} | 戶名：{self.owner} | 餘額：${self.balance:,} 元"


# 測試程式碼
print("========== 銀行帳戶測試 ==========\n")

# 建立帳戶
account = BankAccount("123-456-789", "王小明", 1000)
print(f"建立帳戶：{account}\n")

# 存款
account.deposit(500)
print()

# 提款
account.withdraw(300)
print()

# 嘗試提款超過餘額
account.withdraw(2000)
print()

# 查詢餘額
account.get_balance()
print()

# 顯示帳戶資訊
print(f"最終狀態：{account}")
````

## 第 4 題：Pandas 基礎 - 讀取與分析學生成績資料

### 題目描述
使用 Pandas 讀取學生成績資料，並進行基本的統計分析。

### 資料說明
我們提供以下學生成績資料（先建立 CSV 檔案）：

| 學號 | 姓名 | 數學 | 英文 | 程式設計 | 班級 |
|------|------|------|------|----------|------|
| S001 | 王小明 | 85 | 78 | 92 | A |
| S002 | 李小華 | 92 | 88 | 85 | A |
| S003 | 張小芳 | 78 | 92 | 88 | B |
| S004 | 陳小強 | 88 | 85 | 90 | A |
| S005 | 林小美 | 95 | 90 | 95 | B |

### 要求
1. 將上述資料建立為 DataFrame（可以用字典或直接建立 CSV）
2. 顯示前 3 筆資料
3. 顯示資料的基本資訊（shape, columns, dtypes）
4. 計算每科的平均分數
5. 找出數學成績最高的學生
6. 計算每位學生的總分和平均分數（新增欄位）
7. 找出總分前 3 名的學生
8. 統計各班級的人數

````python
# 請在此作答
import pandas as pd

# 步驟 1：建立資料
data = {
    '學號': ['S001', 'S002', 'S003', 'S004', 'S005'],
    '姓名': ['王小明', '李小華', '張小芳', '陳小強', '林小美'],
    '數學': [85, 92, 78, 88, 95],
    '英文': [78, 88, 92, 85, 90],
    '程式設計': [92, 85, 88, 90, 95],
    '班級': ['A', 'A', 'B', 'A', 'B']
}

df = pd.DataFrame(data)

print("========== Pandas 學生成績分析 ==========\n")

# 2. 顯示前 3 筆資料
print("2. 顯示前 3 筆資料:")
print(df.head(3))
print("\n" + "="*50 + "\n")

# 3. 顯示基本資訊
print("3. 資料基本資訊:")
print(f"資料形狀 (shape): {df.shape}")
print(f"欄位名稱 (columns): {list(df.columns)}")
print(f"\n資料型態 (dtypes):")
print(df.dtypes)
print("\n" + "="*50 + "\n")

# 4. 計算每科平均分數
print("4. 每科平均分數:")
subjects = ['數學', '英文', '程式設計']
for subject in subjects:
    avg = df[subject].mean()
    print(f"{subject}：{avg:.2f} 分")
print("\n" + "="*50 + "\n")

# 5. 找出數學成績最高的學生
print("5. 數學成績最高的學生:")
max_math_idx = df['數學'].idxmax()
best_math_student = df.loc[max_math_idx]
print(f"姓名：{best_math_student['姓名']}")
print(f"數學成績：{best_math_student['數學']} 分")
print("\n" + "="*50 + "\n")

# 6. 計算每位學生的總分和平均
print("6. 計算每位學生的總分和平均:")
df['總分'] = df[subjects].sum(axis=1)
df['平均'] = df[subjects].mean(axis=1)
print(df[['學號', '姓名', '總分', '平均']])
print("\n" + "="*50 + "\n")

# 7. 找出總分前 3 名
print("7. 總分前 3 名的學生:")
top3 = df.nlargest(3, '總分')[['姓名', '總分', '平均']]
print(top3)
print("\n" + "="*50 + "\n")

# 8. 統計各班級人數
print("8. 各班級人數統計:")
class_count = df['班級'].value_counts()
print(class_count)
print("\n完整資料表:")
print(df)
````

## 第 5 題：資料清洗 - 處理髒資料

### 題目描述
你收到一份員工薪資資料，但資料很髒（有缺失值、重複、異常值等）。請進行資料清洗。

### 髒資料說明
資料包含以下問題：
1. 有缺失值（NaN）
2. 有重複的員工記錄
3. 薪資有異常值（負數或過大）
4. 年齡有不合理的值
5. 部門名稱有大小寫不一致

### 要求
1. 載入資料並檢查問題
2. 處理缺失值：
   - 姓名缺失：刪除該筆資料
   - 年齡缺失：填入該部門的平均年齡
   - 薪資缺失：填入該部門的中位數薪資
3. 刪除重複的員工記錄（保留第一筆）
4. 處理異常值：
   - 薪資 < 0 或 > 200000：刪除
   - 年齡 < 18 或 > 65：刪除
5. 統一部門名稱為大寫
6. 顯示清洗前後的資料筆數
7. 顯示清洗後的統計摘要

````python
# 建立髒資料
import pandas as pd
import numpy as np

# 建立包含各種問題的髒資料
dirty_data = {
    '員工ID': ['E001', 'E002', 'E003', 'E004', 'E005', 'E006', 'E007', 'E008', 'E009', 'E010', 'E003', 'E011'],
    '姓名': ['張三', '李四', '王五', None, '趙六', '錢七', '孫八', '周九', '吳十', '鄭十一', '王五', '陳十二'],
    '年齡': [28, 35, 150, 42, None, 30, 25, 38, 10, 45, 41, 33],
    '部門': ['IT', 'hr', 'IT', 'Sales', 'HR', 'it', 'Sales', 'IT', 'sales', 'HR', 'IT', 'Sales'],
    '薪資': [50000, 45000, 60000, -5000, None, 55000, 48000, 300000, 42000, None, 60000, 50000]
}

df_dirty = pd.DataFrame(dirty_data)

print("========== 原始髒資料 ==========\n")
print(df_dirty)
print(f"\n原始資料筆數：{len(df_dirty)}")
print(f"\n缺失值統計：")
print(df_dirty.isnull().sum())

# ========== 開始資料清洗 ==========

# 複製一份資料進行清洗
df_clean = df_dirty.copy()

print("\n\n========== 開始資料清洗 ==========\n")

# 步驟 1：統一部門名稱為大寫（先做這步，方便後續分組計算）
print("步驟 1：統一部門名稱為大寫")
df_clean['部門'] = df_clean['部門'].str.upper()
print(df_clean['部門'].value_counts())

# 步驟 2：處理缺失值
print("\n步驟 2：處理缺失值")

# 2a. 姓名缺失：刪除該筆資料
print("  - 刪除姓名缺失的資料")
before_count = len(df_clean)
df_clean = df_clean.dropna(subset=['姓名'])
print(f"    刪除了 {before_count - len(df_clean)} 筆資料")

# 2b. 年齡缺失：填入該部門的平均年齡
print("  - 年齡缺失：填入該部門的平均年齡")
dept_avg_age = df_clean.groupby('部門')['年齡'].transform('mean')
df_clean['年齡'] = df_clean['年齡'].fillna(dept_avg_age)

# 2c. 薪資缺失：填入該部門的中位數薪資
print("  - 薪資缺失：填入該部門的中位數薪資")
dept_median_salary = df_clean.groupby('部門')['薪資'].transform('median')
df_clean['薪資'] = df_clean['薪資'].fillna(dept_median_salary)

# 步驟 3：刪除重複的員工記錄（保留第一筆）
print("\n步驟 3：刪除重複的員工記錄")
before_count = len(df_clean)
df_clean = df_clean.drop_duplicates(subset=['員工ID'], keep='first')
print(f"  刪除了 {before_count - len(df_clean)} 筆重複資料")

# 步驟 4：處理異常值
print("\n步驟 4：處理異常值")

# 4a. 薪資異常：< 0 或 > 200000
print("  - 刪除薪資異常的資料（< 0 或 > 200000）")
before_count = len(df_clean)
df_clean = df_clean[(df_clean['薪資'] >= 0) & (df_clean['薪資'] <= 200000)]
print(f"    刪除了 {before_count - len(df_clean)} 筆資料")

# 4b. 年齡異常：< 18 或 > 65
print("  - 刪除年齡異常的資料（< 18 或 > 65）")
before_count = len(df_clean)
df_clean = df_clean[(df_clean['年齡'] >= 18) & (df_clean['年齡'] <= 65)]
print(f"    刪除了 {before_count - len(df_clean)} 筆資料")

# 重新設定索引
df_clean = df_clean.reset_index(drop=True)

print("\n" + "="*50)
print("\n========== 清洗後的資料 ==========\n")
print(df_clean)
print(f"\n清洗後資料筆數：{len(df_clean)}")
print(f"清洗掉的資料筆數：{len(df_dirty) - len(df_clean)}")

print("\n========== 清洗後的統計摘要 ==========\n")
print(df_clean.describe())

print("\n========== 各部門統計 ==========\n")
dept_stats = df_clean.groupby('部門').agg({
    '年齡': ['mean', 'min', 'max'],
    '薪資': ['mean', 'median', 'min', 'max'],
    '員工ID': 'count'
})
dept_stats.columns = ['平均年齡', '最小年齡', '最大年齡', '平均薪資', '中位數薪資', '最低薪資', '最高薪資', '人數']
print(dept_stats)
````

- 使用 `plt.title()`, `plt.xlabel()`, `plt.ylabel()` 設定標題和標籤
- 使用 `plt.xticks(rotation=45)` 旋轉 x 軸標籤（避免重疊）
- 使用 `plt.tight_layout()` 自動調整子圖間距

````python
# 請在此作答
import pandas as pd
import matplotlib.pyplot as plt

# 設定中文字型（避免亂碼）
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'Microsoft JhengHei', 'SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 準備資料（使用第 4 題的資料）
data = {
    '學號': ['S001', 'S002', 'S003', 'S004', 'S005'],
    '姓名': ['王小明', '李小華', '張小芳', '陳小強', '林小美'],
    '數學': [85, 92, 78, 88, 95],
    '英文': [78, 88, 92, 85, 90],
    '程式設計': [92, 85, 88, 90, 95],
    '班級': ['A', 'A', 'B', 'A', 'B']
}
df = pd.DataFrame(data)
subjects = ['數學', '英文', '程式設計']
df['總分'] = df[subjects].sum(axis=1)

# 1. 長條圖：每位學生的總分
print("繪製圖表 1：長條圖 - 每位學生的總分")
plt.figure(figsize=(10, 6))
plt.bar(df['姓名'], df['總分'], color='steelblue', alpha=0.8)
plt.title('學生總分長條圖', fontsize=16, fontweight='bold')
plt.xlabel('學生姓名', fontsize=12)
plt.ylabel('總分', fontsize=12)
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
# 在長條上顯示數值
for i, v in enumerate(df['總分']):
    plt.text(i, v + 3, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()

# 2. 折線圖：各科平均分數
print("\n繪製圖表 2：折線圖 - 各科平均分數")
plt.figure(figsize=(10, 6))
avg_scores = df[subjects].mean()
plt.plot(subjects, avg_scores, marker='o', linewidth=2, markersize=10, color='coral')
plt.title('各科平均分數折線圖', fontsize=16, fontweight='bold')
plt.xlabel('科目', fontsize=12)
plt.ylabel('平均分數', fontsize=12)
plt.grid(True, alpha=0.3)
# 在數據點上顯示數值
for i, v in enumerate(avg_scores):
    plt.text(i, v + 1, f'{v:.1f}', ha='center', fontweight='bold')
plt.ylim(80, 92)
plt.tight_layout()
plt.show()

# 3. 圓餅圖：各班級人數比例
print("\n繪製圖表 3：圓餅圖 - 各班級人數比例")
plt.figure(figsize=(8, 8))
class_counts = df['班級'].value_counts()
colors = ['#ff9999', '#66b3ff']
plt.pie(class_counts, labels=class_counts.index, autopct='%1.1f%%', 
        startangle=90, colors=colors, textprops={'fontsize': 14, 'fontweight': 'bold'})
plt.title('各班級人數比例', fontsize=16, fontweight='bold')
plt.legend([f'{cls}班：{count}人' for cls, count in class_counts.items()], 
           loc='upper right')
plt.tight_layout()
plt.show()

# 4. 子圖：三科成績分佈
print("\n繪製圖表 4：子圖 - 三科成績分佈直方圖")
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
colors_hist = ['skyblue', 'lightcoral', 'lightgreen']

for i, subject in enumerate(subjects):
    axes[i].hist(df[subject], bins=5, color=colors_hist[i], alpha=0.7, edgecolor='black')
    axes[i].set_title(f'{subject}成績分佈', fontsize=14, fontweight='bold')
    axes[i].set_xlabel('分數', fontsize=11)
    axes[i].set_ylabel('人數', fontsize=11)
    axes[i].grid(axis='y', alpha=0.3)
    
    # 顯示平均分數線
    mean_score = df[subject].mean()
    axes[i].axvline(mean_score, color='red', linestyle='--', linewidth=2, label=f'平均: {mean_score:.1f}')
    axes[i].legend()

plt.suptitle('三科成績分佈直方圖', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n✅ 所有圖表繪製完成！")
````
