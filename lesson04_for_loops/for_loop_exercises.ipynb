### For Loop 三種常見用法

1. **使用 range()** - 重複執行 n 次
```python
for i in range(5):
    print(i)
```

2. **迭代 list** - 遍歷串列中的每個元素
```python
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)
```

3. **迭代 string** - 遍歷字串中的每個字元
```python
for char in "Hello":
    print(char)
```
