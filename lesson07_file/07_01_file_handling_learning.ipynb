## 2. open() 與 mode

### 語法

```python
f = open(filename, mode='r')
```

### mode 參數說明

| mode | 說明 |
|------|------|
| `'r'` | **讀取模式**（Read）：檔案必須存在，否則會出錯 |
| `'w'` | **寫入模式**（Write）：會覆蓋原有內容，檔案不存在會自動建立 |
| `'a'` | **附加模式**（Append）：從檔案結尾續寫，不會覆蓋原有內容 |

### 補充：相對路徑與同資料夾概念

- 如果只寫檔名（如 `'file1.txt'`），檔案會建立在**目前工作目錄**
- 可以使用相對路徑（如 `'data/file1.txt'`）或絕對路徑
- 建議先確認目前工作目錄：`import os; print(os.getcwd())`

````python
# 查看目前工作目錄
import os
print("目前工作目錄：", os.getcwd())
````

## 3. close() 關閉檔案

### 為什麼要 close()？

使用完檔案後，**一定要關閉檔案**！

```python
f.close()
```

### 如果忘記 close() 可能造成的問題

1. **資料遺失**：寫入的資料可能沒有真正存到硬碟
2. **檔案被鎖定**：其他程式可能無法讀取或修改該檔案
3. **系統資源浪費**：開啟太多檔案會消耗系統資源

### 良好習慣

```python
f = open('file.txt', 'r')
# ... 操作檔案 ...
f.close()  # 用完立刻關閉
```

## 5. 讀取：read()

### read() 方法

- `f.read()`：讀取**全部內容**，回傳字串
- `f.read(size)`：讀取指定的字元數

### 範例 1：讀取全部內容

````python
# 讀取全部內容
f = open('file1.txt', 'r')
content = f.read()
print(content)
f.close()

# 預期輸出：
# I love python
# I love JWClub
````

### 範例 2：讀取指定字元數

⚠️ **重要**：讀取後，游標會移動到讀取結束的位置！

````python
# 讀取前 6 個字元
f = open('file1.txt', 'r')
content = f.read(6)
print(content)
f.close()

# 輸出：I love
````

## 6. 游標概念（重要！）

### 什麼是游標？

當你開啟檔案時，系統會有一個「游標」指向目前讀取的位置：

1. **初始位置**：開啟檔案時，游標在檔案開頭
2. **讀取後移動**：每次讀取，游標會往後移動
3. **讀完在結尾**：讀完整個檔案後，游標在檔案結尾
4. **再讀會空白**：游標在結尾時再讀，會得到空字串

### 重要觀念

- 如果想從頭再讀一次，必須**重新開啟檔案**
- 關閉檔案後，就不能再讀取了

````python
# 示範游標概念
f = open('file1.txt', 'r')

# 第一次讀取
print("第一次讀取：")
print(f.read())

# 第二次讀取（游標已在結尾）
print("\n第二次讀取（沒有重新 open）：")
print(f.read())  # 會是空的

f.close()
````

## 7. 讀取：readline()

### readline() 方法

- `f.readline()`：讀取**一行**（包含結尾的 `\n`）
- 每次呼叫會讀取下一行
- 適合逐行處理檔案

### 範例：讀兩行並交換順序印出

````python
# 讀兩行並交換順序
f = open('file1.txt', 'r')
line1 = f.readline()
line2 = f.readline()
f.close()

# 交換順序印出（使用 end='' 避免多餘換行）
print(line2, end='')
print(line1, end='')

# 預期輸出：
# I love JWClub
# I love python
````

## 8. 讀取：readlines()

### readlines() 方法

- `f.readlines()`：讀取**所有行**，回傳 **list**
- list 中每個元素是一行（包含結尾的 `\n`）
- 適合需要一次處理所有行的情況

### 範例 1：取得 list

````python
# 讀取所有行到 list
f = open('file1.txt', 'r')
a = f.readlines()
print(a)
f.close()

# 預期輸出：['I love python\n', 'I love JWClub']
````

### 範例 2：搭配 for 迴圈逐行處理

````python
# 使用 for 迴圈印出每一行
f = open('file1.txt', 'r')
for line in f.readlines():
    print(line, end='')  # end='' 避免額外換行
f.close()

# 預期輸出：
# I love python
# I love JWClub
````

## 9. 寫入：write()

### write() 方法

- `f.write(string)`：將字串寫入檔案
- **不會自動換行**，需要自己加 `\n`

### mode 的差異

- `'w'` 模式：**覆寫**整個檔案（原內容會被刪除）
- `'a'` 模式：**續寫**到檔案結尾（不刪除原內容）

### 範例：建立 file2.txt

````python
# 建立 file2.txt 並寫入內容
f = open('file2.txt', 'w')
f.write('JWClub is the best\n')  # 注意要加 \n
f.write("Let's learn Python")
f.close()

print("file2.txt 建立完成！")

# 驗證內容
f = open('file2.txt', 'r')
print(f.read())
f.close()

# 預期輸出：
# JWClub is the best
# Let's learn Python
````

## 10. 寫入：writelines()

### writelines() 方法

- `f.writelines(seq)`：將序列（list、tuple 等）中的字串寫入檔案
- **不會自動加換行**，需要自己在每個元素加 `\n`

### 範例：建立 file3.txt

````python
# 建立 file3.txt 並使用 writelines()
seq = ['JWClub is the best\n', "Let's learn Python"]

f = open('file3.txt', 'w')
f.writelines(seq)
f.close()

print("file3.txt 建立完成！")

# 驗證內容
f = open('file3.txt', 'r')
print(f.read())
f.close()

# 預期輸出：
# JWClub is the best
# Let's learn Python
````

## 11. 寫入：print(..., file=f)

### print() 的 file 參數

- `print()` 有一個 `file` 參數，可以指定輸出目標
- 預設是輸出到螢幕，但可以改為輸出到檔案
- **優點**：會自動加換行，而且可以直接印 list、dict 等資料結構

### 語法

```python
print(內容, file=檔案物件)
```

### 範例：將 movies list 寫入 file4.txt

````python
# 建立 file4.txt 並使用 print(file=f)
movies = ['Up', 'Inside Out', 'Titanic', 'Forrest Gump', 'Mad Max: Fury Road']

f = open('file4.txt', 'w')
print(movies, file=f)
f.close()

print("file4.txt 建立完成！")

# 驗證內容
f = open('file4.txt', 'r')
print(f.read())
f.close()

# 預期輸出：
# ['Up', 'Inside Out', 'Titanic', 'Forrest Gump', 'Mad Max: Fury Road']
````

## 12. 本章重點整理

### 檔案操作流程

1. **開啟檔案**：`f = open(filename, mode)`
2. **操作檔案**：讀取或寫入
3. **關閉檔案**：`f.close()`

### 讀取方法

| 方法 | 說明 | 回傳型別 |
|------|------|----------|
| `f.read()` | 讀取全部內容 | str |
| `f.read(size)` | 讀取指定字元數 | str |
| `f.readline()` | 讀取一行 | str |
| `f.readlines()` | 讀取所有行 | list |

### 寫入方法

| 方法 | 說明 | 自動換行 |
|------|------|----------|
| `f.write(str)` | 寫入字串 | ❌ 需自己加 `\n` |
| `f.writelines(seq)` | 寫入序列 | ❌ 需自己加 `\n` |
| `print(..., file=f)` | 使用 print 寫入 | ✅ 自動換行 |

### ⚠️ 新手常見地雷

1. **忘記 close()**
   - 可能導致資料遺失或檔案被鎖定
   - 記得：**用完就關**

2. **mode 用錯覆寫**
   - `'w'` 會刪除原有內容，要續寫請用 `'a'`
   - 先確認是要覆寫還是續寫

3. **忘記加 `\n`**
   - `write()` 和 `writelines()` 不會自動換行
   - 要換行就要自己加 `\n`

4. **路徑錯誤 FileNotFoundError**
   - 確認檔案路徑是否正確
   - 可用 `os.getcwd()` 查看目前工作目錄

5. **游標位置混淆**
   - 讀取後游標會移動，再讀會從目前位置開始
   - 要從頭讀需要重新開啟檔案

6. **檔案不存在**
   - `'r'` 模式需要檔案存在，否則會報錯
   - `'w'` 和 `'a'` 模式會自動建立檔案
