## Lesson 04｜for / while 迴圈

### 題目 1：計算一週總營收

```python
revenues = [1200, 980, 1350, 1500, 1100, 1680, 1420]

total = 0
for revenue in revenues:
    total += revenue

avg = total / len(revenues)
print("總營收 =", total)
print("平均營收 =", avg)
```

### 題目 2：數高單價訂單有幾筆

```python
orders = [120, 560, 80, 999, 430, 1250, 300]

count = 0
for amount in orders:
    if amount >= 500:
        count += 1

print("高單價訂單數 =", count)
```

### 題目 3：找第一個缺貨商品

```python
stocks = [12, 8, 5, 0, 7, 3]

i = 0
while i < len(stocks):
    if stocks[i] == 0:
        print("第一個缺貨索引 =", i)
        break
    i += 1
```

## Lesson 05｜函式、return、錯誤觀念

### 題目 1：寫一個平均訂單金額函式

```python
def calc_avg_order(amounts):
    return sum(amounts) / len(amounts)


data = [120, 200, 350, 400]
print(calc_avg_order(data))
```

### 題目 2：寫一個判斷高價訂單函式

```python
def is_large_order(amount):
    if amount >= 1000:
        return True
    else:
        return False


print(is_large_order(800))
print(is_large_order(1000))
print(is_large_order(2500))
```

### 題目 3：安全轉換輸入資料

```python
def safe_to_int(text):
    try:
        return int(text)
    except ValueError:
        return None


print(safe_to_int("120"))
print(safe_to_int("abc"))
print(safe_to_int("50"))
```

## Lesson 11-12｜資料分析入門：pandas + matplotlib

## 題組 A：讀取與認識訂單資料

```python
import pandas as pd

df = pd.read_csv("../lesson11_12_data/orders.csv")
print(df.head())
print("shape =", df.shape)
print("columns =", list(df.columns))
df.info()
```

## 題組 B：做每日營收分析

```python
df["date"] = pd.to_datetime(df["date"])
daily_revenue = df.groupby("date")["order_amount"].sum()

print(daily_revenue)
print("最高營收日 =", daily_revenue.idxmax())
print("平均每日營收 =", daily_revenue.mean())
```

## 題組 C：畫出每日營收圖

```python
import matplotlib.pyplot as plt

daily_revenue.plot(kind="line", marker="o", figsize=(8, 4))
plt.title("Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Lesson 13-14｜pandas 資料清理與 profiling

## 題組 A：先做 profiling

```python
dirty_df = pd.read_csv("../lesson13_14_pandas/data/dirty_orders.csv")

print(dirty_df.head())
print(dirty_df.tail())
print(dirty_df.sample(3, random_state=42))
print("shape =", dirty_df.shape)
dirty_df.info()
print(dirty_df.describe(include="all"))

print("觀察到的問題：")
print("1. 有些欄位型別不正確")
print("2. 有缺值")
print("3. 類別資料可能有大小寫不一致")
```

## 題組 B：做基礎清理

```python
clean_df = dirty_df.copy()

clean_df["order_date"] = pd.to_datetime(clean_df["order_date"], errors="coerce")

for col in ["order_amount", "qty", "unit_price", "shipping_fee"]:
    if col in clean_df.columns:
        clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

if "channel" in clean_df.columns:
    clean_df["channel"] = clean_df["channel"].astype(str).str.lower().str.strip()

print(clean_df.info())
print(clean_df.isna().sum())
```

## 題組 C：清理後做分析

```python
if "channel" in clean_df.columns:
    print(clean_df["channel"].value_counts(dropna=False))

if "channel" in clean_df.columns and "order_amount" in clean_df.columns:
    print(clean_df.groupby("channel")["order_amount"].mean())

missing_counts = clean_df.isna().sum()
print("缺值最多的欄位 =", missing_counts.idxmax())
print("結論：先處理日期、金額型別與缺值，才能做可靠分析")
```

## Lesson 15-16｜NumPy 資料分析

## 題組 A：月營收陣列分析

```python
import numpy as np

sales = np.array([120, 135, 128, 150, 160, 175, 168, 180, 190, 200, 210, 230])

print("ndim =", sales.ndim)
print("shape =", sales.shape)
print("dtype =", sales.dtype)
print("平均 =", sales.mean())
print("總和 =", sales.sum())
print("最大 =", sales.max())
print("最小 =", sales.min())
print("最大值月份索引 =", sales.argmax())
```

## 題組 B：用布林遮罩找出高於平均的月份

```python
avg_sales = sales.mean()
mask = sales > avg_sales
high_sales = sales[mask]

print("平均營收 =", avg_sales)
print("遮罩 =", mask)
print("高於平均的月份值 =", high_sales)
print("高於平均的月數 =", mask.sum())
```

## 題組 C：向量化計算成長率

```python
diff = sales[1:] - sales[:-1]
growth_rate = diff / sales[:-1]

print("每月差值 =", diff)
print("月成長率 =", growth_rate)
print("成長最多的一次變化 =", diff.max())
print("不用 NumPy 時，要自己處理索引、逐筆相減、組新清單，會更繁瑣")
```
