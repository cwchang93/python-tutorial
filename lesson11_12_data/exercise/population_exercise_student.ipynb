## Section 2：理解資料欄位

### 📋 這份資料有哪些欄位？

#### 1. **統計期**（文字欄位）
- 記錄時間點
- 例如：100年 1月底、100年 2月底

#### 2. **性別**（文字欄位）
- 記錄性別類別
- 有三種：總計、男、女

#### 3. **總計**（數字欄位）
- 這個「統計期 + 性別」的總人口數

#### 4. **年齡層欄位**（數字欄位）
- 未滿1歲
- 1~4歲
- 5~9歲
- 10~14歲
- ... 一直到 100歲以上

### 🤔 每一列代表什麼？

每一列代表：「某個時間點 + 某個性別」的各年齡層人口分布

例如：
- 第 1 列：100年 1月底，總計（男+女），各年齡層的人口數
- 第 2 列：100年 1月底，男，各年齡層的男性人口數
- 第 3 列：100年 1月底，女，各年齡層的女性人口數
