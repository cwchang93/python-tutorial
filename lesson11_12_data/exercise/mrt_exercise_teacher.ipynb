## Section 4：第一次分析 — 計算每個統計期的平均進站人次

### 🎯 任務：用 groupby() 計算每個統計期的平均進站人次

**老師解說：**
1. 以「統計期」當作分組依據
2. 選擇要計算的欄位「進站人次」
3. 用 `.mean()` 計算平均值

```python
# 計算每個統計期的平均進站人次
period_avg_in = df.groupby('統計期')['進站人次'].mean()

# 顯示結果（前 10 個）
print("每個統計期的平均進站人次（前 10 個）：")
print(period_avg_in.head(10))
print(f"\n總共有 {len(period_avg_in)} 個統計期")
```

### 📊 結果說明

- **groupby('統計期')**：把資料按照「統計期」分組
- **['進站人次']**：選擇要計算的欄位
- **.mean()**：計算每組的平均值
- 結果是一個 Series，index 是統計期，值是平均進站人次

## Section 5：畫折線圖 — 觀察進站人次的時間趨勢

### 🎯 任務：把分析結果畫成折線圖

**老師解說：**
- X 軸：統計期（period_avg_in 的 index）
- Y 軸：平均進站人次（period_avg_in 的值）

```python
# 畫折線圖
plt.figure(figsize=(12, 6))
plt.plot(period_avg_in.index, list(period_avg_in), marker='o', linewidth=2, markersize=5, color='forestgreen')
plt.title('各統計期的平均進站人次趨勢', fontsize=14)
plt.xlabel('統計期', fontsize=12)
plt.ylabel('平均進站人次（人）', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("\n✅ 圖表繪製完成！")
```

### 📊 圖表解讀

**觀察重點：**
1. 整體趨勢：進站人次是增加還是減少？
2. 特殊時期：有沒有哪個時期特別高或特別低？
3. 變化幅度：變化大還是相對平穩？

**教學重點：**
- 折線圖能清楚呈現時間序列的變化趨勢
- groupby() 幫助我們快速彙總資料
- 從圖表可以發現資料中的規律或異常

## Section 6：反思問題

### 🤔 問題 1：groupby() 的目的是什麼？

**答案：**

groupby() 的目的是「分組彙總」：
1. 把資料按照某個欄位（如統計期）分成多個群組
2. 對每個群組分別計算統計量（如平均值、總和等）
3. 讓我們能快速看出不同群組之間的差異

就像是「把學生按班級分組，然後計算每班的平均成績」。

---

### 🤔 問題 2：折線圖在看什麼？

**答案：**

從剛才的折線圖可以觀察：
1. **趨勢**：進站人次隨時間是上升、下降還是持平
2. **波動**：人次變化是劇烈還是平穩
3. **異常點**：是否有特別高或特別低的時期

折線圖特別適合用來觀察「連續性變化」，例如：
- 時間序列資料（如每月銷售額）
- 溫度變化
- 人口成長
