## Section 4：數字摘要統計

```python
# 查看數字欄位的統計摘要
df.describe()
```

### 📝 思考題 1：describe() 告訴我們什麼？

**答案：**
- **count**：有多少筆非空值資料
- **mean**：平均值
- **std**：標準差（數據的離散程度）
- **min**：最小值
- **25%、50%、75%**：四分位數
- **max**：最大值

---

### 📝 思考題 2：哪些欄位出現在摘要中？

**答案：**
- 「統計期」和「性別」**沒有**出現在 describe() 的結果中
- **原因**：describe() 只會顯示「數字型別」的欄位，文字欄位不會被包含

## Section 5：第一次畫圖 📊

### 🎯 任務：計算 4 個年齡層的平均人口數並畫圖

**老師解說：**
1. 先選出需要的 4 個欄位
2. 用 `.mean()` 計算每個欄位的平均值
3. 把結果畫成折線圖

```python
# Step 1：選出需要的欄位
age_cols = ['未滿1歲', '1歲', '2歲', '3歲']

# Step 2：計算平均值
mean_values = df[age_cols].mean()

# 顯示平均值
print("各年齡層的平均人口數：")
print(mean_values)
```

```python
# Step 3：畫折線圖
plt.figure(figsize=(8, 5))
plt.plot(age_cols, list(mean_values), marker='o', linewidth=2, markersize=8, color='steelblue')
plt.title('0~3歲各年齡層平均人口數', fontsize=14)
plt.xlabel('年齡層', fontsize=12)
plt.ylabel('平均人口數（人）', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

print("\n✅ 圖表繪製完成！")
```

### 📊 圖表解讀

**觀察重點：**
- 哪個年齡層的平均人口最多？
- 從「未滿1歲」到「3歲」，人口數有什麼變化趨勢？

**教學重點：**
- 這張圖幫助我們快速看出 0~3 歲人口的分布狀況
- 折線圖適合顯示「連續性」的變化趨勢
