## 🤔 函式的基本概念

### 什麼是函式？

**函式（Function）** 就像一台**機器**：

```
輸入（Input） → 【機器處理】 → 輸出（Output）
```

#### 生活化比喻：果汁機 🥤

- **輸入**：放入水果（參數）
- **處理**：機器打碎攪拌（函式內部的程式碼）
- **輸出**：得到果汁（回傳值）

### 重要觀念

- 函式可以接收**參數**（input）
- 函式會執行某些**操作**
- 函式**不一定要有回傳值**（return）
  - 有些函式只是「做事情」（例如印出訊息）
  - 有些函式會「產生結果」並回傳

---

## 📝 def 的基本語法

### 語法格式

```python
def 函式名稱(參數1, 參數2, ...):
    # 函式內部的程式碼
    # 注意：這裡要縮排！
    執行的動作
```

### 重點說明

1. **`def`**：define 的縮寫，表示「定義」一個函式
2. **函式名稱**：自己取的名字，建議使用小寫加底線（snake_case）
3. **參數**：函式需要的輸入資料，可以有 0 個、1 個或多個
4. **冒號 `:`**：不能忘記！
5. **縮排**：函式內部的程式碼必須縮排（通常是 4 個空格）

### 最簡單的範例

讓我們先看一個最基本的函式：

````python
# 定義一個最簡單的函式
def say_hello():
    print("Hello, World!")
    print("歡迎學習 Python 函式！")

# 呼叫函式
say_hello()

# 輸出：
# Hello, World!
# 歡迎學習 Python 函式！
````

## 📌 範例一：沒有 return 的函式

### 情境：判斷是否發燒

讓我們寫一個函式，接收體溫（參數），然後印出判斷結果。

- 體溫 ≥ 37.5°C：發燒
- 體溫 < 37.5°C：正常

````python
# 定義函式：檢查是否發燒
def fever(t):
    """
    檢查體溫是否發燒
    參數 t: 體溫（攝氏）
    """
    if t >= 37.5:
        print(f"體溫 {t}°C，你發燒了！請多休息。")
    else:
        print(f"體溫 {t}°C，正常體溫。")

# 測試 1：正常體溫
print("測試 1：")
fever(36.5)

print("\n測試 2：")
# 測試 2：發燒
fever(38.2)

print("\n測試 3：")
# 測試 3：臨界值
fever(37.5)

# 輸出：
# 測試 1：
# 體溫 36.5°C，正常體溫。
# 
# 測試 2：
# 體溫 38.2°C，你發燒了！請多休息。
# 
# 測試 3：
# 體溫 37.5°C，你發燒了！請多休息。
````

### 💡 重點觀察

- 這個函式**沒有 return**
- 它只是**印出訊息**
- 呼叫函式後，訊息會直接顯示在畫面上

## ⚠️ 呼叫函式的注意事項

### 1. 函式必須先定義，才能呼叫

```python
# ❌ 錯誤示範：先呼叫再定義
greet()  # 這會出錯！因為 greet 還沒定義

def greet():
    print("Hi!")
```

```python
# ✅ 正確示範：先定義再呼叫
def greet():
    print("Hi!")

greet()  # 正確！
```

### 2. 參數型態要正確

```python
def fever(t):
    if t >= 37.5:
        print("發燒了！")

# ✅ 正確
fever(38.5)

# ❌ 錯誤：傳入字串會導致比較失敗
# fever("38.5")  # 會出錯！
```

### 3. 參數數量要符合

```python
def add(a, b):
    return a + b

# ✅ 正確：傳入 2 個參數
add(3, 5)

# ❌ 錯誤：只傳入 1 個參數
# add(3)  # 會出錯！
```

## 🔄 return 的概念

### return 的功能

`return` 的作用是：**將結果回傳給呼叫者**

```python
def 函式名稱(參數):
    # 處理資料
    return 結果  # 把結果回傳
```

### return 的特性

1. **回傳值**：函式執行完後，將結果傳回
2. **中斷函式**：遇到 `return` 就立刻結束函式，後面的程式碼不會執行
3. **可以回傳任何型態**：數字、字串、串列、字典...都可以

### return vs print 的差異

| 比較項目 | `print()` | `return` |
|---------|----------|----------|
| 功能 | 在螢幕上顯示訊息 | 將結果傳回給呼叫者 |
| 回傳值 | 無（None） | 有（指定的值） |
| 後續使用 | 無法繼續使用結果 | 可以接住結果繼續運算 |
| 適用情境 | 除錯、顯示資訊 | 計算結果、資料處理 |

### 示範差異

````python
# 只有 print，沒有 return
def add_print(a, b):
    result = a + b
    print(result)

# 有 return
def add_return(a, b):
    result = a + b
    return result

# 測試 print 版本
print("=== 使用 print 版本 ===")
x = add_print(3, 5)  # 會印出 8
print(f"x 的值是：{x}")  # x 是 None！

print("\n=== 使用 return 版本 ===")
# 測試 return 版本
y = add_return(3, 5)  # 不會印出任何東西
print(f"y 的值是：{y}")  # y 是 8！

# 可以繼續運算
z = y * 2
print(f"y × 2 = {z}")

# 輸出：
# === 使用 print 版本 ===
# 8
# x 的值是：None
# 
# === 使用 return 版本 ===
# y 的值是：8
# y × 2 = 16
````

## 📌 範例二：有 return 的函式

### 改寫發燒檢查函式

讓我們把 `fever()` 改成使用 `return`，這樣可以接住結果做後續處理。

````python
# 使用 return 版本
def check_fever(t):
    """
    檢查體溫是否發燒
    參數 t: 體溫（攝氏）
    回傳: True（發燒）或 False（正常）
    """
    if t >= 37.5:
        return True  # 發燒
    else:
        return False  # 正常

# 測試並使用回傳值
temp1 = 36.5
result1 = check_fever(temp1)
print(f"體溫 {temp1}°C，發燒？{result1}")

temp2 = 38.2
result2 = check_fever(temp2)
print(f"體溫 {temp2}°C，發燒？{result2}")

# 可以根據回傳值做不同處理
print("\n=== 根據回傳值做判斷 ===")
my_temp = 37.8
if check_fever(my_temp):
    print(f"體溫 {my_temp}°C，需要休息！")
else:
    print(f"體溫 {my_temp}°C，健康良好！")

# 輸出：
# 體溫 36.5°C，發燒？False
# 體溫 38.2°C，發燒？True
# 
# === 根據回傳值做判斷 ===
# 體溫 37.8°C，需要休息！
````

### 💡 觀察重點

- 使用 `return` 後，可以**接住結果**存到變數
- 回傳值可以**繼續使用**（例如在 if 判斷中）
- 這讓函式更有彈性！

## 🎯 實戰範例：判斷閏年

### 閏年規則

一個年份是閏年，需要滿足以下條件：

1. **能被 4 整除**，且**不能被 100 整除**
2. **或者**能被 400 整除

### 範例

- 2020：能被 4 整除，不能被 100 整除 → ✅ 閏年
- 1900：能被 4 整除，能被 100 整除，不能被 400 整除 → ❌ 平年
- 2000：能被 400 整除 → ✅ 閏年

### 實作函式

````python
def check_leap_year(year):
    """
    判斷是否為閏年
    參數 year: 年份（整數）
    回傳: True（閏年）或 False（平年）
    """
    # 規則 1：能被 400 整除 → 閏年
    if year % 400 == 0:
        return True
    
    # 規則 2：能被 100 整除（但不能被 400 整除）→ 平年
    if year % 100 == 0:
        return False
    
    # 規則 3：能被 4 整除 → 閏年
    if year % 4 == 0:
        return True
    
    # 其他情況 → 平年
    return False

# 測試多個年份
test_years = [2020, 2021, 1900, 2000, 2024, 2100]

print("=== 閏年判斷結果 ===")
for year in test_years:
    is_leap = check_leap_year(year)
    result = "閏年" if is_leap else "平年"
    print(f"{year} 年：{result}")

# 輸出：
# === 閏年判斷結果 ===
# 2020 年：閏年
# 2021 年：平年
# 1900 年：平年
# 2000 年：閏年
# 2024 年：閏年
# 2100 年：平年
````

### 互動式測試

讓使用者輸入年份，即時判斷：

````python
# 互動式測試（需要手動執行）
print("=== 閏年查詢系統 ===")
year_input = input("請輸入年份：")
year = int(year_input)

if check_leap_year(year):
    print(f"✅ {year} 年是閏年！")
else:
    print(f"❌ {year} 年是平年。")

# 範例輸出（當輸入 2024）：
# === 閏年查詢系統 ===
# 請輸入年份：2024
# ✅ 2024 年是閏年！
````

### 題目：計算母音數量

請仔細閱讀以下程式碼，**先不要執行**，試著猜測輸出結果：

````python
def count_vowels(text):
    """
    計算字串中的母音數量
    母音包括：a, e, i, o, u（不區分大小寫）
    """
    vowels = "aeiouAEIOU"
    count = 0
    
    for char in text:
        if char in vowels:
            count += 1
    
    return count

# 測試
word1 = "Hello"
word2 = "Python"
word3 = "Programming"

print(f'"{word1}" 的母音數量：{count_vowels(word1)}')
print(f'"{word2}" 的母音數量：{count_vowels(word2)}')
print(f'"{word3}" 的母音數量：{count_vowels(word3)}')

# 問題：這三個單字各有幾個母音？
# 先自己想一想，再執行程式碼驗證！
````

### ✅ 正確答案與解析

**正確輸出：**

```
"Hello" 的母音數量：2
"Python" 的母音數量：1
"Programming" 的母音數量：3
```

**詳細解析：**

1. **"Hello"** → 2 個母音
   - 'e' 和 'o' 是母音
   - 'H', 'l', 'l' 是子音

2. **"Python"** → 1 個母音
   - 只有 'o' 是母音
   - 'y' 不算母音（在英文中有時算，但我們的規則只包含 a, e, i, o, u）

3. **"Programming"** → 3 個母音
   - 'o', 'a', 'i' 是母音

**程式碼重點：**

- 使用 `for` 迴圈遍歷字串中的每個字元
- 使用 `if char in vowels` 檢查是否為母音
- 使用 `count += 1` 累計數量
- 使用 `return count` 回傳結果

## 📝 章節重點整理

### 1. def 的用途

- 用來**定義函式**
- 讓程式碼可以**重複使用**
- 提高程式的**可讀性**和**維護性**

### 2. return 的用途

- **回傳結果**給呼叫者
- **中斷函式執行**（遇到 return 就結束）
- 讓函式的結果可以**繼續使用**

### 3. 為什麼要使用函式？

#### ✅ 可重複利用（Reusability）

- 寫一次，到處使用
- 避免重複撰寫相同的程式碼

```python
# 不用函式：重複的程式碼
print(f"Hello, Alice!")
print(f"Hello, Bob!")
print(f"Hello, Charlie!")

# 使用函式：簡潔明瞭
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
greet("Charlie")
```

#### ✅ 提高可讀性（Readability）

- 函式名稱可以**說明功能**
- 讓程式碼更容易理解

```python
# 難以理解
if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("閏年")

# 清楚明瞭
if check_leap_year(year):
    print("閏年")
```

#### ✅ 模組化（Modularity）

- 將大問題**拆解成小問題**
- 每個函式專注處理一件事
- 方便測試和除錯

### 4. 函式設計原則

- **單一職責**：一個函式只做一件事
- **命名清楚**：從函式名稱就能知道功能
- **適當註解**：說明參數和回傳值
- **測試驗證**：確保函式正常運作
