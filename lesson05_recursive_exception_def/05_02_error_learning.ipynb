## 第一部分：錯誤訊息怎麼看？

### 🔍 Traceback 概念

當 Python 遇到錯誤時，會產生一個 **Traceback**（追蹤），告訴你：
- 💬 是什麼錯誤（Error Type）
- 📍 在哪一行（Line Number）
- 🔤 錯誤訊息說什麼（Error Message）

### 📋 讀錯誤訊息的步驟

**例子：假設你寫了這個程式，它出錯了**

```
Traceback (most recent call last):
  File "script.py", line 5, in <module>
    print(name)  # 出錯在這一行
NameError: name 'name' is not defined  # 錯誤類型 + 訊息
```

**重點提取：**
1. **錯誤類型**：`NameError` ← 這告訴你問題是「名稱找不到」
2. **錯誤訊息**：`name 'name' is not defined` ← 具體說是 'name' 這個變數沒定義
3. **出錯位置**：`line 5` ← 看第 5 行
4. **出錯程式碼**：`print(name)` ← 確實使用了未定義的 name

✅ **初學者秘訣**：看不懂的時候，先看「錯誤類型」，再看「出錯位置」，通常就能猜到問題了！

## 第二部分：常見錯誤類型詳解

### (1) SyntaxError（語法錯誤）

**什麼時候出現：** Python 讀不懂你寫的程式碼。通常是標點、括號、冒號等問題。

---

**❌ 範例 1-A：漏掉冒號**

### 學習建議

- 核心先做：先學會讀懂錯誤訊息、辨認 `SyntaxError`、`NameError`、`TypeError`、`IndentationError`。
- 建議延後：太長的錯誤案例分析、連續多層錯誤排查，初學者先能看懂訊息就夠。
- 這一堂的重點不是把所有錯都背起來，而是建立「看到錯誤訊息不要害怕」的習慣。

````python
# 錯誤的程式碼
if 5 > 3
    print('5 大於 3')
````

**💥 錯誤訊息：**
```
  File "...", line 1
    if 5 > 3
           ^
SyntaxError: invalid syntax
```

**🔧 修正方式：** 在 `if` 的條件後面加冒號 `:`

````python
# ✅ 正確的程式碼
if 5 > 3:
    print('5 大於 3')
````

**✅ 正確輸出：**
```
5 大於 3
```

---

**❌ 範例 1-B：括號沒關**

````python
# 錯誤的程式碼
print('Hello World'
````

**💥 錯誤訊息：**
```
  File "...", line 1
    print('Hello World'
                       ^
SyntaxError: '(' was never closed
```

**🔧 修正方式：** 加上結尾的右括號 `)`

````python
# ✅ 正確的程式碼
print('Hello World')
````

**✅ 正確輸出：**
```
Hello World
```

### (2) NameError（名稱錯誤）

**什麼時候出現：** 使用了一個尚未定義的變數名稱，或者打錯變數名。

---

**❌ 範例 2-A：使用尚未宣告的變數**

````python
# 錯誤的程式碼
print(ccClub * 3)  # ccClub 還沒有定義
````

**💥 錯誤訊息：**
```
NameError: name 'ccClub' is not defined
```

**🔧 修正方式：** 先定義變數，再使用

````python
# ✅ 正確的程式碼
ccClub = 'CC'
print(ccClub * 3)
````

**✅ 正確輸出：**
```
CCCCCC
```

---

**❌ 範例 2-B：打錯變數名稱**

````python
# 錯誤的程式碼
universities = ['台大', '台科大', '台北科大']
print(universit)  # 打錯字：少了最後的 'ies'
````

**💥 錯誤訊息：**
```
NameError: name 'universit' is not defined
```

**🔧 修正方式：** 檢查變數名稱的拼字

````python
# ✅ 正確的程式碼
universities = ['台大', '台科大', '台北科大']
print(universities)
````

**✅ 正確輸出：**
```
['台大', '台科大', '台北科大']
```

### (3) TypeError（類型錯誤）

**什麼時候出現：** 對某個資料類型做了它不能做的事。例如字串加整數、列表無法迭代等。

---

**❌ 情境 3-A：字串不可直接修改索引**

````python
# 錯誤的程式碼
slogan = '聯成電腦'
slogan[3] = '不'  # 想修改第 4 個字
````

**💥 錯誤訊息：**
```
TypeError: 'str' object does not support item assignment
```

**💡 解釋：** 字串是「不可變」的（immutable），無法直接改裡面的字元。

**🔧 修正方式：** 重新組合字串

````python
# ✅ 正確的程式碼（方法 1：字串切片 + 串接）
slogan = '聯成電腦'
new_slogan = slogan[:3] + '不' + slogan[4:]
print(new_slogan)
````

**✅ 正確輸出：**
```
聯成電不
```

---

**❌ 情境 3-B：字串加整數串接**

````python
# 錯誤的程式碼
num = 55
result = '55' + num  # 字串和整數無法直接加
````

**💥 錯誤訊息：**
```
TypeError: can only concatenate str (not "int") to str
```

**💡 解釋：** 只能在相同類型間進行 `+` 操作。

**🔧 修正方式：** 轉換成相同類型再串接

````python
# ✅ 正確的程式碼（方法 1：轉成字串）
num = 55
result = '55' + str(num)
print(result)

# ✅ 或者（方法 2：都轉成數字）
num = 55
result = 55 + num
print(result)
````

**✅ 正確輸出：**
```
5555
110
```

---

**❌ 情境 3-C：for 迴圈中 in 後面放整數（不可迭代）**

````python
# 錯誤的程式碼
for i in 3:  # 3 是整數，不能用 for 去迭代
    print(i)
````

**💥 錯誤訊息：**
```
TypeError: 'int' object is not iterable
```

**💡 解釋：** 整數無法被迭代（iteration）。只有字串、列表、range 等可以迭代。

**🔧 修正方式：** 使用 range() 或改用列表

````python
# ✅ 正確的程式碼
for i in range(3):  # 改用 range(3) = [0, 1, 2]
    print(i)
````

**✅ 正確輸出：**
```
0
1
2
```

---

**❌ 情境 3-D：range() 的參數是列表（應該是整數）**

````python
# 錯誤的程式碼
lst = [1, 2, 3]
for i in range(lst):  # range 的參數必須是整數
````

**💥 錯誤訊息：**
```
TypeError: 'list' object cannot be interpreted as an integer
```

**💡 解釋：** `range()` 需要整數，不能放列表。

**🔧 修正方式：** 用 `range(len(lst))` 或直接 `for i in lst`

````python
# ✅ 正確的程式碼（方法 1：使用 len 和 range）
lst = [1, 2, 3]
for i in range(len(lst)):
    print(f'索引 {i}: {lst[i]}')

print('---')

# ✅ 或者（方法 2：直接迭代列表）
for item in lst:
    print(f'元素: {item}')
````

**✅ 正確輸出：**
```
索引 0: 1
索引 1: 2
索引 2: 3
---
元素: 1
元素: 2
元素: 3
```

### (4) ValueError（值錯誤）

**什麼時候出現：** 資料類型正確，但「值」不合法。例如把「python」轉成整數。

---

**❌ 範例 4-A：將非數字字串轉成整數**

````python
# 錯誤的程式碼
age = int('python')  # 'python' 不是數字
````

**💥 錯誤訊息：**
```
ValueError: invalid literal for int() with base 10: 'python'
```

**💡 解釋：** `int()` 只能轉換看起來像數字的字串。

**🔧 修正方式：** 確保輸入是合法的數字字串

````python
# ✅ 正確的程式碼
age = int('20')  # '20' 可以轉成整數
print(f'年齡：{age} 歲')
````

**✅ 正確輸出：**
```
年齡：20 歲
```

---

**❌ 範例 4-B：浮點數轉換錯誤**

````python
# 錯誤的程式碼
score = float('abc')  # 'abc' 無法轉成浮點數
````

**💥 錯誤訊息：**
```
ValueError: could not convert string to float: 'abc'
```

**🔧 修正方式：** 檢查輸入，或使用 `.isdigit()` 驗證

````python
# ✅ 正確的程式碼（方法：先檢查）
user_input = '95.5'
if user_input.replace('.', '', 1).isdigit():  # 檢查是否為數字
    score = float(user_input)
    print(f'分數：{score}')
else:
    print('輸入不是有效的數字')
````

**✅ 正確輸出：**
```
分數：95.5
```

---

### (5) IndentationError（縮排錯誤）

**什麼時候出現：** 程式碼縮排不正確。Python 很在意「空格」！

---

**❌ 情境 5-A：未正確縮排（expected an indented block）**

````python
# 錯誤的程式碼
if 10 > 5:
print('10 大於 5')  # 這行沒有縮排
````

**💥 錯誤訊息：**
```
IndentationError: expected an indented block
```

**💡 解釋：** `if` 或 `for` 後面的程式碼「一定要」縮排（通常 4 個空格）。

**🔧 修正方式：** 加上正確的縮排

````python
# ✅ 正確的程式碼
if 10 > 5:
    print('10 大於 5')  # 這行縮排了 4 個空格
````

**✅ 正確輸出：**
```
10 大於 5
```

---

**❌ 情境 5-B：縮排過多（不規則縮排）**

````python
# 錯誤的程式碼
x = 5
        print(x)  # 這行縮排過多（不符規則）
````

**💥 錯誤訊息：**
```
IndentationError: unexpected indent
```

**🔧 修正方式：** 移除多餘的縮排

````python
# ✅ 正確的程式碼
x = 5
print(x)  # 沒有多餘縮排
````

**✅ 正確輸出：**
```
5
```

### (6) IndexError（索引超出範圍）

**什麼時候出現：** 要存取的索引不存在。例如列表只有 3 個元素，卻要存取第 5 個。

---

**❌ 範例 6-A：字串索引超出範圍**

````python
# 錯誤的程式碼
slogan = '聯成電腦'  # 長度是 4
print(slogan[10])  # 要存取第 11 個？（超出範圍）
````

**💥 錯誤訊息：**
```
IndexError: string index out of range
```

**💡 解釋：** `slogan` 的有效索引是 0, 1, 2, 3，沒有 10。

**🔧 修正方式：** 確認索引在有效範圍內

````python
# ✅ 正確的程式碼
slogan = '聯成電腦'  # 索引：0='聯', 1='成', 2='電', 3='腦'
print(slogan[0])  # 第 1 個字
print(slogan[3])  # 第 4 個字
print(f'長度：{len(slogan)}')
````

**✅ 正確輸出：**
```
聯
腦
長度：4
```

---

**❌ 範例 6-B：for 迴圈中索引超出範圍**

````python
# 錯誤的程式碼
slogan = '聯成電腦'  # 長度是 4
for num in range(len(slogan) + 1):  # 迴圈變成 0, 1, 2, 3, 4
    print(slogan[num])  # 當 num=4 時，超出範圍！
````

**💥 錯誤訊息：**
```
IndexError: string index out of range
```

**💡 解釋：** `range(len(slogan) + 1)` 會產生 [0, 1, 2, 3, 4]，但索引 4 不存在。

**🔧 修正方式：** 不要加 `+ 1`

````python
# ✅ 正確的程式碼
slogan = '聯成電腦'
for num in range(len(slogan)):  # 迴圈變成 0, 1, 2, 3
    print(slogan[num])
````

**✅ 正確輸出：**
```
聯
成
電
腦
```

### (7) ZeroDivisionError（除以零）

**什麼時候出現：** 用某個數字除以 0。

---

**❌ 範例 7-A：直接除以零**

````python
# 錯誤的程式碼
result = 1 / 0
````

**💥 錯誤訊息：**
```
ZeroDivisionError: division by zero
```

**💡 解釋：** 數學上，除以 0 是未定義的。

**🔧 修正方式：** 在執行除法前檢查分母

````python
# ✅ 正確的程式碼
dividend = 1
divisor = 0

if divisor != 0:
    result = dividend / divisor
    print(f'{dividend} / {divisor} = {result}')
else:
    print('分母不能為 0，無法計算')
````

**✅ 正確輸出：**
```
分母不能為 0，無法計算
```

---

**❌ 範例 7-B：計算中出現除以零**

````python
# 錯誤的程式碼
scores = [85, 90, 88]
total = sum(scores)
average = total / (len(scores) - 3)  # (0 - 3) = -3... 等等，len(scores) - 3 = 0
````

**💥 錯誤訊息：**
```
ZeroDivisionError: division by zero
```

**💡 解釋：** `len(scores)` 是 3，3 - 3 = 0，所以會除以 0。

**🔧 修正方式：** 修正分母的計算邏輯

````python
# ✅ 正確的程式碼
scores = [85, 90, 88]
total = sum(scores)
average = total / len(scores)  # 應該是除以個數
print(f'平均分數：{average}')
````

**✅ 正確輸出：**
```
平均分數：87.66666666666667
```

## 第三部分：錯誤類型速查表

| 錯誤類型 | 可能的原因 | 如何修正 |
|---------|----------|--------|
| **SyntaxError** | 冒號漏掉、括號沒關、標點錯誤 | 檢查程式碼標點符號 |
| **NameError** | 變數未定義或拼字錯誤 | 先定義變數，檢查拼字 |
| **TypeError** | 資料類型不相容或操作不適用 | 檢查資料類型，必要時轉換 |
| **ValueError** | 值無法轉換或不合法 | 驗證輸入的值是否合法 |
| **IndentationError** | 縮排不正確或不一致 | 檢查縮排的空格數 |
| **IndexError** | 索引超出範圍 | 確認索引在有效範圍內 |
| **ZeroDivisionError** | 除數為 0 | 在除法前檢查分母 |

**題目 1-2 (簡單)**

```python
name = 'Alice'
age = 25
info = name + age
```

問：執行這段程式會出現什麼錯誤？

**答案 1-2**

**錯誤類型：** `TypeError`

**原因：** 嘗試用 `+` 連接字串和整數，但不同類型無法直接相加。

**修正後的程式碼：**

````python
# 方法 1：都轉成字串
name = 'Alice'
age = 25
info = name + str(age)
print(info)

# 方法 2：使用格式化字符串
info2 = f'{name}{age}'
print(info2)
````


**題目 1-3 (中等)**

```python
numbers = [1, 2, 3]
for i in numbers:
    print(numbers[i])
```

問：執行這段程式會出現什麼錯誤？

**答案 1-3**

**錯誤類型：** `TypeError`

**原因：** 迴圈中的 `i` 會逐次取得列表中的值（1, 2, 3），不是索引（0, 1, 2）。但 `range(numbers)` 是 TypeError，這裡是用值來當索引。實際上當 i=3 時，numbers[3] 會產生 IndexError。

更正：應該是 **IndexError**。原因是當 `i=3` 時，列表的有效索引只有 0, 1, 2。

**修正後的程式碼：**

````python
# 方法 1：用 range(len())
numbers = [1, 2, 3]
for i in range(len(numbers)):
    print(numbers[i])

print('---')

# 方法 2：直接迭代列表
for num in numbers:
    print(num)
````

**✅ 正確輸出：**
```
1
2
3
---
1
2
3
```

**題目 1-4 (中等)**

```python
user_input = input('請輸入你的年齡：')
age = int(user_input)
print(f'明年你 {age + 1} 歲')
```

如果使用者輸入「twenty」（而不是數字），會出現什麼錯誤？

**答案 1-4**

**錯誤類型：** `ValueError`

**原因：** `int('twenty')` 嘗試將字串 'twenty' 轉成整數，但它不是純數字。

**修正後的程式碼：**

````python
# 方法：檢查輸入是否為有效的數字
user_input = '25'  # 示範用

if user_input.isdigit():
    age = int(user_input)
    print(f'明年你 {age + 1} 歲')
else:
    print('請輸入有效的數字')
````

**✅ 正確輸出（當輸入 25）：**
```
明年你 26 歲
```

**題目 1-6 (中等)**

```python
total_score = 450
num_students = 0
average = total_score / num_students
```

問：執行這段程式會出現什麼錯誤？

## 本章重點回顧

### 🎯 核心要點
1. **錯誤是正常的** → 初學者經常出錯，這是成長的一部分
2. **學會讀錯誤訊息** → 這是寫程式最重要的技能之一

### 💡 初學者檢查清單
- 遇到 **SyntaxError** 時 → 檢查冒號、括號、縮排
- 遇到 **NameError** 時 → 檢查變數是否已定義、拼字是否正確
- 遇到 **TypeError** 時 → 檢查資料類型是否相容
- 遇到 **ValueError** 時 → 檢查值是否合法（例如轉換前驗證）
- 遇到 **IndentationError** 時 → 檢查縮排的空格數
- 遇到 **IndexError** 時 → 檢查索引是否超出範圍
- 遇到 **ZeroDivisionError** 時 → 檢查分母是否為 0
