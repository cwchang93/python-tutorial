# Lesson 08｜函式 Function

把重複工作命名，讓程式更清楚、更容易修改。


## 例題 01｜重複程式碼的問題


```python
print('歡迎 Amy')
print('歡迎 Bob')
print('歡迎 Cindy')
```

## 例題 02｜定義並呼叫函式


```python
def say_hello():
    print('Hello, Python!')

say_hello()
say_hello()
```

## 例題 03｜一個參數


```python
def greet(name):
    print('歡迎', name)

greet('Amy')
greet('Bob')
```

## 例題 04｜兩個參數


```python
def show_score(name, score):
    print(name, '的分數：', score)

show_score('Amy', 95)
```

## 例題 05｜位置參數的順序


```python
def introduce(name, city):
    print(name, '住在', city)

introduce('Amy', '台北')
```

## 例題 06｜return 回傳答案


```python
def add(a, b):
    return a + b

result = add(10, 20)
print('總和：', result)
```

## 例題 07｜print 與 return 不同


```python
def show_total(a, b):
    print(a + b)

def get_total(a, b):
    return a + b

answer = get_total(80, 120)
print('再加運費：', answer + 60)
```

## 例題 08｜計算折扣價


```python
def discount_price(price, rate):
    return price * rate

print('折扣後：', discount_price(1000, 0.8))
```

## 例題 09｜回傳布林值


```python
def is_passed(score):
    return score >= 60

print(is_passed(75))
```

## 例題 10｜函式搭配 if


```python
def grade_message(score):
    if score >= 60:
        return '及格'
    return '需要加油'

print(grade_message(58))
```

## 例題 11｜預設參數


```python
def greet(name, message='歡迎'):  
    print(message, name)

greet('Amy')
greet('Bob', '早安')
```

## 例題 12｜關鍵字參數


```python
def order_summary(name, amount):
    print(name, amount)

order_summary(amount=680, name='Amy')
```

## 例題 13｜區域變數


```python
def calculate_total(price, quantity):
    total = price * quantity
    return total

print(calculate_total(80, 3))
# total 只存在函式內
```

## 例題 14｜驗證函式


```python
def is_valid_score(score):
    return 0 <= score <= 100

print(is_valid_score(85))
print(is_valid_score(120))
```

## 例題 15｜處理 List


```python
def average(numbers):
    return sum(numbers) / len(numbers)

print('平均：', average([70, 80, 90]))
```

## 例題 16｜訂單工具組


```python
def subtotal(price, quantity):
    return price * quantity

def shipping_fee(amount):
    return 0 if amount >= 1000 else 80

amount = subtotal(350, 3)
print('商品小計：', amount)
print('運費：', shipping_fee(amount))
print('訂單總額：', amount + shipping_fee(amount))
```

## 例題 17｜回傳多個結果

一次回傳最高、最低與平均。


```python
def score_summary(scores):
    return max(scores), min(scores), sum(scores)/len(scores)

high, low, avg = score_summary([70, 85, 90])
print('最高：', high, '最低：', low, '平均：', avg)
```

## 例題 18｜處理空 List

沒有資料時先回傳 None。


```python
def average(numbers):
    if len(numbers) == 0:
        return None
    return sum(numbers)/len(numbers)

print(average([]))
```

## 例題 19｜會員折扣函式

依會員等級回傳不同折扣。


```python
def member_price(price, level):
    if level == 'gold':
        return price * 0.8
    if level == 'silver':
        return price * 0.9
    return price

print(member_price(1000, 'gold'))
```

## 例題 20｜格式化商品小卡

函式可以回傳整理好的文字。


```python
def product_card(name, price, stock):
    return f'{name}｜價格 {price}｜庫存 {stock}'

print(product_card('咖啡', 120, 8))
```

## 例題 21｜函式呼叫函式

訂單總額可以組合小計與運費函式。


```python
def subtotal(price, quantity):
    return price * quantity

def shipping_fee(amount):
    return 0 if amount >= 1000 else 80

def order_total(price, quantity):
    amount = subtotal(price, quantity)
    return amount + shipping_fee(amount)

print(order_total(350, 3))
```

## 例題 22｜找出及格名單

傳入 List，回傳新的篩選結果。


```python
def passed_scores(scores):
    result = []
    for score in scores:
        if score >= 60:
            result.append(score)
    return result

print(passed_scores([55, 72, 88, 40]))
```

## 例題 23｜計算 Dict 總額

函式也能處理多筆訂單。


```python
def revenue(orders):
    total = 0
    for order in orders:
        total += order['amount']
    return total

orders = [{'amount':680},{'amount':320}]
print(revenue(orders))
```

## 例題 24｜範圍驗證函式

把常用判斷整理成布林值函式。


```python
def in_range(value, minimum, maximum):
    return minimum <= value <= maximum

print(in_range(85, 0, 100))
```

## 例題 25｜完整成績報表

組合多個小函式產生摘要。


```python
def average(scores):
    return sum(scores)/len(scores)

def passed_count(scores):
    count = 0
    for score in scores:
        if score >= 60:
            count += 1
    return count

def report(scores):
    return {'最高分':max(scores),'平均':average(scores),'及格人數':passed_count(scores)}

print(report([55, 72, 88, 95]))
```
