## 解答 11｜⭐⭐ 最高分

不使用 max()，回傳最高分。


```python
def highest(scores):
    best = scores[0]
    for score in scores:
        if score > best:
            best = score
    return best

print(highest([88,65,100,72]))
```

## 解答 12｜⭐⭐ 最低分

不使用 min()，回傳最低分。


```python
def lowest(scores):
    low = scores[0]
    for score in scores:
        if score < low:
            low = score
    return low

print(lowest([88,65,100,72]))
```

## 解答 13｜⭐⭐ 及格人數

## 解答 15｜⭐⭐ 空資料平均

```python
def average(numbers):
    if not numbers:
        return None
    return sum(numbers)/len(numbers)

print(average([]))
```

## 解答 17｜⭐⭐ 會員折扣

依 gold、silver 或一般會員回傳價格。


```python
def member_price(price, level):
    if level == 'gold':
        return price * 0.8
    if level == 'silver':
        return price * 0.9
    return price

print(member_price(1000,'gold'))
```

## 解答 18｜⭐⭐ 及格名單

## 解答 19｜⭐⭐ 訂單營收

接收 List 裡的 Dict，回傳總營收。


```python
def revenue(orders):
    total = 0
    for order in orders:
        total += order['amount']
    return total

print(revenue([{'amount':680},{'amount':320}]))
```

## 解答 21｜⭐⭐⭐ 訂單工具組

拆成小計、運費、總額三個函式。


```python
def subtotal(price, quantity):
    return price * quantity

def shipping_fee(amount):
    return 0 if amount >= 1000 else 80

def order_total(price, quantity):
    amount = subtotal(price, quantity)
    return amount + shipping_fee(amount)

print(order_total(350,3))
```

## 解答 22｜⭐⭐⭐ 分數摘要

回傳最高、最低、平均與及格人數。


```python
def score_summary(scores):
    passed = 0
    for score in scores:
        if score >= 60:
            passed += 1
    return {'最高':max(scores),'最低':min(scores),'平均':sum(scores)/len(scores),'及格':passed}

print(score_summary([55,72,88,95]))
```

## 解答 23｜⭐⭐⭐ BMI 工具

建立 BMI 計算與分類函式。


```python
def bmi(weight, height):
    return weight / height ** 2

def bmi_level(value):
    if value < 18.5:
        return '過輕'
    if value < 24:
        return '正常'
    return '過重'

value = bmi(60,1.7)
print(value,bmi_level(value))
```

## 解答 24｜⭐⭐⭐ 購物車工具

接收價格 List，回傳小計、運費與總額。


```python
def cart_summary(prices):
    amount = sum(prices)
    fee = 0 if amount >= 1000 else 80
    return {'小計':amount,'運費':fee,'總額':amount+fee}

print(cart_summary([350,420,300]))
```

## 解答 25｜⭐⭐⭐ 成績報表

組合多個函式輸出完整摘要。


```python
def average(scores):
    return sum(scores)/len(scores)

def passed_count(scores):
    count = 0
    for score in scores:
        if score >= 60:
            count += 1
    return count

def report(scores):
    return {'最高分':max(scores),'最低分':min(scores),'平均':average(scores),'及格人數':passed_count(scores)}

print(report([55,72,88,95]))
```
