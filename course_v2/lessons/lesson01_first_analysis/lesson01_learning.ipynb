## 例題 05｜生活計算機

Python 可以直接處理加、減、乘、除。

```python
print('早餐總價：', 45 + 35)
print('找零：', 100 - 80)
print('三杯飲料：', 65 * 3)
print('四人平分：', 600 / 4)
```

## 例題 08｜更新盒子裡的內容

同一個變數可以換成新資料。

```python
mood = '有點緊張'
print('上課前：', mood)
mood = '開始有信心'
print('執行成功後：', mood)
```

## 例題 09｜飲料訂單

只修改數量，就能重複使用計算步驟。

```python
drink = '紅茶'
price = 35
quantity = 2
total = price * quantity
print(drink, quantity, '杯，共', total, '元')
```

## 例題 10｜午餐分帳

觀察 / 的計算結果。

```python
bill = 840
people = 4
per_person = bill / people
print('每人支付：', per_person, '元')
```

## 例題 11｜通勤時間

把分鐘拆成去程與回程，算出一天總時間。

```python
one_way_minutes = 35
daily_minutes = one_way_minutes * 2
print('一天通勤', daily_minutes, '分鐘')
```

## 例題 12｜購物找零

總價先保存，再用付款金額減掉它。

```python
item_price = 79
quantity = 3
paid = 500
total = item_price * quantity
change = paid - total
print('總價：', total)
print('找零：', change)
```

## 例題 16｜商品收據

將品名、單價、數量與總價組合成易讀結果。

```python
product = '筆記本'
unit_price = 50
quantity = 3
print('=== 購物收據 ===')
print('商品：', product)
print('數量：', quantity)
print('總價：', unit_price * quantity)
```
