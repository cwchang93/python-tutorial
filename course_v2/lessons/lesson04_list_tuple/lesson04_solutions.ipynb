## 練習 09｜⭐⭐ 安全刪除

只有商品存在時才 remove，否則顯示找不到。


```python
cart = ['筆記本', '鉛筆']
target = '橡皮擦'
if target in cart:
    cart.remove(target)
else:
    print('找不到')
```

## 練習 12｜⭐⭐ 電影收藏

新增兩部、刪除一部，最後排序並印出收藏數。


```python
movies = ['可可夜總會', '全面啟動']
movies.extend(['腦筋急轉彎', '寄生上流'])
movies.remove('全面啟動')
movies.sort()
print(movies, len(movies))
```

## 練習 13｜⭐⭐⭐ 候位管理器

叫號一位、新增一位、插入 VIP，最後顯示下一位。


```python
queue = ['Amy', 'Bob', 'Cindy']
print('叫號：', queue.pop(0))
queue.append('David')
queue.insert(0, 'VIP')
print('下一位：', queue[0], queue)
```

## 練習 14｜⭐⭐⭐ 成績修正器

若分數超過 100 改成 100；低於 0 改成 0。先處理一筆指定位置。


```python
scores = [88, 105, -3, 72]
index = 1
if scores[index] > 100:
    scores[index] = 100
elif scores[index] < 0:
    scores[index] = 0
print(scores)
```
