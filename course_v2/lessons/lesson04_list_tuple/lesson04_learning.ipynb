# lesson04｜List 深入與 Tuple 入門

先預測、再執行，最後換資料測試。


## 例題 01｜一筆資料與多筆資料

資料變多時，用 List 一起管理。


```python
name = 'Amy'
names = ['Amy', 'Bob', 'Cindy']
print(name)
print(names)
```

## 例題 02｜建立 List

中括號內可放多筆資料，並保留順序。


```python
courses = ['Python', 'Excel', 'SQL']
print(courses)
print(len(courses))
```

## 例題 03｜Index 從 0 開始

第一筆是 0，最後一筆可以用 -1。


```python
courses = ['Python', 'Excel', 'SQL']
print(courses[0])
print(courses[1])
print(courses[-1])
```

## 例題 04｜餐廳候位名單

排隊有順序，非常適合 List。


```python
waiting = ['Amy', 'Bob', 'Cindy']
print('下一位：', waiting[0])
print('最後一位：', waiting[-1])
```

## 例題 05｜修改指定位置

List 可以修改原本內容。


```python
drinks = ['紅茶', '綠茶', '奶茶']
drinks[1] = '烏龍茶'
print(drinks)
```

## 例題 06｜append 加到最後

購物清單臨時多一項商品。


```python
shopping = ['牛奶', '麵包']
shopping.append('雞蛋')
print(shopping)
```

## 例題 07｜insert 插入位置

將最重要的事情放到最前面。


```python
tasks = ['回覆 Email', '整理報表']
tasks.insert(0, '參加早會')
print(tasks)
```

## 例題 08｜remove 依內容刪除

知道內容、不知道位置時使用 remove。


```python
cart = ['鉛筆', '筆記本', '橡皮擦']
cart.remove('鉛筆')
print(cart)
```

## 例題 09｜pop 依位置取出

pop 會刪除並回傳那一筆。


```python
queue = ['Amy', 'Bob', 'Cindy']
served = queue.pop(0)
print(served)
print(queue)
```

## 例題 10｜in 檢查是否存在

先檢查再刪除，可以避免錯誤。


```python
menu = ['紅茶', '綠茶', '奶茶']
if '奶茶' in menu:
    print('今天有奶茶')
```

## 例題 11｜切片取得一部分

起點包含、終點不包含。


```python
scores = [60, 70, 80, 90, 100]
print(scores[1:4])
print(scores[:3])
print(scores[-2:])
```

## 例題 12｜切片的步長與反轉

第三個數字是 step，可以跳著取，也可以用負數反轉。


```python
days = ['一', '二', '三', '四', '五', '六', '日']
print(days[::2])
print(days[::-1])
```

## 例題 13｜List 的 + 與 *

List 可以串接，也可以重複內容。


```python
morning = ['早餐', '通勤']
afternoon = ['上課', '複習']
print(morning + afternoon)
print(['提醒'] * 3)
```

## 例題 14｜Mutable、alias 與 copy

List 是 mutable。使用 copy() 或 [:] 才會建立新的 List；直接指定只會共用同一份資料。


```python
shopping = ['牛奶', '麵包']
alias = shopping
copied = shopping.copy()
alias.append('雞蛋')
print(shopping)  # alias 與原本共用資料
print(copied)   # copy 是獨立的新 List
```

## 例題 15｜Immutable Tuple

Tuple 可以索引、切片與查詢，但不能 append、remove 或修改元素。


```python
point = (25.0330, 121.5654)
print(point[0])
print(point[:1])
# point[0] = 0  # 取消註解會出現 TypeError
```

## 例題 16｜排序

sorted 產生新 List；sort 修改原 List。


```python
scores = [88, 65, 100, 72]
new_scores = sorted(scores)
print(scores)
print(new_scores)
```

## 例題 17｜找最大、最小與總和

Python 內建函式可以快速回答問題。


```python
scores = [88, 65, 100, 72]
print(max(scores))
print(min(scores))
print(sum(scores))
```

## 例題 18｜List 小作品：旅行清單

綜合新增、插入、修改與刪除。


```python
places = ['台南', '高雄']
places.append('花蓮')
places.insert(0, '台北')
places[2] = '屏東'
print(places)
```

## 例題 19｜Tuple：固定資料包

Tuple 有順序，但建立後不能修改。


```python
gps = (25.0330, 121.5654)
print(gps[0])
print(gps[1])
```

## 例題 20｜Tuple 解包

一次把固定資料拆到多個變數。


```python
brand_blue = (20, 80, 160)
red, green, blue = brand_blue
print(red, green, blue)
```

## 例題 21｜List 還是 Tuple

會改用 List；不希望改用 Tuple。


```python
shopping = ['牛奶', '麵包']
meeting_point = (25.0330, 121.5654)
shopping.append('雞蛋')
print(shopping, meeting_point)
```

## 例題 18｜小作品：本週任務板

建立、修改並排序自己的任務。


```python
tasks = ['完成作業', '複習 List', '整理筆記']
tasks.append('預習 Tuple')
tasks.remove('整理筆記')
print('任務數：', len(tasks))
print(tasks)
```
