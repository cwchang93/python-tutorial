# Lesson 02-2｜List、Tuple、Dict、Set

從一個值，走向能依照需求整理多筆資料的容器。

本教材共有 **20 個課堂例題**。每題先猜結果，再執行，最後改成自己的版本。

## 例題 01｜一個值與多筆資料

比較單一變數與 List；資料變多時，需要容器協助整理。

```python
name = 'Amy'
names = ['Amy', 'Bob', 'Cindy']
print(name)
print(names)
```

## 例題 02｜建立課程 List

List 使用中括號，內容依照放入順序排列。

```python
courses = ['Python', 'Excel', 'SQL']
print(courses)
print('共有', len(courses), '門課')
```

## 例題 03｜用 Index 取得資料

Index 從 0 開始；-1 可以取得最後一筆。

```python
courses = ['Python', 'Excel', 'SQL']
print('第一門：', courses[0])
print('第二門：', courses[1])
print('最後一門：', courses[-1])
```

## 例題 04｜餐廳候位名單

排隊有先後順序，很適合使用 List。先猜 Bob 的 Index。

```python
waiting = ['Amy', 'Bob', 'Cindy']
print('第一位：', waiting[0])
print('Bob 的 Index：', waiting.index('Bob'))
print('最後一位：', waiting[-1])
```

## 例題 05｜修改 List 內容

List 可以修改；用 Index 指定要換掉哪一筆。

```python
drinks = ['紅茶', '綠茶', '奶茶']
print('修改前：', drinks)
drinks[1] = '烏龍茶'
print('修改後：', drinks)
```

## 例題 06｜append 加到最後

購物清單臨時多一項商品，可以用 append。

```python
shopping = ['牛奶', '麵包']
shopping.append('雞蛋')
shopping.append('水果')
print(shopping)
```

## 例題 07｜insert 插入指定位置

最優先的待辦事項可以插到第 0 格。

```python
tasks = ['回覆 Email', '整理報表']
tasks.insert(0, '參加早會')
print(tasks)
```

## 例題 08｜remove 依內容刪除

知道要刪除的內容時使用 remove。

```python
cart = ['鉛筆', '筆記本', '橡皮擦']
cart.remove('鉛筆')
print(cart)
```

## 例題 09｜pop 依位置取出

pop 會刪除並回傳資料；不填 Index 時取出最後一筆。

```python
queue = ['Amy', 'Bob', 'Cindy']
served = queue.pop(0)
print('已服務：', served)
print('仍在等候：', queue)
```

## 例題 10｜List 小作品：旅行清單

綜合建立、新增、插入、修改與刪除。

```python
travel_list = ['台南', '高雄']
travel_list.append('花蓮')
travel_list.insert(0, '台北')
travel_list[2] = '屏東'
travel_list.remove('台南')
print('我的旅行清單：', travel_list)
```

## 例題 11｜建立學生 Dict

Dict 使用大括號，透過 Key 對應 Value。

```python
student = {
    'name': 'Amy',
    'age': 20,
    'score': 95
}
print(student)
```

## 例題 12｜手機聯絡人

查聯絡人時不用記住第幾格，直接使用欄位名稱。

```python
contact = {
    'name': '媽媽',
    'phone': '0912-345-678'
}
print(contact['name'])
print(contact['phone'])
```

## 例題 13｜讀取與更新 Dict

有相同 Key 時會更新原本的 Value。

```python
student = {'name': 'Amy', 'score': 95}
print('原分數：', student['score'])
student['score'] = 100
print('新分數：', student['score'])
```

## 例題 14｜新增 Dict 欄位

指定不存在的 Key，就會新增一個欄位。

```python
student = {'name': 'Amy', 'score': 95}
student['city'] = 'Taipei'
student['is_student'] = True
print(student)
```

## 例題 15｜用 get 避免 KeyError

get 找不到 Key 時不會讓程式停止，還能提供預設文字。

```python
student = {'name': 'Amy', 'score': 95}
print(student.get('email'))
print(student.get('email', '尚未提供'))
# print(student['email'])  # 取消註解會出現 KeyError
```

## 例題 16｜Dict 小作品：商品小卡

用有意義的欄位名稱管理商品資料。

```python
product = {'name': 'Notebook', 'price': 120, 'stock': 8}
product['stock'] = product['stock'] - 1
product['category'] = 'Stationery'
print('商品：', product['name'])
print('售價：', product['price'])
print('剩餘庫存：', product['stock'])
print('分類：', product['category'])
```

## 例題 17｜Tuple 保存固定座標

Tuple 使用小括號，適合建立後不應修改的資料。

```python
gps = (25.0330, 121.5654)
print('緯度：', gps[0])
print('經度：', gps[1])
# gps[0] = 24.0  # Tuple 不能修改
```

## 例題 18｜Tuple 保存品牌顏色

RGB 是固定的三個數值，可以使用 Tuple 表示。

```python
brand_blue = (20, 80, 160)
red, green, blue = brand_blue
print('R：', red)
print('G：', green)
print('B：', blue)
```

## 例題 19｜Set 移除重複報名

同一個 Email 報名兩次，轉成 Set 後只保留一份。

```python
emails = ['amy@test.com', 'bob@test.com', 'amy@test.com']
unique_emails = set(emails)
unique_emails.add('cindy@test.com')
print(unique_emails)
print('不重複人數：', len(unique_emails))
```

## 例題 20｜綜合小作品：學習儀表板

使用 List、Dict、Tuple 與 Set 整理一位學生的學習資料。

```python
tasks = ['完成 Lesson 01', '練習 List', '複習 Dict']
profile = {'name': 'Amy', 'hours': 4, 'goal': '完成資料分析作品'}
class_time = (9, 30)
skills = {'print', '變數', 'List', 'Dict', 'List'}
print('=== 學習儀表板 ===')
print('學員：', profile['name'])
print('本週任務：', tasks)
print('上課時間：', class_time)
print('已學技能：', skills)
print('技能數量：', len(skills))
print('最終目標：', profile['goal'])
```
