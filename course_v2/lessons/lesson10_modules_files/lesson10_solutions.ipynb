## 例題 07｜使用 w 建立文字檔

用 with open 與 w 模式建立 practice_note.txt，寫入兩行內容。


```python
from pathlib import Path

path = Path('practice_note.txt')
with path.open('w', encoding='utf-8') as file:
    file.write('第一行\n')
    file.write('第二行\n')

print('已建立：', path)
```

## 例題 08｜使用 a 追加紀錄

在 practice_note.txt 結尾追加「第三行」，不可清空前兩行。


```python
with path.open('a', encoding='utf-8') as file:
    file.write('第三行\n')

print(path.read_text(encoding='utf-8'))
```

## 例題 09｜比較三種讀取方法

分別用 read、readline、readlines 讀取 practice_note.txt，清楚標示三種結果。


```python
with path.open('r', encoding='utf-8') as file:
    print('read：', repr(file.read()))

with path.open('r', encoding='utf-8') as file:
    print('readline：', repr(file.readline()))

with path.open('r', encoding='utf-8') as file:
    print('readlines：', file.readlines())
```

## 例題 10｜逐行清理換行

直接使用 for 逐行讀取 practice_note.txt，並用 strip() 移除換行。


```python
with path.open('r', encoding='utf-8') as file:
    for line in file:
        print(line.strip())
```

## 例題 11｜三種寫入方式

在同一個檔案中各使用一次 write、writelines、print(file=)。


```python
from pathlib import Path

path = Path('write_methods.txt')
with path.open('w', encoding='utf-8') as file:
    file.write('第一行\n')
    file.writelines(['第二行\n', '第三行\n'])
    print('第四行', file=file)

print(path.read_text(encoding='utf-8'))
```

## 例題 12｜CSV 小計


```python
import csv
total = 0
with open('products.csv', encoding='utf-8') as file:
    next(file)
    for item, qty, price in csv.reader(file):
        total += int(qty) * int(price)
print('總額：', total)
```

## 例題 13｜DictWriter


```python
import csv
rows = [{'name': 'Amy', 'level': 'A'}, {'name': 'Bob', 'level': 'B'}]
with open('members.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=['name', 'level'])
    writer.writeheader(); writer.writerows(rows)
```

## 例題 14｜DictReader


```python
import csv
with open('members.csv', encoding='utf-8') as file:
    for row in csv.DictReader(file):
        print(row['name'], row['level'])
```

## 例題 15｜CSV 篩選


```python
import csv
with open('sales.csv', encoding='utf-8') as file:
    for row in csv.DictReader(file):
        if float(row['amount']) > 1000:
            print(row)
```

## 例題 21｜CSV 平均


```python
import csv
with open('scores.csv', encoding='utf-8') as file:
    scores = [int(row['score']) for row in csv.DictReader(file)]
print('平均：', sum(scores) / len(scores))
print('最高分：', max(scores))
```

## 例題 22｜random 模擬


```python
import random
random.seed(4)
rolls = [random.randint(1, 6) for _ in range(10)]
print(rolls)
print({n: rolls.count(n) for n in range(1, 7)})
```

## 例題 25｜綜合銷售報表


```python
import csv
rows = [['month', 'sales'], ['Jan', '1200'], ['Feb', '900']]
with open('report.csv', 'w', newline='', encoding='utf-8') as file:
    csv.writer(file).writerows(rows)
with open('report.csv', encoding='utf-8') as file:
    data = list(csv.DictReader(file))
print('總銷售額：', sum(int(row['sales']) for row in data))
```
