## 例題 13｜逐行讀取


```python
with open('log.txt', 'w', encoding='utf-8') as file:
    file.write('A\nB\n')
with open('log.txt', encoding='utf-8') as file:
    for line in file:
        print(line.strip())
```

## 例題 16｜CSV 的概念


```python
import csv
rows = [['name', 'score'], ['Amy', 90]]
with open('scores.csv', 'w', newline='', encoding='utf-8') as file:
    csv.writer(file).writerows(rows)
```

## 例題 17｜csv.writer 寫入資料


```python
import csv
with open('sales.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['item', 'amount'])
    writer.writerow(['coffee', 120])
```

## 例題 18｜csv.reader 讀取資料


```python
import csv
with open('sales.csv', encoding='utf-8') as file:
    for row in csv.reader(file):
        print(row)
```

## 例題 19｜DictWriter 寫入欄位


```python
import csv
rows = [{'name': 'Amy', 'score': 90}]
with open('scores.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=['name', 'score'])
    writer.writeheader()
    writer.writerows(rows)
```

## 例題 20｜DictReader 以欄位讀取


```python
import csv
with open('scores.csv', encoding='utf-8') as file:
    for row in csv.DictReader(file):
        print(row['name'], row['score'])
```

## 例題 21｜清理 CSV 數字


```python
import csv
with open('scores.csv', encoding='utf-8') as file:
    scores = [int(row['score']) for row in csv.DictReader(file)]
print('平均：', sum(scores) / len(scores))
```

## 例題 24｜模組與檔案整合


```python
from pathlib import Path
import csv
path = Path('report.csv')
with path.open('w', newline='', encoding='utf-8') as file:
    csv.writer(file).writerows([['month', 'sales'], ['Jan', 1200]])
print(path.exists())
```

## 例題 25｜用函式整理檔案流程


```python
from pathlib import Path

def save_note(path, text):
    Path(path).write_text(text, encoding='utf-8')

save_note('note.txt', '模組與檔案可以組合使用')
print(Path('note.txt').read_text(encoding='utf-8'))
```
