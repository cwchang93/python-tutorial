## 例題 07｜建立自訂模組

一個 `.py` 檔案就是模組。將可重複使用的函式放在另一個檔案，再由主程式匯入。


```python
from pathlib import Path
import importlib

module_path = Path('lesson10_tools.py')
module_path.write_text(
    "def add(a, b):\n"
    "    return a + b\n",
    encoding='utf-8'
)

import lesson10_tools
importlib.reload(lesson10_tools)
print('加總：', lesson10_tools.add(3, 5))
```

## 例題 10｜檔案模式 r、w、a

`r` 讀取、`w` 重新寫入、`a` 保留舊內容並追加。


```python
file_modes = {
    'r': '讀取；檔案不存在會報錯',
    'w': '重新寫入；舊內容會被清空',
    'a': '追加內容；保留原本資料',
}

for mode, meaning in file_modes.items():
    print(mode, '：', meaning)
```

## 例題 11｜with open 自動關閉

離開 with 區塊後，Python 會自動關閉檔案。


```python
from pathlib import Path

note_path = Path('lesson10_note.txt')
with note_path.open('w', encoding='utf-8') as file:
    file.write('今天開始學檔案處理\n')

print('已寫入：', note_path)
```

## 例題 12｜read、readline、readlines

三種方法的回傳結果不同；重新開啟檔案，避免讀取游標已在結尾。


```python
with note_path.open('r', encoding='utf-8') as file:
    content = file.read()
print('read：', repr(content))

with note_path.open('r', encoding='utf-8') as file:
    first_line = file.readline()
print('readline：', repr(first_line))

with note_path.open('r', encoding='utf-8') as file:
    lines = file.readlines()
print('readlines：', lines)
```

## 例題 13｜使用 for 逐行讀取

大型檔案通常直接逐行處理，不需要先把所有內容放進 List。


```python
with note_path.open('r', encoding='utf-8') as file:
    for line in file:
        clean_line = line.strip()
        print('一行內容：', clean_line)
```

## 例題 14｜write、writelines、print(file=)

write 與 writelines 不會自動換行；print 寫入時會自動加入換行。


```python
write_path = Path('lesson10_write_demo.txt')
with write_path.open('w', encoding='utf-8') as file:
    file.write('第一行\n')
    file.writelines(['第二行\n', '第三行\n'])
    print('第四行', file=file)

print(write_path.read_text(encoding='utf-8'))
```

## 例題 15｜使用 a 模式追加文字

`a` 会從檔案結尾新增內容，不會清空原本資料。


```python
with write_path.open('a', encoding='utf-8') as file:
    file.write('第五行：追加內容\n')

print(write_path.read_text(encoding='utf-8'))
```

## 例題 17｜csv.writer 寫入資料

## 例題 18｜csv.reader 讀取課程資料

課程已提供 `../../data/lesson10_sales.csv`，不需要另外搜尋檔案。


```python
import csv
from pathlib import Path
data_path = Path('../../data/lesson10_sales.csv')
with data_path.open(encoding='utf-8') as file:
    for row in csv.reader(file):
        print(row)
```

## 例題 19｜DictWriter 寫入欄位


```python
import csv
rows = [{'name': 'Amy', 'score': 90}]
with open('scores.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=['name', 'score'])
    writer.writeheader()
    writer.writerows(rows)
```

## 例題 20｜DictReader 以欄位讀取

`DictReader` 會依照 CSV 標題列，讓我們用欄位名稱取值。


```python
import csv
from pathlib import Path
data_path = Path('../../data/lesson10_sales.csv')
with data_path.open(encoding='utf-8') as file:
    for row in csv.DictReader(file):
        print(row['date'], row['product'], row['quantity'])
```

## 例題 21｜清理 CSV 數字


```python
import csv
with open('scores.csv', encoding='utf-8') as file:
    scores = [int(row['score']) for row in csv.DictReader(file)]
print('平均：', sum(scores) / len(scores))
```

## 例題 25｜用函式整理檔案流程


```python
from pathlib import Path

def save_note(path, text):
    Path(path).write_text(text, encoding='utf-8')

save_note('note.txt', '模組與檔案可以組合使用')
print(Path('note.txt').read_text(encoding='utf-8'))
```
