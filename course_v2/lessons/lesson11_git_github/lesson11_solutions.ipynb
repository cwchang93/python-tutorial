## 解答 11｜修改程式

## 解答 14｜建立 README

寫下專案用途與執行方式。


```python
from pathlib import Path
readme = '# 我的第一個 Git 專案\n\n執行方式：`python hello.py`\n'
Path('practice_git_project/README.md').write_text(readme, encoding='utf-8')
print(readme)
```

## 解答 15｜忽略暫存檔

建立 .gitignore，避免提交 cache 與暫存檔。


```python
from pathlib import Path
Path('practice_git_project/.gitignore').write_text('__pycache__/\n*.tmp\n', encoding='utf-8')
print('忽略規則已建立')
```

## 解答 16｜一次檢查所有狀態

確認 README 與 .gitignore 尚未提交。


```python
!git -C practice_git_project status --short
```

## 解答 17｜提交文件

將 README 與 .gitignore 放入同一個文件 Commit。


```python
!git -C practice_git_project add README.md .gitignore
!git -C practice_git_project commit -m "docs: add project instructions"
```

## 解答 18｜理解 GitHub Remote

列出目前遠端；尚未設定時不會顯示內容。


```python
!git -C practice_git_project remote -v
```

## 解答 19｜設定遠端示範

請把網址換成自己的 GitHub Repository；本格只示範指令，不直接執行。


```python
remote_command = 'git remote add origin https://github.com/USERNAME/REPOSITORY.git'
print(remote_command)
```

## 解答 20｜Push 指令示範

第一次 Push 通常要指定 origin 與 main。


```python
push_command = 'git push -u origin main'
print(push_command)
```

## 解答 21｜Clone 指令示範

Clone 是把遠端 Repository 完整複製到本機。


```python
clone_command = 'git clone https://github.com/USERNAME/REPOSITORY.git'
print(clone_command)
```

## 解答 22｜Pull 指令示範

Pull 是把遠端的新版本同步回目前資料夾。


```python
pull_command = 'git pull origin main'
print(pull_command)
```

## 解答 23｜比較 Push、Pull、Clone

用 Dict 整理三個指令的用途。


```python
commands = {'push': '本機 → GitHub', 'pull': 'GitHub → 既有本機專案', 'clone': '第一次完整下載'}
for name, meaning in commands.items():
    print(name, '：', meaning)
```

## 解答 24｜安全操作檢查表

輸出操作前的三項檢查。


```python
checks = ['確認所在資料夾', '先看 git status', 'Commit 訊息說明改了什麼']
for index, item in enumerate(checks, 1):
    print(index, item)
```
