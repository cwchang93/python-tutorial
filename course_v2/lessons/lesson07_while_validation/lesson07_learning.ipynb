## 例題 02｜第一個 while


```python
count = 1
while count <= 3:
    print('第', count, '次')
    count += 1
```

## 例題 03｜倒數計時


```python
number = 5
while number > 0:
    print(number)
    number -= 1
print('開始！')
```

## 例題 04｜累加到指定數字


```python
number = 1
total = 0
while number <= 5:
    total += number
    number += 1
print('總和：', total)
```

## 例題 05｜break 提早結束


```python
while True:
    command = input('輸入 q 離開：')
    if command == 'q':
        break
    print('你輸入：', command)
```

## 例題 06｜continue 重新輸入


```python
while True:
    text = input('不可空白：')
    if text == '':
        print('請重新輸入')
        continue
    print('收到：', text)
    break
```

## 例題 07｜登入重試三次


```python
password = 'python123'
attempts = 0
while attempts < 3:
    answer = input('密碼：')
    if answer == password:
        print('登入成功')
        break
    attempts += 1
    print('密碼錯誤，剩下', 3 - attempts, '次')
else:
    print('登入失敗')
```

## 例題 08｜數字輸入驗證


```python
while True:
    try:
        age = int(input('年齡：'))
        break
    except ValueError:
        print('請輸入整數')
print('你的年齡：', age)
```

## 例題 09｜限制合理範圍


```python
while True:
    try:
        score = int(input('分數 0 到 100：'))
        if 0 <= score <= 100:
            break
        print('分數超出範圍')
    except ValueError:
        print('請輸入整數')
print('有效分數：', score)
```

## 例題 10｜是或否


```python
while True:
    answer = input('是否繼續？(y/n)：').lower()
    if answer in ['y', 'n']:
        break
    print('只能輸入 y 或 n')
print('你的選擇：', answer)
```

## 例題 11｜選單操作


```python
while True:
    print('1. 查詢  2. 儲值  0. 離開')
    choice = input('請選擇：')
    if choice == '1':
        print('目前餘額：500')
    elif choice == '2':
        print('進入儲值功能')
    elif choice == '0':
        print('再見')
        break
    else:
        print('沒有這個選項')
```

## 例題 12｜收集多筆資料


```python
names = []
while True:
    name = input('姓名（輸入 end 結束）：')
    if name == 'end':
        break
    names.append(name)
print('名單：', names)
```

## 例題 13｜ATM 提款驗證


```python
balance = 1000
while True:
    try:
        amount = int(input('提款金額：'))
        if amount <= 0:
            print('金額必須大於 0')
        elif amount > balance:
            print('餘額不足')
        else:
            balance -= amount
            print('提款成功，餘額：', balance)
            break
    except ValueError:
        print('請輸入整數金額')
```

## 例題 14｜訂票數量驗證


```python
available = 6
while True:
    try:
        tickets = int(input('購買張數：'))
        if 1 <= tickets <= available:
            break
        print('可購買 1 到', available, '張')
    except ValueError:
        print('請輸入整數')
print('訂票成功：', tickets, '張')
```

## 例題 15｜猜數字

## 例題 17｜購物預算累加

持續輸入支出，輸入 0 後顯示總額。


```python
total = 0
while True:
    amount = int(input('支出（0 結束）：'))
    if amount == 0:
        break
    total += amount
print('支出總額：', total)
```

## 例題 18｜庫存扣除

每次售出一件，直到庫存為 0。


```python
stock = 3
while stock > 0:
    print('售出 1 件')
    stock -= 1
print('商品售完')
```

## 例題 19｜驗證商品數量

數量必須介於 1 到庫存之間。


```python
stock = 8
while True:
    try:
        quantity = int(input('數量：'))
        if 1 <= quantity <= stock:
            break
        print('可購買 1 到', stock, '件')
    except ValueError:
        print('請輸入整數')
print('購買數量：', quantity)
```

## 例題 20｜累加到超過目標

逐筆累加，總額達 1000 後停止。


```python
amounts = [120, 350, 280, 400]
index = 0
total = 0
while index < len(amounts) and total < 1000:
    total += amounts[index]
    index += 1
print('累計：', total)
```

## 例題 21｜簡易點餐系統

選擇餐點並累加總價，輸入 0 結帳。


```python
prices = {'1': 80, '2': 120}
total = 0
while True:
    choice = input('1 飯糰／2 便當／0 結帳：')
    if choice == '0':
        break
    if choice in prices:
        total += prices[choice]
    else:
        print('沒有這個選項')
print('總價：', total)
```

## 例題 22｜刪除無效資料

遇到負數就跳過，只累加有效金額。


```python
values = [120, -1, 80, -1, 200]
index = 0
total = 0
while index < len(values):
    value = values[index]
    index += 1
    if value < 0:
        continue
    total += value
print('有效總額：', total)
```

## 例題 23｜問卷輸入器

持續收集 1 到 5 分，輸入 0 結束。


```python
ratings = []
while True:
    rating = int(input('評分 1～5（0 結束）：'))
    if rating == 0:
        break
    if 1 <= rating <= 5:
        ratings.append(rating)
    else:
        print('評分超出範圍')
print('有效評分：', ratings)
```

## 例題 24｜最大值輸入器

持續輸入數字，輸入 end 後顯示最大值。


```python
numbers = []
while True:
    text = input('數字（end 結束）：')
    if text == 'end':
        break
    try:
        numbers.append(float(text))
    except ValueError:
        print('格式錯誤')
print('最大值：', max(numbers) if numbers else '沒有資料')
```

## 例題 25｜完整結帳流程

驗證數量、會員選項並輸出結帳結果。


```python
price = 120
while True:
    try:
        quantity = int(input('數量：'))
        if quantity > 0:
            break
        print('數量必須大於 0')
    except ValueError:
        print('請輸入整數')
while True:
    member = input('會員？(y/n)：').lower()
    if member in ['y', 'n']:
        break
total = price * quantity
if member == 'y':
    total *= 0.9
print('應付金額：', total)
```
