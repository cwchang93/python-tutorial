## 解答 01｜⭐ 倒數

使用 while 印出 5、4、3、2、1。


```python
number = 5
while number >= 1:
    print(number)
    number -= 1
```

```python
number = 2
while number <= 10:
    print(number)
    number += 2
```

## 解答 03｜⭐ 累加 1 到 100

使用 while 計算 1 到 100 的總和。


```python
number = 1
total = 0
while number <= 100:
    total += number
    number += 1
print('總和：', total)
```

## 解答 14｜⭐⭐ 跳過負數

逐筆處理 List，負數不加入總和。


```python
values = [120,-1,80,-5,200]
index = 0
total = 0
while index < len(values):
    value = values[index]
    index += 1
    if value < 0:
        continue
    total += value
print('總和：', total)
```

## 解答 15｜⭐⭐ 累加到目標

金額總和達到 1000 時停止。


```python
amounts = [120,350,280,400]
index = 0
total = 0
while index < len(amounts) and total < 1000:
    total += amounts[index]
    index += 1
print(total)
```

## 解答 17｜⭐⭐ 最大值輸入器

輸入 end 後顯示最大值。


```python
numbers = []
while True:
    text = input('數字：')
    if text == 'end':
        break
    try:
        numbers.append(float(text))
    except ValueError:
        print('格式錯誤')
print('最大值：', max(numbers) if numbers else '沒有資料')
```

## 解答 21｜⭐⭐⭐ ATM 選單

完成查詢、存款、提款與離開。


```python
balance = 1000
while True:
    choice = input('1 查詢／2 存款／3 提款／0 離開：')
    if choice == '1':
        print('餘額：', balance)
    elif choice == '2':
        balance += int(input('存款：'))
    elif choice == '3':
        amount = int(input('提款：'))
        if 0 < amount <= balance:
            balance -= amount
        else:
            print('金額錯誤')
    elif choice == '0':
        break
    else:
        print('無效選項')
```

## 解答 22｜⭐⭐⭐ 點餐系統

重複點餐並累加總價。


```python
prices = {'1':80,'2':120}
total = 0
while True:
    choice = input('1 飯糰／2 便當／0 結帳：')
    if choice == '0':
        break
    if choice in prices:
        total += prices[choice]
    else:
        print('無效選項')
print('總價：', total)
```

## 解答 23｜⭐⭐⭐ 登入與鎖定

連續錯誤三次顯示帳號鎖定。


```python
account = 'amy'
password = '1234'
attempts = 0
while attempts < 3:
    user = input('帳號：')
    pwd = input('密碼：')
    if user == account and pwd == password:
        print('登入成功')
        break
    attempts += 1
else:
    print('帳號鎖定')
```

## 解答 24｜⭐⭐⭐ 成績輸入器

輸入 end 後顯示總數、最高分與平均。


```python
scores = []
while True:
    text = input('分數（end 結束）：')
    if text == 'end':
        break
    try:
        score = int(text)
        if 0 <= score <= 100:
            scores.append(score)
    except ValueError:
        print('格式錯誤')
if scores:
    print('總數：', len(scores))
    print('最高分：', max(scores))
    print('平均：', sum(scores)/len(scores))
```

## 解答 25｜⭐⭐⭐ 完整結帳

驗證數量與會員選項，計算折扣後金額。


```python
price = 120
while True:
    try:
        quantity = int(input('數量：'))
        if quantity > 0:
            break
    except ValueError:
        print('請輸入整數')
while True:
    member = input('會員？(y/n)：').lower()
    if member in ['y','n']:
        break
total = price * quantity
if member == 'y':
    total *= 0.9
print('應付金額：', total)
```
