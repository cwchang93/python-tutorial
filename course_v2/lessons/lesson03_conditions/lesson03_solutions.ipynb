## 練習 09｜⭐⭐ 有幾個數字相等

三個整數全部相同輸出 3；只有兩個相同輸出 2；都不同輸出 0。


```python
a = 7
b = 7
c = 3
if a == b == c:
    print(3)
elif a == b or a == c or b == c:
    print(2)
else:
    print(0)
```

## 練習 11｜⭐⭐ 閏年判斷

年份可被 400 整除，或可被 4 整除但不可被 100 整除，就是閏年。


```python
year = 2024
if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print('閏年')
else:
    print('平年')
```

## 練習 12｜⭐⭐ 飲料店會員折扣

會員且滿 150 元打九折，否則原價。


```python
price = 65
quantity = 3
is_member = True
total = price * quantity
if total >= 150 and is_member:
    final_total = total * 0.9
else:
    final_total = total
print('應付：', final_total)
```

## 練習 13｜⭐⭐⭐ 三數排序

輸入三個互不相同的數字，由小到大印出。可以交換變數內容。


```python
a = 9
b = 2
c = 6
if a > b:
    a, b = b, a
if b > c:
    b, c = c, b
if a > b:
    a, b = b, a
print(a, b, c)
```

## 練習 14｜⭐⭐⭐ 巧克力能不能剛好折下來

一塊 n×m 的巧克力，只沿格線折一次，判斷能否剛好取得 k 格。k 必須小於整塊。


```python
n = 4
m = 6
k = 8
if 0 < k < n * m and (k % n == 0 or k % m == 0):
    print('可以')
else:
    print('不可以')
```

## 練習 15｜⭐⭐⭐ 自動售票機

依年齡計算票價：未滿 6 歲免費、6–11 歲 150 元、12–64 歲 300 元、65 歲以上 150 元。


```python
age = 10
if age < 6:
    price = 0
elif age < 12:
    price = 150
elif age < 65:
    price = 300
else:
    price = 150
print('票價：', price)
```
