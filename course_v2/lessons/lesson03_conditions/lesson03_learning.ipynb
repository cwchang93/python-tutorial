# Lesson 03｜if、elif、else 條件判斷

先預測、再執行、最後修改條件。


## 例題 01｜比較會得到 bool

先預測每一題會得到 True 還是 False。


```python
print(20 >= 18)
print(20 == 18)
print(20 != 18)
```

## 例題 02｜生活中的真假狀態

布林值就像開關，只有兩種狀態。


```python
is_raining = True
is_logged_in = False
print(is_raining, is_logged_in)
```

## 例題 03｜第一個 if

條件成立時，才會執行縮排區塊。


```python
score = 80
if score >= 60:
    print('及格')
print('檢查完成')
```

## 例題 04｜if 不成立時

把溫度改成 25，再觀察哪一行沒有執行。


```python
temperature = 32
if temperature >= 30:
    print('今天很熱')
print('記得補充水分')
```

## 例題 05｜成年判斷

年齡是生活中很直觀的邊界條件。


```python
age = 20
if age >= 18:
    print('可以觀看')
```

## 例題 06｜if / else 兩條路

兩個分支只會執行其中一個。


```python
age = 16
if age >= 18:
    print('成年')
else:
    print('未成年')
```

## 例題 07｜密碼登入

字串也可以使用 == 比較。


```python
password = 'python123'
if password == 'python123':
    print('登入成功')
else:
    print('密碼錯誤')
```

## 例題 08｜奇數或偶數

餘數為 0 代表可以被 2 整除。


```python
number = 7
if number % 2 == 0:
    print('偶數')
else:
    print('奇數')
```

## 例題 09｜elif 多種結果

Python 從上往下，只執行第一個符合的分支。


```python
score = 85
if score >= 90:
    print('A')
elif score >= 80:
    print('B')
else:
    print('C')
```

## 例題 10｜條件順序陷阱

先放寬鬆條件，會讓後面的嚴格條件失去機會。


```python
score = 95
if score >= 60:
    print('及格')
elif score >= 90:
    print('優秀')
```

## 例題 11｜正確的成績順序

由嚴格到寬鬆排列。


```python
score = 95
if score >= 90:
    print('優秀')
elif score >= 60:
    print('及格')
else:
    print('再加油')
```

## 例題 12｜邊界值

18 歲是否包含在成年規則中？


```python
for_age = 18
print(for_age >= 18)
print(for_age > 18)
```

## 例題 13｜and：而且

兩個條件都成立，才能進場。


```python
age = 20
has_ticket = True
if age >= 18 and has_ticket:
    print('可以入場')
else:
    print('不能入場')
```

## 例題 14｜or：或者

學生或會員，符合任一條件即可。


```python
is_student = True
is_member = False
if is_student or is_member:
    print('享有優惠')
```

## 例題 15｜not：反轉狀態

尚未登入時顯示提醒。


```python
is_logged_in = False
if not is_logged_in:
    print('請先登入')
```

## 例題 16｜範圍判斷

and 也能判斷數字是否位於範圍內。


```python
temperature = 25
if 20 <= temperature <= 28:
    print('體感舒適')
else:
    print('注意溫度變化')
```

## 例題 18｜消費免運判斷

把門檻改成不同數字看看。


```python
amount = 680
if amount >= 500:
    print('符合免運')
else:
    print('還差', 500 - amount, '元')
```

## 例題 19｜會員折扣

整合金額與會員兩個條件。


```python
amount = 600
is_member = True
if amount >= 500 and is_member:
    print('會員九折：', amount * 0.9)
else:
    print('原價：', amount)
```

## 例題 20｜小作品：飲料店結帳

修改杯數、會員狀態與優惠門檻，做出自己的版本。


```python
price = 65
quantity = 3
is_member = True
total = price * quantity
if total >= 150 and is_member:
    final_total = total * 0.9
    message = '會員滿額九折'
else:
    final_total = total
    message = '原價計算'
print('原價：', total)
print('優惠：', message)
print('應付：', final_total)
```
