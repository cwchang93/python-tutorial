## 例題 01｜print 顯示多個內容

逗號適合把文字與資料一起輸出。

```python
name = 'Amy'
score = 95
print('姓名：', name, '分數：', score)
```

## 例題 02｜sep 自訂分隔符號

sep 可以控制多個內容之間放什麼。

```python
print('2026', '07', '19', sep='-')
print('Python', 'Excel', 'SQL', sep=' / ')
```

## 例題 03｜end 控制換行

end 可以讓下一個 print 接在同一行。

```python
print('Hello', end=' ')
print('Python')
```

## 例題 04｜變數讓程式可重用

只換資料，不必重寫顯示邏輯。

```python
product_name = 'Coffee'
unit_price = 120
quantity = 3
print(product_name, '小計：', unit_price * quantity)
```

## 例題 05｜int：數幾個

年齡、人數、件數與庫存通常是沒有小數點的整數。

```python
age = 25
students = 18
stock = 120
print(age, students, stock)
print(type(age))
```

## 例題 06｜float：精確測量

身高、重量、價格與折扣常需要小數。

```python
height = 168.5
price = 99.9
discount = 0.8
print(height, price, discount)
print(type(height))
```

## 例題 07｜int 與 float 一起計算

運算中出現 float 時，結果通常也會是 float；除法 `/` 也通常得到 float。

```python
price = 80
tax_rate = 0.05
total = price * (1 + tax_rate)
print(total)
print(10 / 2)
print(type(total))
```

## 例題 08｜str：不只是一般文字

電話、學號與郵遞區號雖然有數字，但不拿來計算，所以仍是字串。

```python
name = 'Amy'
phone = '0912-345-678'
student_id = '00123'
print(name, phone, student_id)
print(type(phone))
```

## 例題 09｜str 的常見操作

字串可以串接、重複，也能用 len() 計算長度。

```python
first_name = 'Amy'
last_name = 'Chen'
print(first_name + ' ' + last_name)
print('Python! ' * 3)
print(len(first_name))
```

## 例題 10｜bool：生活中的是與否

付款、登入、有沒有票，都可以用 True 或 False 表示。

```python
is_paid = True
is_logged_in = False
has_ticket = True
print(is_paid, is_logged_in, has_ticket)
print(type(is_paid))
```

## 例題 11｜比較運算產生 bool

問 Python 一個是非題，它會回答 True 或 False。

```python
age = 20
print(age >= 18)
print(age == 20)
print(age != 20)
```

## 例題 12｜用 type 偵查資料

不確定資料型態時，直接請 Python 告訴你。

```python
age = 20
price = 99.5
name = 'Amy'
is_paid = True
print(type(age), type(price), type(name), type(is_paid))
```

## 例題 13｜同樣外觀，不同型態

數字會相加，字串會接在一起。

```python
print(10 + 5)
print('10' + '5')
```

## 例題 14｜數字轉成 str

要用 + 串接文字時，可以先用 str() 把數字轉成字串。

```python
age = 20
message = '我今年 ' + str(age) + ' 歲'
print(message)
print(type(str(age)))
```

## 例題 15｜字串轉成數字

網頁或 input 取得的數字常需用 int() 或 float() 轉型。

```python
quantity = int('3')
rate = float('0.8')
print(quantity + 2)
print(500 * rate)
```

## 例題 19｜除錯：input 和數字

input 是 str，必須先轉成 int 才能加 1。

```python
age_text = '20'  # 模擬 input 得到的文字
# print(age_text + 1)  # TypeError
age = int(age_text)
print('明年', age + 1, '歲')
```

## 例題 20｜小作品：活動報名確認單

綜合輸出、變數、型態與計算，改成自己的活動版本。

```python
event = 'Python 入門工作坊'
name = 'Amy'
ticket_price = 500
quantity = 2
is_paid = True
total = ticket_price * quantity
print('=== 報名確認單 ===')
print('活動：', event)
print('參加者：', name)
print('票數：', quantity)
print('總金額：', total)
print('已付款：', is_paid)
```
