## 25｜完整查詢流程

請依照說明完成：回答：台北或新北、營業額至少 150 元的訂單，依營業額排序。


```python
from pathlib import Path
import pandas as pd

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]
data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break
if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
city_condition = df["city"].isin(["台北", "新北"] )
revenue_condition = df["revenue"] >= 150
result = df.loc[city_condition & revenue_condition, ["order_id", "product", "city", "revenue"]]
result = result.sort_values("revenue", ascending=False)
print(result)
```
