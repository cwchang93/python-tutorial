## 04｜loc：用標籤或名稱找資料

loc 是 label-based indexing。看到 loc，就先看 index 標籤與欄位名稱。


```python
import pandas as pd

df_people = pd.DataFrame(
    {
        "name": ["JW", "Amy", "Bob"],
        "age": [30, 25, 28],
    },
    index=["a", "b", "c"],
)

print(df_people)
print("標籤 a 的整列：")
print(df_people.loc["a"])
print("標籤 a、name 欄：", df_people.loc["a", "name"])
```

## 05｜iloc：用整數位置找資料

iloc 是 integer location。列與欄都從位置 0 開始計算。


```python
import pandas as pd

df_people = pd.DataFrame(
    {
        "name": ["JW", "Amy", "Bob"],
        "age": [30, 25, 28],
    },
    index=["a", "b", "c"],
)

print(df_people)
print("第 0 列：")
print(df_people.iloc[0])
print("第 0 列、第 1 欄：", df_people.iloc[0, 1])
```

## 06｜loc 與 iloc 的切片終點

loc 的標籤切片包含終點；iloc 和 Python slicing 一樣，不包含結尾位置。數字 index 也要先分清楚標籤與位置。


```python
import pandas as pd

df_people = pd.DataFrame(
    {
        "name": ["JW", "Amy", "Bob"],
        "age": [30, 25, 28],
    },
    index=["a", "b", "c"],
)

loc_result = df_people.loc["a":"b"]
iloc_result = df_people.iloc[0:2]

print("loc 從 a 到 b，包含 b：")
print(loc_result)
print("iloc 從位置 0 到 2，不包含位置 2：")
print(iloc_result)

number_index_df = df_people.copy()
number_index_df.index = [10, 20, 30]

print("loc[10]：找標籤 10")
print(number_index_df.loc[10])
print("iloc[0]：找第 0 個位置")
print(number_index_df.iloc[0])
```

## 12｜排除指定值

```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
result = df[df["city"] != "台北"]
print(result[["order_id", "city"]].head())
```

## 13｜同時符合兩個條件：&

每個條件都要加括號，& 代表而且。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
condition = (df["city"] == "台北") & (df["quantity"] >= 2)
print(df[condition])
```

## 14｜符合其中一個條件：|

| 代表或者，只要一邊是 True 就留下。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
condition = (df["city"] == "台北") | (df["city"] == "高雄")
print(df[condition].head())
```

## 15｜反轉條件：~

~ 會把 True 與 False 對調。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
condition = ~(df["category"] == "飲品")
print(df[condition])
```

## 26｜串接選取、篩選與排序

每一步存成清楚變數，比一行寫完容易除錯。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
selected = df[["order_id", "product", "city", "revenue"]]
filtered = selected[selected["revenue"] >= 200]
result = filtered.sort_values("revenue", ascending=False)
print(result)
```

## 27｜排序後取前幾名

先 sort_values()，再 head() 取得 Top N。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
top5 = df.sort_values("revenue", ascending=False).head(5)
print(top5[["order_id", "product", "revenue"]])
```

## 28｜完整查詢流程

回答：台北或新北、營業額至少 150 元的訂單，依營業額排序。


```python
from pathlib import Path
import pandas as pd

data_path = Path("lesson13_orders.csv")
df = pd.read_csv(data_path)
df["revenue"] = df["quantity"] * df["unit_price"]
city_condition = df["city"].isin(["台北", "新北"] )
revenue_condition = df["revenue"] >= 150
result = df.loc[city_condition & revenue_condition, ["order_id", "product", "city", "revenue"]]
result = result.sort_values("revenue", ascending=False)
print(result)
```
