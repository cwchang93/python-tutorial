# Lesson 09｜錯誤處理

先讀懂錯誤，再用 try、except、else、finally 保護程式。


## 例題 01｜安全轉換年齡


```python
try:
    age = int('20')
except ValueError:
    print('年齡格式錯誤')
```

## 例題 02｜安全計算平均


```python
total, people = 100, 0
if people == 0:
    print('沒有資料')
else:
    print(total / people)
```

## 例題 03｜清單索引防呆


```python
items = ['a', 'b']
try:
    index = int('3')
    print(items[index])
except ValueError:
    print('索引格式錯誤')
except IndexError:
    print('索引超出範圍')
```

## 例題 04｜字典查詢防呆


```python
members = {'Amy': 'amy@example.com'}
print(members.get('Bob', '查無會員'))
```

## 例題 05｜分數範圍驗證


```python
try:
    score = int('101')
    if not 0 <= score <= 100:
        raise ValueError
except ValueError:
    print('分數錯誤')
```

## 例題 06｜輸入折扣


```python
try:
    discount = float('x')
except ValueError:
    discount = 0
print(discount)
```

## 例題 07｜try else


```python
try:
    quantity = int('2')
except ValueError:
    print('數量格式錯誤')
else:
    print('小計：', quantity * 50)
```

## 例題 08｜finally 提示


```python
try:
    print('開始工作')
finally:
    print('流程結束')
```

## 例題 09｜負數數量


```python
def set_quantity(q):
    if q < 0:
        raise ValueError('數量不能為負')
    return q
```

## 例題 10｜捕捉自訂錯誤


```python
try:
    set_quantity(-1)
except ValueError as error:
    print(error)
```

## 例題 11｜多筆分數


```python
values = ['80', 'x', '95']
scores = []
for value in values:
    try:
        scores.append(int(value))
    except ValueError:
        continue
print(scores)
```

## 例題 12｜錯誤統計


```python
success = failed = 0
for value in ['1', 'x', '2']:
    try:
        int(value); success += 1
    except ValueError:
        failed += 1
print(success, failed)
```

## 例題 13｜訂單編號


```python
orders = {'A001': 300}
order_id = 'A002'
try:
    print(orders[order_id])
except KeyError:
    print('查無訂單')
```

## 例題 14｜檔案不存在


```python
try:
    open('missing.txt', encoding='utf-8')
except FileNotFoundError:
    print('檔案不存在')
```

## 例題 15｜多種例外


```python
try:
    result = int('x') / 0
except ValueError:
    print('格式錯誤')
except ZeroDivisionError:
    print('不能除以 0')
```

## 例題 16｜安全平均函式


```python
def average(values):
    if not values:
        return None
    return sum(values) / len(values)
print(average([]))
```

## 例題 17｜安全價格函式


```python
def parse_price(text):
    try:
        price = float(text)
        if price < 0:
            raise ValueError
        return price
    except ValueError:
        return None
```

## 例題 18｜錯誤訊息紀錄


```python
errors = []
for value in ['1', 'x']:
    try:
        int(value)
    except ValueError:
        errors.append(value)
print(errors)
```

## 例題 19｜輸入三次


```python
invalid = []
for text in ['20', 'abc', '30']:
    try:
        int(text)
    except ValueError:
        invalid.append(text)
print('無效：', invalid)
```

## 例題 20｜購物車小計


```python
def line_total(price_text, qty_text):
    try:
        price, qty = float(price_text), int(qty_text)
        return price * qty
    except (ValueError, TypeError):
        return None
print(line_total('50', '2'))
```

## 例題 21｜成績報表


```python
scores, errors = [], 0
for text in ['90', 'x', '70']:
    try:
        scores.append(int(text))
    except ValueError:
        errors += 1
print('最高分：', max(scores))
print('最低分：', min(scores))
print('錯誤筆數：', errors)
```

## 例題 22｜安全除法


```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return None
print(safe_divide(10, 0))
```

## 例題 23｜錯誤分類


```python
try:
    value = int('abc')
except ValueError:
    print('格式錯誤')
else:
    if not 0 <= value <= 100:
        print('範圍錯誤')
```

## 例題 24｜資料驗證流程


```python
valid = []
for text in ['10', '120', '30']:
    try:
        value = int(text)
        if not 0 <= value <= 100:
            raise ValueError
        valid.append(value)
    except ValueError:
        pass
print(valid)
```

## 例題 25｜綜合訂單檢查


```python
orders = [('咖啡', '50', '2'), ('蛋糕', 'x', '1')]
total = 0
for name, price, qty in orders:
    try:
        total += float(price) * int(qty)
    except ValueError:
        print('略過：', name)
print('有效訂單總額：', total)
```
