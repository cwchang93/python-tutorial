# Lesson 09｜錯誤處理

先讀懂錯誤，再用 try、except、else、finally 保護程式。


## 例題 01｜錯誤訊息是線索


```python
print('先看最後一行錯誤類型')
print('再回頭找發生錯誤的程式碼')
```

## 例題 02｜SyntaxError 語法錯誤


```python
# print('少一個右括號'
print('修正括號後就能執行')
```

## 例題 03｜NameError 名稱不存在


```python
user_name = 'Amy'
print(user_name)
```

## 例題 04｜TypeError 型別不相容


```python
price = 100
quantity = 2
print(price * quantity)
```

## 例題 05｜ValueError 內容無法轉換


```python
text = '42'
number = int(text)
print(number + 8)
```

## 例題 06｜IndexError 索引超出範圍


```python
colors = ['紅', '藍']
print(colors[0])
print(len(colors))
```

## 例題 07｜KeyError 字典 key 不存在


```python
profile = {'name': 'Amy'}
print(profile.get('email', '未提供 email'))
```

## 例題 08｜ZeroDivisionError 除以零


```python
total = 100
people = 0
if people != 0:
    print(total / people)
else:
    print('人數不能是 0')
```

## 例題 09｜try 保護可能失敗的程式


```python
try:
    age = int('20')
    print(age)
except ValueError:
    print('年齡格式錯誤')
```

## 例題 10｜except 捕捉指定錯誤


```python
try:
    score = int('abc')
except ValueError:
    print('請輸入整數分數')
```

## 例題 11｜多種錯誤分開處理


```python
try:
    numbers = [10, 20]
    index = int('1')
    print(numbers[index])
except ValueError:
    print('索引必須是整數')
except IndexError:
    print('索引超出範圍')
```

## 例題 12｜except Exception 作為最後防線


```python
try:
    result = 10 / 2
except Exception as error:
    print('發生錯誤：', error)
else:
    print('結果：', result)
```

## 例題 13｜else 只在成功時執行


```python
try:
    number = int('25')
except ValueError:
    print('轉換失敗')
else:
    print('轉換成功：', number)
```

## 例題 14｜finally 一定會執行


```python
try:
    print('開始處理')
finally:
    print('清理或結束提示')
```

## 例題 15｜錯誤時提供預設值


```python
try:
    discount = float('abc')
except ValueError:
    discount = 0
print('折扣：', discount)
```

## 例題 16｜安全輸入整數


```python
def read_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(read_int('18'))
print(read_int('十八'))
```

## 例題 17｜驗證分數範圍


```python
def read_score(text):
    try:
        score = int(text)
        if not 0 <= score <= 100:
            return None
        return score
    except ValueError:
        return None

print(read_score('88'))
```

## 例題 18｜raise 主動提出錯誤


```python
def set_quantity(quantity):
    if quantity < 0:
        raise ValueError('數量不能是負數')
    return quantity

print(set_quantity(3))
```

## 例題 19｜捕捉自己提出的錯誤


```python
try:
    raise ValueError('資料不完整')
except ValueError as error:
    print('請修正：', error)
```

## 例題 20｜清楚顯示錯誤原因


```python
try:
    amount = int('100')
    print(1000 / amount)
except (ValueError, ZeroDivisionError) as error:
    print('輸入或計算錯誤：', error)
```

## 例題 21｜清單逐筆轉換並略過錯誤


```python
raw_scores = ['90', 'abc', '75']
scores = []
for raw in raw_scores:
    try:
        scores.append(int(raw))
    except ValueError:
        print('略過：', raw)
print(scores)
```

## 例題 22｜保護字典查詢


```python
orders = {'A001': 120}
order_id = 'A002'
try:
    print(orders[order_id])
except KeyError:
    print('找不到訂單：', order_id)
```

## 例題 23｜保護檔案開啟概念


```python
try:
    file = open('not_found.txt', encoding='utf-8')
except FileNotFoundError:
    print('檔案不存在')
```

## 例題 24｜錯誤紀錄清單


```python
errors = []
for value in ['10', 'x', '20']:
    try:
        int(value)
    except ValueError:
        errors.append(value)
print('錯誤資料：', errors)
```

## 例題 25｜完整的資料輸入流程


```python
def parse_price(text):
    try:
        price = float(text)
        if price < 0:
            raise ValueError('價格不能小於 0')
        return price
    except ValueError as error:
        print('價格錯誤：', error)
        return None

print(parse_price('199.5'))
```
