## 例題 01｜錯誤訊息是線索

先觸發並捕捉一個錯誤，觀察錯誤類型與訊息。實際 traceback 最後一行通常最重要。


```python
try:
    int('abc')
except ValueError as error:
    print('錯誤類型：', type(error).__name__)
    print('錯誤訊息：', error)
    print('閱讀順序：先看錯誤類型，再回頭找出錯位置')
```

## 例題 04｜TypeError 型別不相容

數字與文字不能直接相加；先觀察 TypeError，再把文字轉成數字。

### 第一步：先執行下一格，觀察真正的錯誤訊息


```python
price = 100
quantity = '2'
print(price + quantity)
```

### 第二步：修正或安全處理

將文字轉成整數後，才能與其他數字進行加法。


```python
price = 100
quantity_text = '2'
quantity = int(quantity_text)
print(price + quantity)
```

## 例題 05｜ValueError 內容無法轉換

資料型態可以轉換，不代表內容一定符合格式。

### 第一步：先執行下一格，觀察真正的錯誤訊息


```python
text = '四十二'
number = int(text)
print(number + 8)
```

### 第二步：修正或安全處理

確認文字內容是可轉換的數字，或在不確定資料時使用 try/except。


```python
text = '四十二'

try:
    number = int(text)
    print(number + 8)
except ValueError as error:
    print('內容無法轉成整數：', error)
```

## 例題 06｜IndexError 索引超出範圍

兩個元素的 List 只有索引 0 與 1，讀取索引 2 會出錯。

### 第一步：先執行下一格，觀察真正的錯誤訊息


```python
colors = ['紅', '藍']
print(colors[2])
```

### 第二步：修正或安全處理

兩個元素只有索引 0 與 1。可先檢查長度，或捕捉 IndexError。


```python
colors = ['紅', '藍']

try:
    print(colors[2])
except IndexError:
    print('索引超出範圍，目前長度：', len(colors))
```

## 例題 07｜KeyError 字典 key 不存在

使用不存在的 key 會觸發 KeyError；不確定欄位是否存在時可使用 get()。

### 第一步：先執行下一格，觀察真正的錯誤訊息


```python
profile = {'name': 'Amy'}
print(profile['email'])
```

### 第二步：修正或安全處理

不確定 key 是否存在時，可使用 get() 提供預設值。


```python
profile = {'name': 'Amy'}
email = profile.get('email', '未提供 email')
print(email)
```

## 例題 08｜ZeroDivisionError 除以零

先實際觸發除以零，再示範用條件判斷預防。

### 第一步：先執行下一格，觀察真正的錯誤訊息


```python
total = 100
people = 0
print(total / people)
```

### 第二步：修正或安全處理

可以先用條件阻止除以零；也可以在輸入不可預期時捕捉 ZeroDivisionError。


```python
total = 100
people = 0

try:
    print(total / people)
except ZeroDivisionError:
    print('人數不能是 0')
```

## 例題 09｜try 保護可能失敗的程式

把可能失敗的轉換放進 try；錯誤發生時由 except 接手。


```python
try:
    age = int('二十')
    print('年齡：', age)
except ValueError as error:
    print('年齡格式錯誤：', error)
```

## 例題 10｜except 捕捉指定錯誤

只捕捉預期的 ValueError，並保留清楚的使用者提示。


```python
try:
    score = int('abc')
except ValueError as error:
    print('請輸入整數分數')
    print('技術訊息：', error)
```

## 例題 11｜多種錯誤分開處理

同一段流程可能遇到格式錯誤或索引錯誤，應分開處理並提供不同提示。


```python
numbers = [10, 20]

for raw_index in ['abc', '5', '1']:
    print('測試輸入：', raw_index)
    try:
        index = int(raw_index)
        print('取得數字：', numbers[index])
    except ValueError:
        print('索引必須是整數')
    except IndexError:
        print('索引超出範圍')
```

## 例題 12｜except Exception 作為最後防線

Exception 可以作為最後防線，但平常仍應優先捕捉具體錯誤。


```python
for divisor in [2, 0]:
    try:
        result = 10 / divisor
    except Exception as error:
        print('最後防線捕捉：', type(error).__name__, error)
    else:
        print('成功結果：', result)
```

## 例題 13｜else 只在成功時執行

try 沒有發生錯誤時才會執行 else。


```python
for text in ['25', '二十五']:
    try:
        number = int(text)
    except ValueError:
        print(text, '轉換失敗')
    else:
        print(text, '轉換成功：', number)
```

## 例題 14｜finally 一定會執行

不論成功或失敗，finally 都會執行，適合放清理與結束工作。


```python
for divisor in [2, 0]:
    print('測試除數：', divisor)
    try:
        print('結果：', 10 / divisor)
    except ZeroDivisionError:
        print('不能除以 0')
    finally:
        print('finally：本次處理結束')
```

## 例題 16｜安全輸入整數


```python
def read_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(read_int('18'))
print(read_int('十八'))
```

## 例題 17｜驗證分數範圍

同時測試有效分數、超出範圍與格式錯誤。


```python
def read_score(text):
    try:
        score = int(text)
        if score < 0 or score > 100:
            return None
        return score
    except ValueError:
        return None

for text in ['88', '120', '八十八']:
    print(text, '→', read_score(text))
```

## 例題 18｜raise 主動提出錯誤

當資料不符合商業規則時，可以用 raise 主動提出清楚的錯誤。


```python
def set_quantity(quantity):
    if quantity < 0:
        raise ValueError('數量不能是負數')
    return quantity

for quantity in [3, -1]:
    try:
        print('設定數量：', set_quantity(quantity))
    except ValueError as error:
        print('設定失敗：', error)
```

## 例題 19｜捕捉自己提出的錯誤


```python
try:
    raise ValueError('資料不完整')
except ValueError as error:
    print('請修正：', error)
```

## 例題 20｜清楚顯示錯誤原因

測試成功、格式錯誤與除以零三種情況。


```python
for text in ['100', 'abc', '0']:
    print('測試輸入：', text)
    try:
        amount = int(text)
        print('計算結果：', 1000 / amount)
    except ValueError as error:
        print('格式錯誤：', error)
    except ZeroDivisionError as error:
        print('計算錯誤：', error)
```

## 例題 23｜保護檔案開啟概念

使用 with open 確保檔案成功開啟時也會自動關閉。


```python
try:
    with open('not_found.txt', encoding='utf-8') as file:
        content = file.read()
        print(content)
except FileNotFoundError as error:
    print('檔案不存在：', error.filename)
```

## 例題 25｜完整的資料輸入流程

整合格式轉換、範圍規則、raise 與清楚錯誤訊息。


```python
def parse_price(text):
    try:
        price = float(text)
        if price < 0:
            raise ValueError('價格不能小於 0')
        return price
    except ValueError as error:
        print(text, '價格錯誤：', error)
        return None

for text in ['199.5', '-20', '一百元']:
    result = parse_price(text)
    print(text, '→', result)
```
