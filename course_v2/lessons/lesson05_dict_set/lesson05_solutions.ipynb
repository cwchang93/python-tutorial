## 練習 05｜⭐ 商品庫存

賣出一件後庫存減 1；有庫存才處理。


```python
product = {'name': '筆記本', 'stock': 3}
if product['stock'] > 0:
    product['stock'] -= 1
print(product)
```

## 練習 08｜⭐⭐ 安全刪除欄位

如果 coupon 存在就 pop，否則顯示沒有優惠券。


```python
order = {'id': 1001, 'total': 680}
if 'coupon' in order:
    print(order.pop('coupon'))
else:
    print('沒有優惠券')
```

## 練習 09｜⭐⭐ 技能集合

## 練習 10｜⭐⭐ 共同技能

找出兩組共同技能與全部技能。


```python
a = {'Python', 'Excel'}
b = {'Python', 'SQL'}
print(a & b)
print(a | b)
```

## 練習 11｜⭐⭐ 會員折扣

Dict 保存會員資料，會員且消費滿 500 打九折。


```python
member = {'name': 'Amy', 'is_member': True}
amount = 600
if member['is_member'] and amount >= 500:
    amount *= 0.9
print(amount)
```

## 練習 12｜⭐⭐ 多位學生

取得 Bob 的成績並判斷是否及格。


```python
students = [{'name': 'Amy', 'score': 95}, {'name': 'Bob', 'score': 58}]
score = students[1]['score']
print(score)
if score >= 60:
    print('及格')
else:
    print('再加油')
```

## 練習 13｜⭐⭐⭐ 商品小卡

建立商品 Dict、更新庫存、加入分類，最後格式化輸出。


```python
product = {'name': 'Notebook', 'price': 120, 'stock': 8}
product['stock'] -= 1
product['category'] = 'Stationery'
print(product['name'], product['price'], product['stock'], product['category'])
```

## 練習 14｜⭐⭐⭐ 報名整理器

用 Set 去除重複報名，再用 Dict 保存活動名稱與人數。


```python
emails = ['amy@test.com', 'bob@test.com', 'amy@test.com']
unique = set(emails)
event = {'name': 'Python 工作坊', 'people': len(unique)}
print(event)
```

## 練習 15｜⭐⭐⭐ 個人學習儀表板

至少使用一個 Dict、一個 Set、一個 List 與一個 if。


```python
profile = {'name': 'Amy', 'hours': 4}
skills = {'Python', 'List', 'Dict'}
tasks = ['完成作業', '複習']
if profile['hours'] >= 3:
    print(profile['name'], skills, tasks)
```
