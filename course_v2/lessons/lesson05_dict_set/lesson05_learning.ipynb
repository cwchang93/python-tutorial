# lesson05｜Dict 深入與 Set 入門

先預測、再執行，最後換資料測試。


## 例題 01｜List 與 Dict 的選擇

位置清單用 List；有欄位的一筆資料用 Dict。


```python
courses = ['Python', 'Excel']
student = {'name': 'Amy', 'score': 95}
print(courses)
print(student)
```

## 例題 02｜建立 Dict

大括號內用 Key: Value 表示欄位。


```python
student = {'name': 'Amy', 'age': 20, 'score': 95}
print(student)
```

## 例題 03｜用 Key 讀取

不用記位置，直接使用欄位名稱。


```python
student = {'name': 'Amy', 'score': 95}
print(student['name'])
print(student['score'])
```

## 例題 04｜修改 Value

相同 Key 會更新原本資料。


```python
student = {'name': 'Amy', 'score': 95}
student['score'] = 100
print(student)
```

## 例題 05｜新增欄位

指定不存在的 Key 就會新增。


```python
student = {'name': 'Amy'}
student['city'] = 'Taipei'
print(student)
```

## 例題 06｜get 安全查詢

找不到 Key 時回傳預設值，不讓程式停止。


```python
student = {'name': 'Amy'}
print(student.get('email'))
print(student.get('email', '未提供'))
```

## 例題 07｜刪除欄位

pop 可以取出並刪除欄位。


```python
product = {'name': '筆記本', 'price': 120, 'stock': 8}
stock = product.pop('stock')
print(stock, product)
```

## 例題 08｜keys、values、items

查看欄位、內容與配對。


```python
student = {'name': 'Amy', 'score': 95}
print(student.keys())
print(student.values())
print(student.items())
```

## 例題 09｜用 in 檢查 Key

in 檢查的是 Key。


```python
student = {'name': 'Amy', 'score': 95}
if 'score' in student:
    print('有成績資料')
```

## 例題 10｜商品小卡

Dict 適合描述一個具有多個欄位的物件。


```python
product = {'name': 'Notebook', 'price': 120, 'stock': 8}
product['stock'] -= 1
print(product)
```

## 例題 11｜List 裡放 Dict

多位學生就是多筆相同欄位的資料。


```python
students = [
    {'name': 'Amy', 'score': 95},
    {'name': 'Bob', 'score': 82}
]
print(students[0]['name'])
print(students[1]['score'])
```

## 例題 12｜Set 去除重複

相同 Email 只保留一份。


```python
emails = ['amy@test.com', 'bob@test.com', 'amy@test.com']
unique_emails = set(emails)
print(unique_emails)
```

## 例題 13｜建立空 Set

空 Set 要用 set()，因為 {} 是空 Dict。


```python
skills = set()
skills.add('Python')
skills.add('Python')
print(skills)
```

## 例題 14｜Set 新增與刪除

add 新增；discard 在不存在時也不報錯。


```python
skills = {'Python', 'Excel'}
skills.add('SQL')
skills.discard('Word')
print(skills)
```

## 例題 15｜交集與聯集

交集找共同，聯集找全部不重複。


```python
team_a = {'Python', 'Excel'}
team_b = {'Python', 'SQL'}
print(team_a & team_b)
print(team_a | team_b)
```

## 例題 16｜Dict 還是 Set

需要對應欄位用 Dict；只在乎唯一項目用 Set。


```python
profile = {'name': 'Amy', 'city': 'Taipei'}
skills = {'Python', 'Excel', 'Python'}
print(profile, skills)
```

## 例題 17｜資料整理小作品

用 Dict 描述商品，用 Set 管理標籤。


```python
product = {'name': '筆記本', 'price': 120, 'stock': 8}
tags = {'文具', '新品', '文具'}
if product['stock'] > 0:
    print(product['name'], product['price'], tags)
```

## 例題 18｜學習名片

整合 Dict、Set、List 與條件判斷。


```python
profile = {'name': 'Amy', 'hours': 4}
skills = {'print', 'List', 'Dict'}
tasks = ['完成練習', '複習']
if profile['hours'] >= 3:
    print(profile['name'], skills, tasks)
```

## 例題 19｜生活案例：更新商品資料

`update()` 一次更新多個欄位；`items()` 適合列印資料卡片。


```python
product = {'name': '保溫杯', 'price': 480, 'stock': 10}
product.update({'price': 450, 'stock': 8})
for key, value in product.items():
    print(key, ':', value)
```

## 例題 20｜生活案例：比較兩堂課的學員

Set 的交集找共同學員，差集找只出現在其中一組的人。


```python
python_class = {'Amy', 'Bob', 'Cindy'}
sql_class = {'Bob', 'Cindy', 'David'}
print('共同學員：', python_class & sql_class)
print('只上 Python：', python_class - sql_class)
```
