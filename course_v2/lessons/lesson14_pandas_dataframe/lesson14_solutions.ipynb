## 25｜完成健康檢查函式

建立 inspect_dataframe()，印出 head(3)、shape、columns、dtypes、缺值與重複列。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import pandas as pd

def inspect_dataframe(dataframe):
    print(dataframe.head(3))
    print("資料形狀：", dataframe.shape)
    print("欄位：", list(dataframe.columns))
    print("資料型態：")
    print(dataframe.dtypes)
    print("缺值：")
    print(dataframe.isna().sum())
    print("完整重複列：", dataframe.duplicated().sum())

df = pd.read_csv(data_path)
inspect_dataframe(df)
```
