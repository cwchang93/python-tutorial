## 練習 04｜⭐ 計算及格人數

計算 scores 中 >=60 的人數。


```python
scores=[55,72,88,40,95]
count=0
for score in scores:
    if score>=60:
        count += 1
print(count)
```

```python
scores=[70,55,80]
total=0
for score in scores:
    total += score
print(total/len(scores))
```

## 練習 07｜⭐⭐ 找最高分

不使用 max()，用 for 找最高分。


```python
scores=[88,65,100,72]
best=scores[0]
for score in scores:
    if score>best:
        best=score
print(best)
```

## 練習 08｜⭐⭐ 偶數總和

計算 numbers 中所有偶數的總和。


```python
numbers=[3,8,10,15,22]
total=0
for number in numbers:
    if number%2==0:
        total += number
print(total)
```

## 練習 09｜⭐⭐ 建立及格清單

## 練習 11｜⭐⭐ 會員訂單總額

orders 是 Dict List，只加總會員訂單。


```python
orders=[{'member':True,'amount':680},{'member':False,'amount':320}]
total=0
for order in orders:
    if order['member']:
        total += order['amount']
print(total)
```

## 練習 12｜⭐⭐ break 搜尋

找到第一筆 >=500 的訂單就停止。


```python
orders=[120,320,680,950]
for amount in orders:
    if amount>=500:
        print(amount)
        break
```

## 練習 13｜⭐⭐⭐ 庫存警示器

逐筆檢查商品，庫存 <5 印出警告。


```python
products=[{'name':'筆記本','stock':8},{'name':'鉛筆','stock':3}]
for product in products:
    if product['stock']<5:
        print('庫存警告：',product['name'])
```

## 練習 14｜⭐⭐⭐ 銷售統計

計算總營收、高價訂單數與平均訂單金額。


```python
orders=[{'name':'Amy','amount':680},{'name':'Bob','amount':320},{'name':'Cindy','amount':950}]
total=0
high=0
for order in orders:
    total += order['amount']
    if order['amount']>=500:
        high += 1
print('總營收：',total)
print('高價訂單數：',high)
print('平均：',total/len(orders))
```

## 練習 15｜⭐⭐⭐ 成績報表

逐筆印出排名、姓名、分數與及格狀態。


```python
students=[{'name':'Amy','score':95},{'name':'Bob','score':58}]
for index,student in enumerate(students,1):
    status='及格' if student['score']>=60 else '再加油'
    print(index,student['name'],student['score'],status)
```
