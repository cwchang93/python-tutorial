## 例題 13A｜break 與 continue 的差別

`break` 會結束整個迴圈；`continue` 只跳過目前這一輪，接著處理下一筆。
要停止全部搜尋用 `break`，只忽略一筆資料用 `continue`。


```python
scores=[45,52,68,75,90]
for score in scores:
    if score>=60:
        print('第一個及格分數：',score)
        break
```

## 例題 13B｜以字串使用 for 迴圈

字串也可以逐字走訪，每一輪取得一個字元。


```python
word='Python'
for index, char in enumerate(word):
    print('index',index,'→',char)
```

## 例題 06｜計算總和

```python
prices=[80,120,65,200]
total=0
for price in prices:
    total += price
print('總金額：',total)
```

## 例題 07｜計數器

每符合一次條件，就增加 1。


```python
scores=[55,72,88,40,95]
passed=0
for score in scores:
    if score>=60:
        passed += 1
print('及格人數：',passed)
```

## 例題 08｜計算平均

總和除以資料筆數。


```python
scores=[70,55,80]
total=0
for score in scores:
    total += score
average=total/len(scores)
print(average)
```

## 例題 09｜篩選資料

for 搭配 if，找出符合條件的資料。


```python
prices=[120,680,320,950]
for price in prices:
    if price>=500:
        print('高價訂單：',price)
```

## 例題 10｜建立新 List

把符合條件的資料收集起來。


```python
prices=[120,680,320,950]
high_prices=[]
for price in prices:
    if price>=500:
        high_prices.append(price)
print(high_prices)
```

## 例題 11｜找最大值

先設定目前最佳答案，再逐筆比較。


```python
scores=[88,65,100,72]
best=scores[0]
for score in scores:
    if score>best:
        best=score
print('最高分：',best)
```

## 例題 13｜continue 跳過本輪

遇到不需要的資料，直接進入下一輪。


```python
scores=[55,72,88,40,95]
for score in scores:
    if score<60:
        continue
    print('及格：',score)
```

## 例題 16｜List 裡的 Dict

多筆訂單可以逐筆讀取欄位。


```python
orders=[{'name':'Amy','amount':680},{'name':'Bob','amount':320}]
for order in orders:
    print(order['name'],order['amount'])
```

## 例題 17｜銷售統計器

整合 List、Dict、for、if、累加與計數。


```python
orders=[{'name':'Amy','amount':680},{'name':'Bob','amount':320},{'name':'Cindy','amount':950}]
total=0
high_value_count=0
for order in orders:
    total += order['amount']
    if order['amount']>=500:
        high_value_count += 1
print('總營收：',total)
print('高價訂單數：',high_value_count)
```

## 例題 18｜生活案例：成績排行榜

`sorted()` 產生新 List；`sort()` 會直接修改原本 List。


```python
scores = [88, 65, 100, 72]
print('由小到大：', sorted(scores))
print('由大到小：', sorted(scores, reverse=True))
print('原始成績仍在：', scores)
```

## 例題 19｜排序後逐筆製作報表

排序後搭配 `for`，就能逐筆列出排行榜。


```python
sales = [680, 320, 950, 510]
for rank, amount in enumerate(sorted(sales, reverse=True), 1):
    print(rank, '名：', amount, '元')
```

## 例題 22｜跳過缺貨商品


```python
products = [{'name': '筆記本', 'stock': 3}, {'name': '貼紙', 'stock': 0}, {'name': '筆', 'stock': 5}]
for product in products:
    if product['stock'] == 0:
        continue
    print(product['name'])
```

## 例題 25｜每日銷售摘要


```python
orders = [{'amount': 120}, {'amount': 180}, {'amount': 80}]
total = 0
for order in orders:
    total += order['amount']
print('訂單數：', len(orders))
print('今日營收：', total)
```

## 例題 26｜雙層迴圈：列印座位

外層處理排數，內層處理每排的座位。


```python
for row in range(1, 3):
    for seat in range(1, 4):
        print('第', row, '排 - 第', seat, '號')
```
