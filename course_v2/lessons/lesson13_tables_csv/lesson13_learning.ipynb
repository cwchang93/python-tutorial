# Lesson 13｜資料分析入門：表格與 CSV

從列、欄、儲存格一路走到第一次 CSV 分析。

共 25 個逐步示範；請依序執行。


## 01｜資料表：把同一類資料排整齊

先用 list 裝多筆訂單；每個 dict 就是一列。


```python
orders = [
    {"order_id": "A001", "product": "美式咖啡", "quantity": 2},
    {"order_id": "A002", "product": "拿鐵", "quantity": 1},
]
print(orders)
```

## 02｜欄（column）：固定的一種意義

欄名像表格上方的標籤，告訴我們每個位置代表什麼。


```python
columns = ["order_id", "product", "quantity"]
for column in columns:
    print("欄位：", column)
```

## 03｜列（row）：一筆完整紀錄

一列應該能描述一張訂單，而不是把不同訂單混在一起。


```python
row = {"order_id": "A003", "product": "貝果", "quantity": 3}
print("這一列代表訂單：", row["order_id"])
print(row)
```

## 04｜儲存格（cell）：列與欄交會的值

指定一列，再指定一個欄名，就能取得一個儲存格。


```python
row = {"order_id": "A003", "product": "貝果", "quantity": 3}
cell = row["product"]
print("product 儲存格：", cell)
```

## 05｜同一欄要維持同一種意義

quantity 欄都應該放數量，不能有時放數量、有時放備註。


```python
rows = [
    {"product": "美式咖啡", "quantity": 2},
    {"product": "拿鐵", "quantity": 1},
]
for row in rows:
    print(row["product"], "數量：", row["quantity"])
```

## 06｜資料型態會影響能做的事

數字可以計算，字串適合呈現文字。


```python
row = {"product": "美式咖啡", "quantity": 2, "unit_price": 80}
print(type(row["product"]))
print(type(row["quantity"]))
print(type(row["unit_price"]))
```

## 07｜從原始欄位算出新資訊

單筆營業額 = 數量 × 單價。這是分析中常見的衍生值。


```python
row = {"product": "美式咖啡", "quantity": 2, "unit_price": 80}
revenue = row["quantity"] * row["unit_price"]
print("單筆營業額：", revenue)
```

## 08｜先確認一共有幾筆資料

`len()` 回答表格有幾列；這是最基本的資料量檢查。


```python
orders = [
    {"order_id": "A001"},
    {"order_id": "A002"},
    {"order_id": "A003"},
]
print("訂單筆數：", len(orders))
```

## 09｜分析從問題開始

先把想知道的事情寫成清楚的問題，再決定需要什麼資料。


```python
question = "哪一項商品賣出的數量最多？"
print("分析問題：", question)
```

## 10｜問題要對應需要的欄位

要回答商品銷量，就需要 product 與 quantity，不需要把所有欄位都拿來算。


```python
question = "哪一項商品賣出的數量最多？"
needed_columns = ["product", "quantity"]
print("問題：", question)
print("需要欄位：", needed_columns)
```

## 11｜CSV 是用逗號分隔的純文字

第一行通常是標題列，後面每一行是一筆資料。


```python
csv_text = """order_id,product,quantity
A001,美式咖啡,2
A002,拿鐵,1"""
print(csv_text)
```

## 12｜用 StringIO 模擬一個 CSV 檔案

課堂先在記憶體中練習，不會改到硬碟裡的原始資料。


```python
from io import StringIO

csv_text = "order_id,product,quantity\nA001,美式咖啡,2"
csv_file = StringIO(csv_text)
print(csv_file.read())
```

## 13｜csv.reader：每一列會變成 list

`reader` 只知道位置，不知道欄位意義；第一列也會當成一般資料。


```python
import csv
from io import StringIO

csv_text = "order_id,product,quantity\nA001,美式咖啡,2"
reader = csv.reader(StringIO(csv_text))
for row in reader:
    print(row)
```

## 14｜csv.DictReader：每一列會變成 dict

標題列會自動成為 key，讀程式時更容易看懂欄位意義。


```python
import csv
from io import StringIO

csv_text = "order_id,product,quantity\nA001,美式咖啡,2"
reader = csv.DictReader(StringIO(csv_text))
for row in reader:
    print(row["product"], row["quantity"])
```

## 18｜計算前先轉換數字型態

用 `int()` 把數量和單價從文字轉為整數。

## 19｜逐列計算總營業額

每次處理一筆訂單，再把結果累加到 total_revenue。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

total_revenue = 0
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        quantity = int(order["quantity"])
        unit_price = int(order["unit_price"])
        total_revenue = total_revenue + quantity * unit_price

print("總營業額：", total_revenue)
```

## 20｜用 dict 累計每項商品數量

第一次看到商品就從 0 開始，之後再把數量加上去。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

product_quantity = {}
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        product = order["product"]
        quantity = int(order["quantity"])
        if product not in product_quantity:
            product_quantity[product] = 0
        product_quantity[product] = product_quantity[product] + quantity

print(product_quantity)
```

## 21｜用 set 找出不重複城市

set 會自動留下唯一值，適合回答『資料中出現哪些城市？』。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

cities = set()
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        cities.add(order["city"])

print("城市數量：", len(cities))
print("城市：", sorted(cities))
```

## 22｜資料品質檢查：是否有空白值

逐列逐欄檢查；空字串代表這個儲存格沒有內容。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

missing_count = 0
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        for value in order.values():
            if value == "":
                missing_count = missing_count + 1

print("空白儲存格：", missing_count)
```

## 23｜資料品質檢查：訂單編號是否重複

唯一編號如果重複，可能代表同一筆資料被匯入兩次。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

seen_ids = set()
duplicate_ids = []
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        order_id = order["order_id"]
        if order_id in seen_ids:
            duplicate_ids.append(order_id)
        else:
            seen_ids.add(order_id)

print("重複編號：", duplicate_ids)
```

## 24｜原始資料先保留，不直接覆寫

分析時保留 raw_orders，再建立 cleaned_orders，出錯時才有資料可回頭比對。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

with data_path.open(encoding="utf-8", newline="") as file:
    raw_orders = list(csv.DictReader(file))

cleaned_orders = []
for order in raw_orders:
    cleaned_order = order.copy()
    cleaned_order["product"] = cleaned_order["product"].strip()
    cleaned_orders.append(cleaned_order)

print("原始筆數：", len(raw_orders))
print("整理後筆數：", len(cleaned_orders))
```

## 25｜完整分析流程：問題 → 欄位 → 方法 → 結果

把分析步驟說清楚，比只丟出一個數字更容易驗證。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv

question = "全部訂單的總營業額是多少？"
needed_columns = ["quantity", "unit_price"]
total_revenue = 0

with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        total_revenue = total_revenue + int(order["quantity"]) * int(order["unit_price"])

print("問題：", question)
print("需要欄位：", needed_columns)
print("方法：逐列計算數量 × 單價，再全部加總")
print("結果：", total_revenue)
```
