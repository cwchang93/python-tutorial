## 21｜統計付款方式次數

用 dict 累計 payment 出現的次數。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv
payment_counts = {}
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        payment = order["payment"]
        if payment not in payment_counts:
            payment_counts[payment] = 0
        payment_counts[payment] = payment_counts[payment] + 1
print(payment_counts)
```

## 23｜檢查重複訂單編號

使用 set 找出重複的 order_id。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv
seen_ids = set()
duplicate_ids = []
with data_path.open(encoding="utf-8", newline="") as file:
    for order in csv.DictReader(file):
        order_id = order["order_id"]
        if order_id in seen_ids:
            duplicate_ids.append(order_id)
        else:
            seen_ids.add(order_id)
print(duplicate_ids)
```

## 24｜保留原始資料

建立 cleaned_orders；每筆用 copy() 複製後再 strip product。


```python
from pathlib import Path

possible_paths = [
    Path("course_v2/data/lesson13_orders.csv"),
    Path("data/lesson13_orders.csv"),
    Path("../../data/lesson13_orders.csv"),
    Path("lesson13_orders.csv"),
]

data_path = possible_paths[0]
for path in possible_paths:
    if path.exists():
        data_path = path
        break

if not data_path.exists():
    raise FileNotFoundError("找不到 lesson13_orders.csv，請確認已下載 course_v2/data 資料夾")

print("資料位置：", data_path)

import csv
with data_path.open(encoding="utf-8", newline="") as file:
    raw_orders = list(csv.DictReader(file))
cleaned_orders = []
for order in raw_orders:
    cleaned_order = order.copy()
    cleaned_order["product"] = cleaned_order["product"].strip()
    cleaned_orders.append(cleaned_order)
print(len(raw_orders), len(cleaned_orders))
```
