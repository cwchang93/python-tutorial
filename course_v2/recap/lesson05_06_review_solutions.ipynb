## Lesson 05｜題目 3：安全查詢
使用 get() 查詢會員的 email；若沒有這個欄位，顯示「未提供」。

```python
member = {'name': 'Amy', 'level': '金卡'}

email = member.get('email', '未提供')
print(email)
```

## Lesson 05｜題目 4：走訪會員卡
使用 items() 逐筆印出會員資料的欄位名稱與內容。

```python
member = {'name': 'Bob', 'level': '銀卡', 'points': 120}

for key, value in member.items():
    print(key, value)
```

## Lesson 05｜題目 5：安全刪除欄位
如果 coupon 存在，就使用 pop() 取出並印出；否則顯示「沒有優惠券」。

```python
order = {'id': 'A001', 'amount': 850, 'coupon': 'SALE50'}

if 'coupon' in order:
    coupon = order.pop('coupon')
    print(coupon)
else:
    print('沒有優惠券')

print(order)
```

## Lesson 05｜題目 6：Email 去重
使用 Set 去除重複的 Email，並印出不重複的人數。

```python
emails = ['amy@example.com', 'bob@example.com', 'amy@example.com']

unique_emails = set(emails)
unique_count = len(unique_emails)

print(unique_emails)
print(unique_count)
```

## Lesson 05｜題目 7：課程名單比較
找出兩堂課的共同學員，以及所有不重複的學員。

```python
python_students = {'Amy', 'Bob', 'Cindy'}
sql_students = {'Bob', 'Cindy', 'David'}

common_students = python_students & sql_students
all_students = python_students | sql_students

print('共同學員：', common_students)
print('所有學員：', all_students)
```

## Lesson 05｜題目 8：會員折扣
會員且消費滿 500 元時打九折，否則維持原價。印出最後金額。

```python
customer = {'name': 'Cindy', 'is_member': True, 'amount': 800}

if customer['is_member'] == True and customer['amount'] >= 500:
    final_amount = customer['amount'] * 0.9
else:
    final_amount = customer['amount']

print(final_amount)
```

## Lesson 06｜題目 11：購物車總價

## Lesson 06｜題目 12：計算及格人數
使用 for 與 if 計算分數大於等於 60 的人數。

```python
scores = [95, 58, 76, 43, 88]
passed_count = 0

for score in scores:
    if score >= 60:
        passed_count = passed_count + 1

print(passed_count)
```

## Lesson 06｜題目 13：建立及格清單
將大於等於 60 的分數加入 passed，最後印出 passed。

```python
scores = [72, 45, 90, 58, 63]
passed = []

for score in scores:
    if score >= 60:
        passed.append(score)

print(passed)
```

## Lesson 06｜題目 14：找到第一筆大額訂單
找到第一筆金額大於等於 500 的訂單後印出金額，並使用 break 停止搜尋。

```python
amounts = [120, 360, 780, 950]

for amount in amounts:
    if amount >= 500:
        print(amount)
        break
```

## Lesson 06｜題目 15：跳過缺貨商品
逐筆檢查商品；遇到庫存為 0 時使用 continue 跳過，只印出有庫存的商品名稱。

```python
products = [
    {'name': '鉛筆', 'stock': 10},
    {'name': '橡皮擦', 'stock': 0},
    {'name': '筆記本', 'stock': 5}
]

for product in products:
    if product['stock'] == 0:
        continue

    print(product['name'])
```

## Lesson 05–06｜題目 16：學生成績報表
使用 enumerate() 從 1 開始編號，逐筆印出學生的編號、姓名、分數與狀態。

- 分數大於等於 60：`及格`
- 分數低於 60：`再加油`
- 請使用一般的 if / else 區塊完成判斷。

```python
students = [
    {'name': 'Amy', 'score': 95},
    {'name': 'Bob', 'score': 58}
]

for index, student in enumerate(students, 1):
    if student['score'] >= 60:
        status = '及格'
    else:
        status = '再加油'

    print(index, student['name'], student['score'], status)
```
