## Git 指令複習表

Git 指令要在 **Terminal／終端機** 執行，不是 Python 程式碼。先看懂用途，再照順序操作。

| 指令 | 用途 | 什麼時候使用 |
|---|---|---|
| `git --version` | 確認電腦是否已安裝 Git | 第一次使用 Git 時 |
| `git init` | 把目前資料夾初始化成 Git Repository | 新專案第一次啟用版本控制 |
| `git status` | 查看新增、修改、已暫存與未暫存的檔案 | 每個步驟前後都可以使用 |
| `git add 檔名` | 把指定檔案放進暫存區 | 只準備提交某個檔案 |
| `git add .` | 把目前資料夾的變更放進暫存區 | 確認所有變更都要提交時 |
| `git diff` | 查看尚未放入暫存區的修改內容 | Commit 前檢查改了什麼 |
| `git diff --staged` | 查看已放入暫存區、即將提交的內容 | `git add` 之後再次確認 |
| `git commit -m "訊息"` | 建立一筆本機版本紀錄 | 暫存區內容確認完成後 |
| `git log` | 查看完整 Commit 歷史 | 想查看作者、日期與訊息時 |
| `git log --oneline` | 用一行顯示一筆 Commit | 想快速查看版本歷史時 |
| `git branch -M main` | 將目前分支命名為 `main` | 第一次準備推送 GitHub 時 |
| `git remote add origin 網址` | 連結本機 Repository 與 GitHub | 第一次設定遠端位置時 |
| `git remote -v` | 查看目前設定的遠端網址 | Push 前確認 GitHub 網址 |
| `git push -u origin main` | 第一次把 `main` 推送到 GitHub | Remote 設定完成後 |
| `git push` | 推送後續 Commit | 已設定上游分支後 |
| `git pull` | 取得 GitHub 上的新版本並合併 | 遠端可能有其他更新時 |

### 最常用的流程

`修改檔案 → git status → git diff → git add → git status → git commit → git push`

- 工作區：你正在修改的檔案。
- 暫存區：這一次準備 Commit 的內容。
- 本機 Repository：已經 Commit、保存在自己電腦的版本。
- GitHub：Push 後保存於網路上的版本。
