## 題目 1｜及格判斷
請設定一個分數。當分數大於等於 60 時，顯示「及格」；否則顯示「需要補強」。


```python
score = 72
if score >= 60:
    result = '及格'
else:
    result = '需要補強'

print(result)
```

## 題目 2｜奇偶數
請設定一個整數，判斷它是偶數還是奇數。能被 2 整除的是偶數，否則是奇數。


```python
number = 17
if number % 2 == 0:
    result = '偶數'
else:
    result = '奇數'

print(result)
```

## 題目 3｜溫度提醒
請設定目前溫度：低於 18 顯示「偏冷」、18 到 27 顯示「舒適」、高於 27 顯示「偏熱」。


```python
temperature = 25
if temperature < 18:
    print('偏冷')
elif temperature <= 27:
    print('舒適')
else:
    print('偏熱')
```

## 題目 4｜會員折扣
會員且消費滿 1000 元享 9 折，請計算折扣後金額；其他情況維持原價。


```python
is_member = True
amount = 1200
if is_member and amount >= 1000:
    final_amount = amount * 0.9
else:
    final_amount = amount

print('應付金額：', final_amount)
```

## 題目 5｜外送費
離島運費為 150 元；本島消費滿 800 元免運；其他本島訂單運費為 80 元。請計算運費。


```python
region = '本島'
amount = 650
if region == '離島':
    shipping = 150
elif amount >= 800:
    shipping = 0
else:
    shipping = 80

print('運費：', shipping)
```

## 題目 7｜ATM 提款
只有餘額足夠，而且提款金額是 100 的倍數時才能提款，請顯示結果。


```python
balance = 3000
withdraw = 1200
if balance >= withdraw and withdraw % 100 == 0:
    print('提款成功')
else:
    print('無法提款')
```

## 題目 8｜電量提醒
電量低於 20 顯示「請充電」、20 到 79 顯示「電量普通」、80 以上顯示「電量充足」。

## 題目 9｜索引與負索引
請從課程清單取出第一個課程、最後一個課程，並顯示清單長度。


```python
courses = ['Python', 'Excel', 'SQL']
first_course = courses[0]
last_course = courses[-1]
course_count = len(courses)
print('第一個課程：', first_course)
print('最後一個課程：', last_course)
print('課程數量：', course_count)
```

## 題目 10｜切片
請從數字清單取出前兩個、後兩個，以及反轉後的清單，分別顯示結果。


```python
numbers = [10, 20, 30, 40, 50]
first_two = numbers[:2]
last_two = numbers[-2:]
reversed_numbers = numbers[::-1]
print('前兩個：', first_two)
print('後兩個：', last_two)
print('反轉後：', reversed_numbers)
```

## 題目 11｜購物車增刪
請在購物車最後加入雞蛋、最前面加入牛奶，再移除麵包，最後顯示購物車。


```python
cart = ['麵包', '水果']
cart.append('雞蛋')
cart.insert(0, '牛奶')
cart.remove('麵包')
print('購物車：', cart)
```

## 題目 12｜候位叫號

## 題目 13｜List 運算
請合併早上的行程與下午的行程，並把喝水提醒重複三次。


```python
morning = ['早餐', '通勤']
afternoon = ['上課', '運動']
reminder = ['記得喝水']
daily_tasks = morning + afternoon
water_reminders = reminder * 3
print('每日行程：', daily_tasks)
print('提醒：', water_reminders)
```

## 題目 14｜copy 與 alias
請建立同一份清單的 alias 與 copy，修改 alias 後觀察原清單與複製清單的差異。


```python
shopping = ['牛奶', '麵包']
alias = shopping
copied = shopping.copy()
alias.append('雞蛋')
print('原清單：', shopping)
print('複製清單：', copied)
```

## 題目 15｜Tuple 拆包
請將座標 Tuple 拆成緯度與經度兩個變數，再分別顯示。
