## Lesson 07｜題目 1：while 基本計數

使用 while 依序印出 1、2、3、4、5。記得在迴圈中更新計數變數。


```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

## Lesson 07｜題目 2：倒數計時

使用 while 印出 5、4、3、2、1，最後顯示「時間到」。


```python
count = 5

while count >= 1:
    print(count)
    count -= 1

print('時間到')
```

## Lesson 07｜題目 3：累加到指定數字

使用 while 計算 1 到 5 的總和，最後清楚顯示總和。


```python
number = 1
total = 0

while number <= 5:
    total += number
    number += 1

print('總和：', total)
```

## Lesson 07｜題目 4：密碼重試三次

密碼為 python123。最多輸入三次；答對顯示「登入成功」並停止，三次都錯則顯示「帳號已鎖定」。


```python
password = 'python123'
attempts = 0

while attempts < 3:
    user_input = input('請輸入密碼：')
    attempts += 1

    if user_input == password:
        print('登入成功')
        break
    else:
        print('密碼錯誤')
else:
    print('帳號已鎖定')
```

## Lesson 07｜題目 5：數字格式驗證

持續要求輸入整數。格式錯誤時顯示提示並重新輸入；成功後印出該數字並停止。


```python
while True:
    text = input('請輸入整數：')

    try:
        number = int(text)
        print('你輸入的數字：', number)
        break
    except ValueError:
        print('格式錯誤，請重新輸入')
```

## Lesson 07｜題目 6：合理範圍驗證

持續要求輸入 1 到 5 的滿意度。不是整數或超出範圍時重新輸入。


```python
while True:
    text = input('滿意度（1～5）：')

    try:
        score = int(text)
    except ValueError:
        print('請輸入整數')
        continue

    if score < 1 or score > 5:
        print('請輸入 1 到 5')
        continue

    print('已記錄：', score)
    break
```

## Lesson 07｜題目 8：收集多筆成績

持續輸入成績並加入 scores；輸入 end 時停止。只接受 0 到 100 的數字，最後印出全部成績。


```python
scores = []

while True:
    text = input('成績（end 結束）：')

    if text == 'end':
        break

    try:
        score = float(text)
    except ValueError:
        print('格式錯誤')
        continue

    if score < 0 or score > 100:
        print('成績必須介於 0 到 100')
        continue

    scores.append(score)

print('全部成績：', scores)
```

## Lesson 08｜題目 15：處理 List

建立 calculate_average(scores)。空 List 回傳 0；有資料時回傳平均分數。請測試兩種情況。


```python
def calculate_average(scores):
    if len(scores) == 0:
        return 0

    total = 0
    for score in scores:
        total += score

    average = total / len(scores)
    return average

print('平均：', calculate_average([80, 90, 70]))
print('空資料平均：', calculate_average([]))
```

## Lesson 07–08｜題目 16：整合練習：安全成績報表

持續輸入 0 到 100 的成績，輸入 end 結束。請用函式驗證範圍、計算平均，最後顯示筆數與平均；沒有資料時顯示「沒有成績」。


```python
def is_valid_score(score):
    if score >= 0 and score <= 100:
        return True
    else:
        return False


def calculate_average(scores):
    if len(scores) == 0:
        return None

    total = 0
    for score in scores:
        total += score

    return total / len(scores)


scores = []

while True:
    text = input('成績（end 結束）：')

    if text == 'end':
        break

    try:
        score = float(text)
    except ValueError:
        print('格式錯誤')
        continue

    if is_valid_score(score) == False:
        print('成績必須介於 0 到 100')
        continue

    scores.append(score)

average = calculate_average(scores)

if average is None:
    print('沒有成績')
else:
    print('成績筆數：', len(scores))
    print('平均分數：', average)
```
