## Lesson 13｜題目 3：從原始欄位算出營業額

使用 for 逐筆計算 `quantity * unit_price`，把結果存入每筆訂單的 `revenue`，最後印出訂單。


```python
orders = [
    {'product': '拿鐵', 'quantity': 2, 'unit_price': 120},
    {'product': '貝果', 'quantity': 3, 'unit_price': 65},
]

for order in orders:
    revenue = order['quantity'] * order['unit_price']
    order['revenue'] = revenue

for order in orders:
    print(order)
```

## Lesson 13｜題目 4：使用 DictReader 讀取 CSV

讀取本複習提供的 CSV，印出欄名、資料筆數與第一筆資料。提醒：DictReader 讀出的每一列是 Dict。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print('欄名：', reader.fieldnames)
print('資料筆數：', len(rows))
print('第一筆：', rows[0])
```

## Lesson 13｜題目 5：CSV 數字欄位需要轉型

CSV 的 quantity 與 unit_price 讀進來是字串。請轉成 int，逐筆計算並累加總營業額。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

total_revenue = 0
for row in rows:
    quantity = int(row['quantity'])
    unit_price = int(row['unit_price'])
    revenue = quantity * unit_price
    total_revenue += revenue

print('總營業額：', total_revenue)
```

## Lesson 13｜題目 6：用 Dict 累計商品銷量

逐筆讀取 CSV，用 Dict 累加每種商品的 quantity，最後逐項印出商品與總數量。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

product_quantity = {}
for row in rows:
    product = row['product']
    quantity = int(row['quantity'])

    if product not in product_quantity:
        product_quantity[product] = 0

    product_quantity[product] += quantity

for product, quantity in product_quantity.items():
    print(product, quantity)
```

## Lesson 13｜題目 7：用 Set 找不重複城市

逐筆把 city 加入 Set，最後印出城市數量與排序後的城市。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import csv

with csv_path.open(encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

cities = set()
for row in rows:
    cities.add(row['city'])

print('城市數量：', len(cities))
print('城市：', sorted(cities))
```

## Lesson 13｜題目 8：檢查空白與重複編號

檢查 dirty_orders：找出含空白欄位的資料，並找出重複的 order_id。請保留完整步驟。


```python
dirty_orders = [
    {'order_id': 'A001', 'product': '拿鐵', 'city': '台北'},
    {'order_id': 'A002', 'product': '', 'city': '台中'},
    {'order_id': 'A001', 'product': '貝果', 'city': '高雄'},
]

blank_rows = []
duplicate_ids = []
seen_ids = set()

for order in dirty_orders:
    has_blank = False
    for value in order.values():
        if value == '':
            has_blank = True

    if has_blank:
        blank_rows.append(order)

    order_id = order['order_id']
    if order_id in seen_ids:
        duplicate_ids.append(order_id)
    else:
        seen_ids.add(order_id)

print('空白資料：', blank_rows)
print('重複編號：', duplicate_ids)
```

## Lesson 13–14｜題目 16：整合練習：從 CSV 到第一份資料健檢報告

建立 `check_dataframe(df)`：回傳資料大小、欄名、型態、缺值、重複列、重複 order_id 與各城市訂單筆數。呼叫後印出前三列和報告，最後用一句話說明資料是否適合進入下一步分析。


```python
from pathlib import Path

csv_path = Path("lesson13_14_review_orders.csv")

import pandas as pd

df = pd.read_csv(csv_path)

def check_dataframe(df):
    rows, columns = df.shape
    report = {
        'rows': rows,
        'columns': columns,
        'column_names': list(df.columns),
        'data_types': df.dtypes.astype(str).to_dict(),
        'missing': int(df.isna().sum().sum()),
        'duplicate_rows': int(df.duplicated().sum()),
        'duplicate_order_ids': int(df.duplicated(subset=['order_id']).sum()),
        'orders_by_city': df['city'].value_counts().to_dict(),
    }
    return report


report = check_dataframe(df)
print('前三列：')
print(df.head(3))
print('健康檢查報告：')
for item, value in report.items():
    print(item, '：', value)

if report['missing'] == 0 and report['duplicate_rows'] == 0:
    print('初步檢查完成：資料可以進入下一步分析。')
else:
    print('資料還有品質問題，應先確認與清理。')
```
