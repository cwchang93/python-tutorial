## Git 指令複習表

| 指令 | 用途 |
|---|---|
| `git init` | 初始化 Git Repository |
| `git status` | 查看目前檔案狀態 |
| `git add 檔名`／`git add .` | 把變更放進暫存區 |
| `git diff` | 查看尚未暫存的修改 |
| `git diff --staged` | 查看即將 Commit 的修改 |
| `git commit -m "訊息"` | 建立本機版本紀錄 |
| `git log`／`git log --oneline` | 查看 Commit 歷史 |
| `git branch -M main` | 將目前分支命名為 main |
| `git remote add origin 網址` | 連結 GitHub Repository |
| `git remote -v` | 確認遠端網址 |
| `git push -u origin main` | 第一次推送 main |
| `git push` | 推送後續 Commit |
| `git pull` | 取得並合併遠端更新 |

常用流程：`修改 → status → diff → add → status → commit → push`

## Lesson 12｜解答 9：建立記帳資料

使用 List 與 Dict 建立三筆交易，欄位包含 `date`、`category`、`amount`、`note`；收入為正數，支出為負數。

```python
records = [
    {'date': '2026-08-01', 'category': '薪資', 'amount': 30000, 'note': '八月薪水'},
    {'date': '2026-08-02', 'category': '餐飲', 'amount': -150, 'note': '午餐'},
    {'date': '2026-08-03', 'category': '交通', 'amount': -50, 'note': '捷運'},
]
print('交易筆數：', len(records))
for record in records:
    print(record)
```

## Lesson 12｜解答 10：計算目前餘額

使用 `for` 逐筆累加 `amount`，算出目前餘額。不要直接寫死答案。

```python
balance = 0
for record in records:
    balance += record['amount']
print('目前餘額：', balance)
```

## Lesson 12｜解答 11：安全新增交易

建立 `add_record()`。分類去除前後空白後不可為空；金額文字需能轉成 float。資料正確時加入 List 並回傳新交易，錯誤時提出 ValueError。

```python
def add_record(records, date, category, amount_text, note):
    clean_category = category.strip()
    if clean_category == '':
        raise ValueError('分類不可空白')
    try:
        amount = float(amount_text)
    except ValueError:
        raise ValueError('金額必須是數字')
    record = {
        'date': date,
        'category': clean_category,
        'amount': amount,
        'note': note.strip(),
    }
    records.append(record)
    return record

new_record = add_record(records, '2026-08-04', ' 餐飲 ', '-180', '晚餐')
print('新增：', new_record)
```

## Lesson 12｜解答 12：收入、支出與餘額摘要

建立 `calculate_summary(records)`，回傳包含 `income`、`expense`、`balance` 的 Dict。支出請顯示為正數。

```python
def calculate_summary(records):
    income = 0
    expense = 0
    for record in records:
        if record['amount'] >= 0:
            income += record['amount']
        else:
            expense += abs(record['amount'])
    return {
        'income': income,
        'expense': expense,
        'balance': income - expense,
    }

summary = calculate_summary(records)
print('收入：', summary['income'])
print('支出：', summary['expense'])
print('餘額：', summary['balance'])
```

## Lesson 12｜解答 13：分類支出統計

建立 `expense_by_category(records)`，只統計支出，使用 Dict 累加每個分類的支出金額。

```python
def expense_by_category(records):
    summary = {}
    for record in records:
        if record['amount'] < 0:
            category = record['category']
            amount = abs(record['amount'])
            current_amount = summary.get(category, 0)
            summary[category] = current_amount + amount
    return summary

print(expense_by_category(records))
```

## Lesson 12｜解答 14：保存與讀回 CSV

在練習專案的 `data/records.csv` 保存交易，再讀回成 List。請使用 UTF-8、`DictWriter`、`DictReader`，並把讀回的 `amount` 轉回 float。

```python
import csv

data_dir = project / 'data'
data_dir.mkdir(exist_ok=True)
csv_path = data_dir / 'records.csv'
fieldnames = ['date', 'category', 'amount', 'note']

with csv_path.open('w', newline='', encoding='utf-8-sig') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(records)

loaded_records = []
with csv_path.open(encoding='utf-8-sig') as file:
    reader = csv.DictReader(file)
    for row in reader:
        row['amount'] = float(row['amount'])
        loaded_records.append(row)

print('保存位置：', csv_path.resolve())
print('讀回筆數：', len(loaded_records))
```

## Lesson 12｜解答 15：產生文字摘要

建立 `build_report(records)`，回傳多行文字，清楚顯示收入、支出、餘額與各分類支出。將結果保存為 `summary.txt`。

```python
def build_report(records):
    summary = calculate_summary(records)
    categories = expense_by_category(records)
    lines = [
        '個人記帳摘要',
        f"收入：{summary['income']:.0f}",
        f"支出：{summary['expense']:.0f}",
        f"餘額：{summary['balance']:.0f}",
        '分類支出：',
    ]
    for category, amount in categories.items():
        lines.append(f'- {category}：{amount:.0f}')
    return '\n'.join(lines)

report = build_report(loaded_records)
report_path = project / 'summary.txt'
report_path.write_text(report, encoding='utf-8')
print(report)
print('報表位置：', report_path.resolve())
```

把記帳程式、CSV、報表與 README 加入同一個練習 Repository，建立 `feat: complete expense tracker` Commit，確認歷史與工作區乾淨，最後只印出 GitHub 發布指令。**不要在 Notebook 自動 push。**
