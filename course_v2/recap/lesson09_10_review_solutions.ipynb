## 題目 16｜安全 CSV 摘要


```python
import csv
from pathlib import Path
def summarize(path):
    total=0; errors=[]
    with path.open(encoding='utf-8') as file:
        for row in csv.DictReader(file):
            try: total+=int(row['quantity'])*int(row['unit_price'])
            except (KeyError,ValueError) as error: errors.append(str(error))
    return total, errors
csv_path=Path('lesson10_sales.csv')
print(summarize(csv_path))
```
