## 🔧 Step 1: Import Module

我們要 import 自己寫的 `dog` 模組。但因為它在 `src/` 資料夾，需要先把 `src/` 加入 Python 的搜尋路徑。

```python
import sys
from pathlib import Path

# Add src folder to Python path
project_root = Path.cwd().parent  # Go up from notebooks/ to project root
src_path = project_root / "src"
sys.path.insert(0, str(src_path))

# Now we can import our dog module!
import dog

# Also import other libraries
import pandas as pd
import matplotlib.pyplot as plt

print("✅ Modules imported successfully!")
print(f"📂 Project root: {project_root}")
print(f"📂 Source path: {src_path}")
```

## 📂 Step 2: Read and Explore Raw Data

使用 `dog.read_dogs_csv()` 讀取原始資料，並查看基本資訊。

```python
# Define data path
data_path = project_root / "data" / "dogs.csv"

# Read data using our module function
df_raw = dog.read_dogs_csv(data_path)

# Display basic info
print(f"\n📊 Dataset shape: {df_raw.shape[0]} rows × {df_raw.shape[1]} columns")
print(f"\n📋 Column names:")
print(df_raw.columns.tolist())
```

```python
# View first few rows
print("👀 First 5 rows:")
df_raw.head()
```

```python
# Check data types and missing values
print("🔍 Data Info:")
df_raw.info()
```

```python
# Basic statistics
print("📈 Descriptive Statistics:")
df_raw.describe()
```

## 🧹 Step 3: Clean Data

使用 `dog.clean_dogs_df()` 清理資料，並比較清理前後的差異。

```python
# Clean data using our module function
df_clean = dog.clean_dogs_df(df_raw)

# Compare data types before and after
print("🔄 Data Type Comparison:")
print("\nBefore cleaning:")
print(df_raw.dtypes)
print("\n" + "="*50)
print("\nAfter cleaning:")
print(df_clean.dtypes)
```

```python
# Check for missing values
print("❓ Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
# View cleaned data
print("✨ Cleaned data (first 10 rows):")
df_clean.head(10)
```

## 📊 Step 4: Generate Statistics

使用 `dog.stats_summary()` 生成統計摘要。

```python
# Generate statistics
stats = dog.stats_summary(df_clean)

# Display in a nice format
print("=" * 60)
print("📊  STATISTICS SUMMARY".center(60))
print("=" * 60)

print(f"\n📈 Overall:")
print(f"   • Total Dogs: {stats['total_dogs']}")
print(f"   • Adopted: {stats['adopted_count']} ({stats['adoption_rate']:.1f}%)")
print(f"   • Vaccinated: {stats['vaccinated_count']}")

print(f"\n🐕 Physical Stats:")
print(f"   • Average Age: {stats['avg_age']:.1f} years")
print(f"   • Average Weight: {stats['avg_weight']:.1f} kg")
print(f"   • Most Common Breed: {stats['most_common_breed']}")

print(f"\n🏙️  By City:")
for city, count in sorted(stats['cities'].items(), key=lambda x: x[1], reverse=True):
    percentage = (count / stats['total_dogs']) * 100
    print(f"   • {city}: {count} dogs ({percentage:.1f}%)")

print(f"\n🐾 By Breed:")
for breed, count in list(sorted(stats['breeds'].items(), key=lambda x: x[1], reverse=True))[:5]:
    percentage = (count / stats['total_dogs']) * 100
    print(f"   • {breed}: {count} dogs ({percentage:.1f}%)")
```

## 🔍 Step 5: Filter and Sort Data

展示如何使用 `dog.filter_dogs()` 和 `dog.sort_dogs()` 進行資料篩選與排序。

```python
# Example 1: Filter by city
print("🔍 Example 1: Dogs in Taipei (台北)")
taipei_dogs = dog.filter_dogs(df_clean, city="台北")
print(taipei_dogs[['name', 'breed', 'age', 'adopted']])
```

```python
# Example 2: Filter by age range
print("🔍 Example 2: Young dogs (1-3 years old)")
young_dogs = dog.filter_dogs(df_clean, min_age=1, max_age=3)
print(young_dogs[['name', 'breed', 'age', 'city']])
```

```python
# Example 3: Filter adopted dogs and sort by weight
print("🔍 Example 3: Adopted dogs, sorted by weight (descending)")
adopted_dogs = dog.filter_dogs(df_clean, adopted=True)
adopted_sorted = dog.sort_dogs(adopted_dogs, by="weight_kg", ascending=False)
print(adopted_sorted[['name', 'breed', 'age', 'weight_kg', 'city']])
```

```python
# Example 4: Complex filter - Unadopted dogs in Taipei, age 2-5
print("🔍 Example 4: Complex filter")
result = dog.filter_dogs(df_clean, city="台北", min_age=2, max_age=5, adopted=False)
result_sorted = dog.sort_dogs(result, by="age")
print(result_sorted[['name', 'breed', 'age', 'weight_kg', 'vaccinated']])
```

## 📈 Step 6: Visualizations

使用 `dog.plot_top_breeds()` 生成圖表，也可以選擇是否存檔。

```python
# Generate and display chart (without saving)
print("📊 Generating Top 5 Breeds Chart...")
dog.plot_top_breeds(df_clean, top_n=5)
```

```python
# Generate and save chart
output_path = project_root / "output" / "top_breeds_from_notebook.png"
print(f"💾 Saving chart to: {output_path}")
dog.plot_top_breeds(df_clean, top_n=5, save_path=output_path)
print("✅ Chart saved!")
```

```python
# Create additional visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Age distribution
axes[0, 0].hist(df_clean['age'], bins=10, color='skyblue', edgecolor='black')
axes[0, 0].set_title('Age Distribution', fontweight='bold')
axes[0, 0].set_xlabel('Age (years)')
axes[0, 0].set_ylabel('Frequency')

# 2. Weight distribution
axes[0, 1].hist(df_clean['weight_kg'], bins=15, color='lightcoral', edgecolor='black')
axes[0, 1].set_title('Weight Distribution', fontweight='bold')
axes[0, 1].set_xlabel('Weight (kg)')
axes[0, 1].set_ylabel('Frequency')

# 3. Adoption status
adoption_counts = df_clean['adopted'].value_counts()
axes[1, 0].bar(['Not Adopted', 'Adopted'], adoption_counts.values, color=['salmon', 'lightgreen'])
axes[1, 0].set_title('Adoption Status', fontweight='bold')
axes[1, 0].set_ylabel('Count')

# 4. City distribution
city_counts = df_clean['city'].value_counts()
axes[1, 1].barh(city_counts.index, city_counts.values, color='plum')
axes[1, 1].set_title('Dogs by City', fontweight='bold')
axes[1, 1].set_xlabel('Count')

plt.tight_layout()
plt.show()

print("📊 All visualizations generated!")
```

## 🎯 Step 7: Save Results

將處理後的資料儲存起來。

```python
# Save cleaned data
output_csv = project_root / "data" / "dogs_cleaned_from_notebook.csv"
dog.save_cleaned(df_clean, output_csv)
print(f"✅ Cleaned data saved to: {output_csv}")
```

## 🎓 學習重點總結

### 1. Module 的使用
- ✅ 用 `sys.path.insert()` 將 `src/` 加入搜尋路徑
- ✅ `import dog` 匯入自訂模組
- ✅ 用 `dog.function_name()` 呼叫模組函式

### 2. 資料分析流程
1. **讀取** → `dog.read_dogs_csv()`
2. **清理** → `dog.clean_dogs_df()`
3. **統計** → `dog.stats_summary()`
4. **篩選** → `dog.filter_dogs()`
5. **排序** → `dog.sort_dogs()`
6. **視覺化** → `dog.plot_top_breeds()`
7. **儲存** → `dog.save_cleaned()`

### 3. Notebook vs Module
- **Notebook**：探索、視覺化、實驗
- **Module (dog.py)**：可重用的邏輯、測試、生產

### 4. 跨平台開發
- ✅ 使用 `pathlib.Path` 處理路徑（Windows/macOS 都可用）
- ✅ 避免寫死斜線（`/` 或 `\`）
- ✅ 使用相對路徑
