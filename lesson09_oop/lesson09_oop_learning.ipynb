## 1️⃣ Class 與 Object 是什麼？

### 🏗️ 用「咖啡店」來理解

想像你走進一家咖啡店，點了一杯拿鐵：

- **Class（類別）** = 咖啡的「食譜」或「設計圖」
  - 定義了咖啡有什麼特性（size、price、name）
  - 定義了咖啡可以做什麼（喝一口、顯示資訊）

- **Object（物件）** = 實際做出來的「那一杯咖啡」
  - 你的拿鐵是一個 object
  - 你朋友的美式也是一個 object
  - 它們都是用同一份「食譜（Class）」做出來的

**重點**：
- 一個 Class 可以產生無數個 Object
- 每個 Object 有自己的資料（你的拿鐵是大杯，他的美式是中杯）
- 但它們都遵循同樣的「設計圖」

### 學習建議

- 核心先做：Class / Object、`__init__`、`self`、屬性與方法。
- 可先延後：封裝、繼承、多型，這些概念可以放到熟悉 class 之後再學。
- 這一堂對零基礎來說偏抽象，建議多用生活化比喻，不要一開始就要求完整理解所有名詞。

````python
# 定義一個 Coffee 類別（設計圖）
class Coffee:
    def __init__(self, name, size, price):
        self.name = name    # 咖啡名稱
        self.size = size    # 大小
        self.price = price  # 價格
    
    def display_info(self):
        print(f"☕ {self.name} ({self.size}) - ${self.price}")

# 用這個設計圖做出兩杯實際的咖啡（兩個 Object）
coffee1 = Coffee("拿鐵", "大杯", 120)
coffee2 = Coffee("美式", "中杯", 80)

# 顯示資訊
coffee1.display_info()
coffee2.display_info()

print(f"\n✨ coffee1 和 coffee2 都是用 Coffee 這個類別建立出來的！")
````

## 2️⃣ `__init__` 與 `self` 是什麼？

### 🍼 用「出生登記」來理解

當一個嬰兒出生時，會登記：姓名、性別、生日等「一出生就有的資料」。

- **`__init__`** = 物件的「出生登記程序」
  - 每次建立新 object 時會自動執行
  - 負責設定物件的初始資料

- **`self`** = 「這個物件自己」
  - `self.name` 表示「這個物件的 name」
  - 讓每個 object 都有自己獨立的資料

**白話文**：
- `__init__` 就像嬰兒的出生證明填寫流程
- `self` 就是在說「我自己的資料」

````python
class Person:
    def __init__(self, name, age):
        # __init__ 會在建立物件時自動執行
        print(f"🎉 正在建立 {name} 的資料...")
        self.name = name  # self.name 表示「這個人的名字」
        self.age = age    # self.age 表示「這個人的年齡」
    
    def introduce(self):
        # 在方法裡也用 self 來存取自己的資料
        print(f"嗨！我是 {self.name}，今年 {self.age} 歲。")

# 建立兩個人
person1 = Person("小明", 25)  # 這裡會自動呼叫 __init__
person2 = Person("小華", 30)  # 又會再呼叫一次 __init__

print("\n--- 自我介紹 ---")
person1.introduce()
person2.introduce()

print("\n✨ 注意：每個人都有自己獨立的 name 和 age！")
````

## 3️⃣ 屬性（Attribute）vs 方法（Method）

### 📱 用「手機」來理解

一支手機有：

- **屬性（Attribute）** = 手機的「狀態」或「特徵」
  - 例如：電量、音量、品牌、顏色
  - 這些是「名詞」，描述手機「是什麼樣子」

- **方法（Method）** = 手機的「功能」或「行為」
  - 例如：調整音量、打電話、拍照
  - 這些是「動詞」，描述手機「能做什麼」

**記憶法**：
- 屬性 = 名詞（音量**是**多少？）
- 方法 = 動詞（**調整**音量、**打**電話）

````python
class Phone:
    def __init__(self, brand, color):
        # 屬性（Attributes）— 手機的特徵
        self.brand = brand
        self.color = color
        self.volume = 50     # 預設音量 50
        self.battery = 100   # 預設電量 100%
    
    # 方法（Methods）— 手機的功能
    def increase_volume(self):
        if self.volume < 100:
            self.volume += 10
            print(f"🔊 音量增加到 {self.volume}")
        else:
            print("⚠️ 音量已達最大值！")
    
    def decrease_volume(self):
        if self.volume > 0:
            self.volume -= 10
            print(f"🔉 音量降低到 {self.volume}")
        else:
            print("⚠️ 已靜音！")
    
    def show_status(self):
        print(f"📱 {self.brand} ({self.color})")
        print(f"   🔋 電量：{self.battery}%")
        print(f"   🔊 音量：{self.volume}")

# 建立一支手機
my_phone = Phone("iPhone", "黑色")

# 使用屬性和方法
my_phone.show_status()
print("\n--- 測試音量調整 ---")
my_phone.increase_volume()
my_phone.increase_volume()
my_phone.decrease_volume()
````

## 4️⃣ 封裝（Encapsulation）

### 🏦 用「銀行帳戶」來理解

想像你去銀行：

- ❌ 你**不能**直接修改資料庫裡的餘額數字
- ✅ 你**只能**透過「存款」或「提款」來改變餘額

這就是**封裝**的概念：

1. **隱藏內部資料**（用 `_` 開頭表示「請不要直接改我」）
2. **提供安全的方法**來存取資料（deposit、withdraw）
3. **加入驗證邏輯**（不能提款超過餘額）

**為什麼要封裝？**
- 防止資料被亂改（不能直接改餘額為 100 萬）
- 加入商業邏輯（提款要檢查餘額夠不夠）
- 保護資料安全

````python
class BankAccount:
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        # _balance 用底線開頭，表示「這是內部資料，請不要直接改」
        self._balance = initial_balance
    
    # 存款方法（唯一安全改變餘額的方式）
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"💰 存款成功！存入 ${amount}，目前餘額：${self._balance}")
        else:
            print("⚠️ 存款金額必須大於 0")
    
    # 提款方法（有驗證邏輯）
    def withdraw(self, amount):
        if amount <= 0:
            print("⚠️ 提款金額必須大於 0")
        elif amount > self._balance:
            print(f"❌ 餘額不足！目前餘額：${self._balance}")
        else:
            self._balance -= amount
            print(f"💸 提款成功！提出 ${amount}，剩餘餘額：${self._balance}")
    
    # 唯讀屬性（只能看，不能改）
    @property
    def balance(self):
        return self._balance

# 建立帳戶
account = BankAccount("小明", 1000)

print(f"帳戶擁有者：{account.owner}")
print(f"目前餘額：${account.balance}")  # 用 @property 可以像屬性一樣存取

print("\n--- 測試交易 ---")
account.deposit(500)
account.withdraw(300)
account.withdraw(2000)  # 會失敗，餘額不足

print(f"\n💡 注意：不能直接寫 account.balance = 9999")
print(f"   必須透過 deposit/withdraw 來改變餘額！")
````

## 5️⃣ 繼承（Inheritance）

### 🐾 用「動物」來理解

想像動物世界：

- **所有動物**都有：名字、年齡、會發出聲音
- **狗**是一種動物，但會「汪汪叫」
- **貓**也是一種動物，但會「喵喵叫」

**繼承的好處**：

1. **避免重複寫 code**（name、age 不用寫兩次）
2. **建立階層關係**（Dog 是一種 Animal）
3. **共用基本功能**，但可以**客製化特殊行為**

**術語**：
- **父類別（Parent Class）** = Animal（基礎類別）
- **子類別（Child Class）** = Dog、Cat（繼承自 Animal）

````python
# 父類別：動物（定義所有動物共同的特性）
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def eat(self):
        print(f"🍖 {self.name} 正在吃飯...")
    
    def sleep(self):
        print(f"😴 {self.name} 正在睡覺...")

# 子類別：狗（繼承 Animal 的所有功能）
class Dog(Animal):  # Dog 繼承 Animal
    def speak(self):
        print(f"🐕 {self.name} 說：汪汪汪！")
    
    def fetch(self):
        print(f"🎾 {self.name} 去撿球了！")

# 子類別：貓（繼承 Animal 的所有功能）
class Cat(Animal):  # Cat 繼承 Animal
    def speak(self):
        print(f"🐱 {self.name} 說：喵喵喵～")
    
    def scratch(self):
        print(f"🪑 {self.name} 正在抓沙發...")

# 建立物件
dog = Dog("阿福", 3)
cat = Cat("小花", 2)

print("--- 狗的行為 ---")
dog.eat()     # 繼承自 Animal
dog.sleep()   # 繼承自 Animal
dog.speak()   # Dog 自己的方法
dog.fetch()   # Dog 自己的方法

print("\n--- 貓的行為 ---")
cat.eat()     # 繼承自 Animal
cat.sleep()   # 繼承自 Animal
cat.speak()   # Cat 自己的方法
cat.scratch() # Cat 自己的方法

print("\n✨ Dog 和 Cat 都繼承了 Animal 的 eat() 和 sleep()！")
````

## 6️⃣ 多型（Polymorphism）

### 🎭 用「說話方式」來理解

**多型**的意思是：「同樣的動作，不同的結果」。

例如：
- 你對**狗**說「叫一聲」→ 汪汪汪
- 你對**貓**說「叫一聲」→ 喵喵喵
- 你對**鴨**說「叫一聲」→ 呱呱呱

**同樣都是「叫一聲」**，但**每種動物的實作不同**。

**多型的好處**：

1. **統一介面**（都用 `speak()` 這個名字）
2. **靈活擴充**（要加新動物很簡單）
3. **程式碼更乾淨**（不用寫一堆 if-else）

````python
# 繼續用剛剛的 Animal, Dog, Cat

# 再加一個 Duck 類別
class Duck(Animal):
    def speak(self):
        print(f"🦆 {self.name} 說：呱呱呱～")

# 建立一群動物
animals = [
    Dog("阿福", 3),
    Cat("小花", 2),
    Duck("唐老鴨", 5)
]

print("=== 讓所有動物講話 ===")
for animal in animals:
    # 同樣都呼叫 speak()，但每種動物的實作不同
    animal.speak()

print("\n✨ 這就是多型！")
print("   同樣的方法名稱（speak），不同的實作結果。")
print("   我們不需要知道是什麼動物，只要呼叫 speak() 就好！")
````

## 7️⃣ dataclass 簡介

### 📦 什麼時候用 dataclass？

如果你的 class **主要是用來儲存資料**，而不是做複雜的邏輯，可以用 `dataclass` 簡化寫法。

**傳統寫法 vs dataclass**：

傳統：
```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

dataclass：
```python
@dataclass
class Point:
    x: int
    y: int
```

**dataclass 自動幫你做的事**：
- ✅ 自動產生 `__init__`
- ✅ 自動產生 `__repr__`（好看的顯示格式）
- ✅ 自動產生 `__eq__`（可以比較兩個物件）

**何時用？**
- ✅ 主要用來儲存資料（像資料庫的一筆記錄）
- ✅ 不需要複雜的初始化邏輯
- ❌ 需要封裝、驗證邏輯 → 用傳統 class

````python
from dataclasses import dataclass

# 用 dataclass 定義一個座標點
@dataclass
class Point:
    x: int
    y: int

# 用 dataclass 定義一個產品
@dataclass
class Product:
    name: str
    price: float
    stock: int = 0  # 可以設定預設值

# 建立物件（超簡潔！）
p1 = Point(10, 20)
p2 = Point(10, 20)

product = Product("iPhone", 30000, 50)

# 自動產生的好看顯示
print(f"點 1：{p1}")
print(f"點 2：{p2}")
print(f"產品：{product}")

# 自動可以比較
print(f"\np1 == p2? {p1 == p2}")

print("\n✨ 注意看，我們沒寫 __init__，但它自動幫我們做好了！")
````

## 8️⃣ 迷你專案：Todo 待辦事項管理系統

### 🎯 專案目標

建立一個簡單的 Todo 系統，包含：

1. **TodoItem**：單一待辦事項（有標題、完成狀態）
2. **TodoList**：管理多個待辦事項（新增、完成、顯示）

**用到的 OOP 概念**：
- ✅ Class 與 Object
- ✅ 屬性與方法
- ✅ 封裝（用方法來改變狀態）
- ✅ dataclass（簡化 TodoItem）

````python
from dataclasses import dataclass
from datetime import datetime

# 單一待辦事項
@dataclass
class TodoItem:
    title: str
    done: bool = False
    created_at: str = None
    
    def __post_init__(self):
        # dataclass 的特殊方法，在 __init__ 後執行
        if self.created_at is None:
            self.created_at = datetime.now().strftime("%Y-%m-%d %H:%M")
    
    def mark_done(self):
        self.done = True
    
    def mark_undone(self):
        self.done = False
    
    def __str__(self):
        status = "✅" if self.done else "⬜"
        return f"{status} {self.title} (建立於 {self.created_at})"

# 測試 TodoItem
item = TodoItem("學習 Python OOP")
print(item)
item.mark_done()
print(item)
````

````python
# Todo 清單管理器
class TodoList:
    def __init__(self, owner):
        self.owner = owner
        self._items = []  # 用 list 儲存所有 TodoItem
    
    def add(self, title):
        """新增待辦事項"""
        item = TodoItem(title)
        self._items.append(item)
        print(f"➕ 已新增：{title}")
    
    def complete(self, index):
        """完成指定編號的事項"""
        if 0 <= index < len(self._items):
            self._items[index].mark_done()
            print(f"✅ 已完成：{self._items[index].title}")
        else:
            print("⚠️ 編號錯誤！")
    
    def show_all(self):
        """顯示所有事項"""
        print(f"\n📋 {self.owner} 的待辦清單")
        print("=" * 50)
        if not self._items:
            print("（空空如也～）")
        else:
            for i, item in enumerate(self._items):
                print(f"{i}. {item}")
        print("=" * 50)
    
    def show_pending(self):
        """只顯示未完成的事項"""
        print(f"\n⏳ {self.owner} 的待辦事項（未完成）")
        print("=" * 50)
        pending = [item for item in self._items if not item.done]
        if not pending:
            print("🎉 全部完成了！")
        else:
            for item in pending:
                print(f"  {item}")
        print("=" * 50)
    
    @property
    def total(self):
        """總事項數"""
        return len(self._items)
    
    @property
    def completed_count(self):
        """已完成數量"""
        return sum(1 for item in self._items if item.done)
    
    @property
    def pending_count(self):
        """待完成數量"""
        return self.total - self.completed_count
````

````python
# 實際使用 Todo 系統
my_todos = TodoList("小明")

# 新增待辦事項
my_todos.add("學習 Python OOP")
my_todos.add("寫作業")
my_todos.add("運動 30 分鐘")
my_todos.add("閱讀一章書")

# 顯示所有事項
my_todos.show_all()

# 完成幾個事項
print("\n--- 開始完成事項 ---")
my_todos.complete(0)  # 完成「學習 Python OOP」
my_todos.complete(2)  # 完成「運動 30 分鐘」

# 再次顯示所有事項
my_todos.show_all()

# 只顯示待辦事項
my_todos.show_pending()

# 顯示統計資訊
print(f"\n📊 統計資訊")
print(f"總共：{my_todos.total} 項")
print(f"已完成：{my_todos.completed_count} 項")
print(f"待完成：{my_todos.pending_count} 項")
````

## 9️⃣ 本課重點總整理

### 🎯 核心概念速查表

| 概念 | 白話解釋 | 記憶法 |
|------|----------|--------|
| **Class** | 設計圖、食譜 | 一份藍圖可以蓋很多房子 |
| **Object** | 實際做出來的東西 | 每棟房子都是獨立的 |
| **`__init__`** | 出生登記程序 | 寶寶一出生就有名字、性別 |
| **`self`** | 「我自己」的資料 | self.name = 我的名字 |
| **屬性（Attribute）** | 名詞、狀態 | 手機的顏色、音量 |
| **方法（Method）** | 動詞、行為 | 打電話、拍照 |
| **封裝（Encapsulation）** | 保護資料，提供安全接口 | 銀行不讓你直接改餘額 |
| **繼承（Inheritance）** | 子類別繼承父類別的功能 | 狗是一種動物 |
| **多型（Polymorphism）** | 同名方法，不同實作 | 都叫「speak」，但汪/喵不同 |
| **dataclass** | 簡化資料類別的寫法 | 主要用來儲存資料時使用 |

### 📝 常見錯誤與解決

#### 1. 忘記寫 `self`
```python
❌ def show_info():          # 錯誤！
       print(name)

✅ def show_info(self):      # 正確！
       print(self.name)
```

#### 2. 屬性沒加 `self`
```python
❌ def __init__(self, name):
       name = name           # 錯誤！這只是區域變數

✅ def __init__(self, name):
       self.name = name      # 正確！
```

#### 3. 呼叫方法忘記加 `()`
```python
❌ dog.speak              # 錯誤！這只是取得方法本身
✅ dog.speak()            # 正確！執行方法
```

#### 4. 直接修改應該被保護的屬性
```python
❌ account._balance = 999999  # 不好的做法
✅ account.deposit(500)       # 正確！透過方法
```

````python
# 最後一個練習：建立你自己的第一個 Class！
# 試著建立一個「學生」類別，包含：
# - 屬性：姓名、年級、成績列表
# - 方法：新增成績、計算平均分數、顯示學生資訊

class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
        self._scores = []
    
    def add_score(self, subject, score):
        self._scores.append({"科目": subject, "分數": score})
        print(f"✅ 已新增 {subject} 成績：{score} 分")
    
    def get_average(self):
        if not self._scores:
            return 0
        total = sum(item["分數"] for item in self._scores)
        return round(total / len(self._scores), 2)
    
    def show_info(self):
        print(f"\n👨‍🎓 學生資訊")
        print(f"姓名：{self.name}")
        print(f"年級：{self.grade}")
        print(f"成績記錄：")
        for item in self._scores:
            print(f"  - {item['科目']}：{item['分數']} 分")
        print(f"平均分數：{self.get_average()} 分")

# 測試你的 Student 類別
student = Student("小明", "高一")
student.add_score("數學", 85)
student.add_score("英文", 90)
student.add_score("物理", 78)
student.show_info()

print("\n🎉 恭喜！你已經完成 Python OOP 入門課程！")
print("💪 繼續練習，你會越來越厲害的！")
````
